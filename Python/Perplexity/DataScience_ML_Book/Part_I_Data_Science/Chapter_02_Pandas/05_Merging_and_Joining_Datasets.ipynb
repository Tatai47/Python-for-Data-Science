```python
# Real-world application example
# Load and analyze actual dataset

# Sample data creation
np.random.seed(42)
sample_data = np.random.randn(100, 5)
df = pd.DataFrame(sample_data, columns=[f'Feature_{i}' for i in range(1, 6)])

# Display basic statistics
print('Dataset Shape:', df.shape)
print('\nBasic Statistics:')
df.describe()
```

## 6. Visualization

Create meaningful visualizations to understand the data and results better.


```python
# Create visualizations
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# Plot 1: Distribution
axes[0, 0].hist(df['Feature_1'], bins=30, edgecolor='black', alpha=0.7)
axes[0, 0].set_title('Feature 1 Distribution')
axes[0, 0].set_xlabel('Value')
axes[0, 0].set_ylabel('Frequency')

# Plot 2: Correlation Heatmap
corr_matrix = df.corr()
im = axes[0, 1].imshow(corr_matrix, cmap='coolwarm', aspect='auto')
axes[0, 1].set_title('Correlation Matrix')
plt.colorbar(im, ax=axes[0, 1])

# Plot 3: Box Plot
df.boxplot(ax=axes[1, 0])
axes[1, 0].set_title('Feature Distributions')
axes[1, 0].tick_params(axis='x', rotation=45)

# Plot 4: Scatter Plot
axes[1, 1].scatter(df['Feature_1'], df['Feature_2'], alpha=0.6)
axes[1, 1].set_title('Feature 1 vs Feature 2')
axes[1, 1].set_xlabel('Feature 1')
axes[1, 1].set_ylabel('Feature 2')

plt.tight_layout()
plt.show()
```
