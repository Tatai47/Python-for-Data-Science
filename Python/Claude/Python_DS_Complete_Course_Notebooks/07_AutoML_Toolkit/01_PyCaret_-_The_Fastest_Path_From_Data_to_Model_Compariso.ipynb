# Chapter 1: PyCaret — The Fastest Path From Data to Model Comparison

### Why This Matters

PyCaret is the most beginner-friendly entry point into AutoML precisely because its API mirrors how you'd *describe* the ML workflow in a meeting: setup, compare, tune, finalize.

```python
from pycaret.classification import *

# 1. SETUP: point PyCaret at your data and target; it infers types, handles missing values,
#    and internally creates its own train/test split
s = setup(data=train_df, target="churn", session_id=42,
          categorical_features=["contract", "internet", "tech_support"])

# 2. COMPARE: train and cross-validate every reasonable model PyCaret knows about, ranked by your chosen metric
best = compare_models(include=["lr", "rf", "xgboost", "lightgbm", "catboost"], sort="AUC")

# 3. TUNE: run automated hyperparameter search specifically on the winning model
tuned = tune_model(best)

# 4. FINALIZE: refit the tuned model on ALL available training data (not just the internal CV folds)
final = finalize_model(tuned)

# 5. PREDICT / EVALUATE
predictions = predict_model(final, data=test_df)
```

Walking through why this five-step shape matters: `setup()` is where you'd normally spend the most manual effort (Volumes 02, 04, 05) — missing value handling, encoding, scaling — and PyCaret handles sensible defaults for all of it in one call, while still letting you override specifics (like explicitly naming categorical columns, as shown). `compare_models()` is the headline feature: rather than you manually writing and running the Logistic Regression → Decision Tree → XGBoost → LightGBM comparison loop from Volume 06 by hand, one line trains and cross-validates every model in the list and returns a ranked leaderboard. `tune_model()` runs the Optuna-style search from Volume 06 automatically on whichever model won. `finalize_model()` matters for a subtle but important reason: everything up to this point was evaluated using internal cross-validation *splits* of your training data — `finalize_model` retrains the winning configuration on the *entire* training set before you deploy it, since more training data (once you've already validated the approach) generally means a stronger final model.

💡 PyCaret's `compare_models()` leaderboard is genuinely one of the fastest ways to get an honest first read on "is this problem even predictable with reasonable accuracy from the features I have" — a question worth answering in the first hour of a project, not the third week.

### Cheat Sheet — PyCaret

| Step | Function |
|---|---|
| Initialize experiment | `setup(data, target, categorical_features=[...])` |
| Compare models | `compare_models(include=[...], sort="AUC")` |
| Tune the winner | `tune_model(best)` |
| Retrain on full data | `finalize_model(tuned)` |
| Predict on new data | `predict_model(final, data=new_df)` |
