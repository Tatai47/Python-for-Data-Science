## Why This Volume Matters

Every volume before this one worked with numbers in neat rows and columns. But an enormous amount of real business information — customer reviews, support tickets, contracts, internal reports, Slack messages — exists as unstructured **text**. This volume takes you from classic text analytics (extracting structured information from text) through to the current generative AI stack: using large language models grounded in your own company's documents, safely and specifically, rather than relying on their general training knowledge alone.

**The five libraries and where each fits in the pipeline:**

| Library | Role |
|---|---|
| **spaCy** | Fast, classic NLP — entities, parts of speech, pattern matching. No generative AI involved. |
| **Hugging Face Transformers** | Access to pretrained transformer models for classification, sentiment, embeddings, and generation |
| **LangChain** | Orchestrates LLM calls into applications — prompts, chains, tools, agents |
| **LlamaIndex** | Specializes in connecting your own documents to an LLM (the "R" in RAG) |
| **ChromaDB** | A vector database — the storage and search engine that makes semantic retrieval possible |
