# Chapter 7: Conditionals and Match

### Why This Matters

Every business rule ("if the customer's spend exceeds X, flag as VIP," "if sales missed target, escalate") is, at the code level, a conditional. This chapter is where your programs start actually *deciding* things rather than just calculating them.

### 7.1 if / elif / else

```python
sales, target = 1350000, 1200000

if sales > target * 1.1:
    status = "Excellent – significantly above target"
elif sales >= target:
    status = "Met target"
else:
    status = "Below target – action needed"
print(status)
```

**Read the structure carefully:** Python checks conditions top to bottom and executes the *first* branch that matches, then skips the rest entirely — even if a later condition would also be true. Order your conditions from most-specific to least-specific (or, as here, from "best case" downward) to avoid a subtle class of logic bugs where a later, broader condition never gets reached, or an earlier, broader one swallows a case it shouldn't.

Indentation is not a style preference in Python — it **is** the syntax that defines which lines belong to which block. This is different from languages that use `{}` or `end` keywords. Four spaces per level is the near-universal convention (PEP 8); mixing tab and space indentation in the same file will crash your script.

### 7.2 Ternary Expressions and Nesting

```python
sales = 950000
status = "Above target" if sales >= 1000000 else "Below target"     # a full if/else compressed to one line

score = 78
if score >= 90:
    grade = "A"
elif score >= 75:
    grade = "B"
elif score >= 60:
    grade = "C"
else:
    grade = "D"
print(grade)
```

The **ternary expression** (`value_if_true if condition else value_if_false`) is best used for short, simple assignments. If the logic needs more than one condition or spans multiple lines, use a full `if/elif/else` block instead — cramming complex logic into a ternary hurts readability, which defeats the purpose of writing clear code in the first place.

### 7.3 match-case (Python 3.10+)

```python
region = "West"

match region:
    case "North":
        manager = "Aisha"
    case "South":
        manager = "Rahul"
    case "East" | "West":          # the | here means "either of these values"
        manager = "Shared Team"
    case _:                        # the wildcard — matches anything not caught above
        manager = "Unassigned"
print(manager)
```

`match`/`case` is Python's newer answer to what other languages call a `switch` statement. It shines when you're comparing one variable against many discrete possible values — it reads cleaner than an equivalent chain of `elif region == "North": ... elif region == "South": ...`. 💡 Check your Python version before relying on it in shared code (`python --version`) — it only exists from 3.10 onward, and some corporate/legacy environments still run 3.8 or 3.9.

### ⚠️ Common Mistakes

- Using `=` instead of `==` inside an `if` (Python will raise a `SyntaxError` here, so at least you'll be told immediately).
- Forgetting the colon `:` at the end of `if`, `elif`, `else`, or `match`/`case` lines.
- Inconsistent indentation within the same block (Python is strict about this — pick one indent width, typically 4 spaces, and never deviate within a block).

### Cheat Sheet — Chapter 7

```python
if condition1:
    ...
elif condition2:
    ...
else:
    ...

value = a if condition else b        # ternary

match variable:
    case pattern1:
        ...
    case pattern2 | pattern3:
        ...
    case _:
        ...
```

### 🎯 Business Challenge — Chapter 7

*Task:* Commission calculator — <50,000 → 2%, 50,000–100,000 → 5%, >100,000 → 8%.

```python
def commission_rate(sales):
    if sales < 50000:
        return 0.02
    elif sales <= 100000:
        return 0.05
    else:
        return 0.08

sales_figure = 87000
rate = commission_rate(sales_figure)
amount = sales_figure * rate

print(f"Sales: ₹{sales_figure:,} | Rate: {rate:.0%} | Commission: ₹{amount:,.2f}")
```

(Note: this challenge is written here using a function, previewing Chapter 9 — you could equally write it as a standalone `if/elif/else` block without a function; both are correct, but wrapping business logic in a named function is the habit worth building early, since it makes the rule reusable and testable.)
