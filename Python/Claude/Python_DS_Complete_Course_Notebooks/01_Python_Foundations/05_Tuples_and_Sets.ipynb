# Chapter 5: Tuples and Sets

### Why This Matters

Lists aren't always the right tool. Sometimes you want to *guarantee* data can't be accidentally modified (a fixed record like a database row) — that's a **tuple**. Sometimes you only care about unique values and fast membership checks — that's a **set**. Knowing when to reach for each makes your code both safer and, for sets specifically, dramatically faster for certain operations.

### 5.1 Tuples — Immutable Sequences

```python
point = (10, 20)
employee = ("Priya", "Analytics", 1250000)
single = (42,)          # the trailing comma is REQUIRED — (42) alone is just the integer 42 in parentheses

print(employee[0])
print(employee[1:])

name, dept, salary = employee     # "unpacking" — assign each element to a variable in one line
print(name, dept, salary)
```

⚠️ `single = (42)` (no comma) does **not** create a tuple — it's just `42` wrapped in redundant parentheses. Python needs that comma to know you mean a one-element tuple. This is a genuinely common gotcha.

Tuples support far fewer methods than lists precisely *because* they're immutable — there's no `.append()`, `.remove()`, or `.sort()` on a tuple, since none of those make sense for something that can't change:

```python
t = (1, 2, 2, 3, 2, 4)
print(t.count(2))    # 3
print(t.index(3))    # 3 (position of first 3)
```

**Why use a tuple instead of a list at all?** Three real reasons: (1) it signals intent — "this data is a fixed record, not a growing collection," (2) it's slightly faster and uses less memory than an equivalent list, (3) tuples can be used as dictionary keys and set elements (lists cannot, because they're mutable). You'll see this third reason matter again when you group data by multiple columns in Pandas — the group key is internally a tuple.

### 5.2 Sets — Unique Elements, Fast Membership

```python
regions = {"North", "South", "East", "West", "North"}
print(regions)          # {'North', 'South', 'East', 'West'} — duplicate silently dropped, order not guaranteed

regions.add("Central")
regions.discard("East")     # like .remove() but does NOT error if the item is missing
print("North" in regions)   # membership check — this is O(1) for a set, vs O(n) for a list
```

💡 **Why the `in` check matters at scale:** checking `"North" in regions` for a set is (on average) instant regardless of how many items the set holds, because sets use a hash table internally — the same underlying structure as dictionaries. Checking `"North" in regions_list` for an equivalent list has to potentially scan every element. For a handful of regions this difference is invisible; for "is this customer ID in my 2-million-row blocklist," it's the difference between milliseconds and seconds.

**Set algebra** — genuinely useful for business questions about overlapping groups:

```python
customers_q1 = {"C001", "C002", "C003", "C004"}
customers_q2 = {"C003", "C004", "C005", "C006"}

print("New customers:", customers_q2 - customers_q1)          # in Q2 but not Q1
print("Retained customers:", customers_q1 & customers_q2)     # in BOTH
print("All customers:", customers_q1 | customers_q2)          # in EITHER
print("Churned or new only:", customers_q1 ^ customers_q2)    # in exactly ONE, not both
```

This is exactly the kind of question an analyst gets asked constantly ("how many customers did we retain quarter over quarter, how many are new, how many churned") and set operators answer it in one line each, with no loop required.

### Cheat Sheet — Chapter 5

| Operator | Meaning | Example |
|---|---|---|
| `-` | Difference | `A - B` = in A, not in B |
| `&` | Intersection | `A & B` = in both |
| `\|` | Union | `A \| B` = in either |
| `^` | Symmetric difference | `A ^ B` = in exactly one |
| `in` | Membership | `"x" in A` |

### 🎯 Business Challenge — Chapter 5

*Task:* Given two campaign-response sets, find who responded to both, who responded only to the first, and total unique reach.

```python
campaign_A = {"C101", "C102", "C103", "C104", "C105"}
campaign_B = {"C103", "C104", "C106", "C107"}

both = campaign_A & campaign_B
only_A = campaign_A - campaign_B
total_reach = campaign_A | campaign_B

print(f"Responded to both      : {both}")
print(f"Only responded to A    : {only_A}")
print(f"Total unique customers : {len(total_reach)}")
```
