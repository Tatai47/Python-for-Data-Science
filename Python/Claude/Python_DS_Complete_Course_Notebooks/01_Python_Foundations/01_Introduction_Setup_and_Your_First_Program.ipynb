### 1.3 Anatomy of Your First Program

```python
print("Hello, Data Analyst!")
print("Welcome to your complete Python journey.")
print("Today you start turning data into decisions.")
```

`print()` is a **function** — a named, reusable block of behavior that Python ships with. You're *calling* it, and the text in quotes is the **argument** you're passing it. Every single line of Python you'll ever write is some combination of: define data, call a function on that data, and (optionally) store the result. That's the entire mental model — everything else is vocabulary.

### 1.4 Your First Business Calculation

```python
revenue = 2_450_000
cost = 1_875_000
profit = revenue - cost
margin = (profit / revenue) * 100

print(f"Revenue      : ₹{revenue:,}")
print(f"Cost         : ₹{cost:,}")
print(f"Profit       : ₹{profit:,}")
print(f"Profit Margin: {margin:.2f}%")
```

Two things worth noticing immediately, because they'll recur constantly:

- **The underscore in `2_450_000`** is purely cosmetic — Python ignores underscores in numeric literals. It exists so *you*, the human reading the code six months from now, can tell at a glance this is 2.45 million and not 245,000 or 24.5 million. Use it liberally in any large literal number.
- **The `f` before the string** (an **f-string**) lets you embed live Python expressions inside `{curly braces}` directly in text. The `:,` inside the braces is a **format spec** — it inserts thousand-separator commas automatically. You'll use f-strings on nearly every line of output you ever write, so it's worth internalizing this syntax now rather than treating it as magic.

### ⚠️ Common Mistakes at This Stage

- Forgetting the `f` before a string that contains `{}` — Python will print the literal text `{revenue}` instead of its value.
- Mixing tabs and spaces for indentation (more relevant from Chapter 7 onward, but worth flagging now): configure your editor to insert spaces, never tabs, when you press Tab. This single setting prevents a whole category of frustrating `IndentationError`s later.

### 🎯 Business Challenge — Chapter 1

*Scenario:* Last quarter your company generated ₹37,50,000 in sales. Total expenses were ₹28,90,000. Calculate absolute profit, profit margin (%), and print a clean executive-style message.

**Worked solution:**

```python
sales = 37_50_000       # Indian numbering — you can group digits however is readable to you
expenses = 28_90_000

profit = sales - expenses
margin = (profit / sales) * 100

print("=" * 40)
print("QUARTERLY EXECUTIVE SUMMARY")
print("=" * 40)
print(f"Sales     : ₹{sales:,}")
print(f"Expenses  : ₹{expenses:,}")
print(f"Profit    : ₹{profit:,}")
print(f"Margin    : {margin:.2f}%")
```

Notice `"=" * 40` — multiplying a string by an integer repeats it. This is a small, cheap trick professional scripts use constantly to print visual section dividers in console reports.
