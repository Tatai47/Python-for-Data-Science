# Chapter 2: Variables, Operators & Input/Output

### Why This Matters

Every analysis you'll ever do is, underneath, "store some numbers under meaningful names, then combine them with arithmetic and comparisons." Getting fluent with variables and operators is what turns Python from "a calculator with extra steps" into a tool you *think* in.

### 2.1 Variables — Names for Values

```python
company = "UrbanCart"
revenue = 12500000
growth_rate = 0.18
is_profitable = True

print(type(company))       # <class 'str'>
print(type(revenue))       # <class 'int'>
print(type(growth_rate))   # <class 'float'>
print(type(is_profitable)) # <class 'bool'>
```

Python is **dynamically typed**: you never declare `int x = 5` like in Java or C#. The variable is just a label pointing at a value, and that value carries its own type. This is convenient for fast analysis work, but it's also *the* source of most bugs experienced analysts hit when a column they expect to be numeric turns out to contain the string `"N/A"` — you'll see exactly this problem again in Volume 02 with Pandas' `dtype` object columns.

**Naming rules** (enforced by Python) vs. **naming conventions** (enforced by your own discipline):

- *Rules:* names can't start with a digit, can't contain spaces or most symbols, and can't be one of Python's ~35 reserved keywords (`if`, `for`, `class`, `return`, etc.).
- *Conventions:* use `snake_case` (`monthly_revenue`, not `MonthlyRevenue` or `monthlyRevenue` — those styles are reserved by convention for classes and, in some other ecosystems, for different roles). Prefer descriptive names — `avg_order_value` beats `aov` beats `x` every time you revisit the code a week later.

```python
# Multiple assignment — assign several variables in one line
x, y, z = 10, 20, 30

# The classic "swap" idiom — Python evaluates the right side fully before assigning
a, b = 5, 9
a, b = b, a
print(a, b)   # 9 5
```

💡 That swap idiom (`a, b = b, a`) is genuinely idiomatic Python — in most other languages swapping two variables needs a temporary third variable. You'll see this pattern reused inside sorting algorithms and data-shuffling code later.

### 2.2 Operators

**Arithmetic operators** — note two that trip up people coming from Excel or other languages:

```python
print(15 / 4)    # 3.75  -> "true division", ALWAYS returns a float
print(15 // 4)   # 3     -> "floor division", drops the remainder
print(15 % 4)    # 3     -> "modulus", the remainder itself
print(15 ** 4)   # 50625 -> exponentiation (NOT ^, which means something else in Python)

# Business example: compound growth — the same formula behind any CAGR calculation
principal = 100000
rate = 0.12
years = 5
future_value = principal * (1 + rate) ** years
print(f"Future value after {years} years: ₹{future_value:,.0f}")
```

⚠️ **The single most common gotcha for people who learned another language first:** `^` is *not* exponentiation in Python (it's the bitwise XOR operator, rarely used in analytics work). Always use `**` for powers.

**Comparison and logical operators** — these return `True`/`False` and are the backbone of every filtering operation you'll do in Pandas later:

```python
sales, target = 1250000, 1200000
print(sales > target)    # True
print(sales == target)   # False  (== compares equality; = assigns — mixing these up is a classic bug)

high_sales = sales > 1000000
met_target = sales >= target
print(high_sales and met_target)   # True only if BOTH are True
print(high_sales or met_target)    # True if AT LEAST ONE is True
print(not met_target)              # flips True <-> False
```

⚠️ `=` assigns a value; `==` checks equality. Typing `if sales = target:` is a syntax error Python will catch, but the mix-up costs beginners real debugging time until the distinction becomes automatic.

**Assignment (compound) operators** — shorthand you'll use constantly to update a running total:

```python
total = 100
total += 50   # same as: total = total + 50   -> 150
total *= 2    # same as: total = total * 2     -> 300
```

### 2.3 Input & Output

`print()` sends text out; `input()` reads text in — and **always** returns a string, even if the user types a number, so you must convert it explicitly.

```python
print("Sales Report", "Q3 2025", sep=" | ", end="\n\n")
print(f"Revenue: ₹{1250000:,}")
print("Growth: {:.1%}".format(0.185))     # the older .format() style — you'll see this in legacy code

# Reading user input (uncomment to try interactively):
# name = input("Enter your name: ")
# units = int(input("Units sold: "))      # explicit conversion from str -> int
# price = float(input("Price per unit: "))
# print(f"{name} generated ₹{units * price:,.2f}")
```

💡 Format-spec quick reference you'll reuse for the rest of your career: `{value:,}` adds thousand separators, `{value:.2f}` fixes 2 decimal places, `{value:.1%}` converts a fraction (0.185) into a percentage string (18.5%) automatically — no need to multiply by 100 yourself.

### Cheat Sheet — Chapter 2

| Operation | Symbol | Example |
|---|---|---|
| True division | `/` | `15/4 → 3.75` |
| Floor division | `//` | `15//4 → 3` |
| Modulus | `%` | `15%4 → 3` |
| Power | `**` | `2**10 → 1024` |
| Equality test | `==` | `a == b` |
| Compound assign | `+=  -=  *=  /=` | `total += 50` |
| Percent format | `{x:.1%}` | `0.185 → "18.5%"` |
| Thousands format | `{x:,}` | `1250000 → "1,250,000"` |

### 🎯 Business Challenge — Chapter 2

*Task:* Create variables for product name, units sold, unit price, and a discount rate (e.g., 0.10). Calculate gross sales, discount amount, and net sales; print a clean f-string summary.

```python
product = "Wireless Mouse"
units_sold = 340
unit_price = 799.0
discount_rate = 0.12

gross_sales = units_sold * unit_price
discount_amount = gross_sales * discount_rate
net_sales = gross_sales - discount_amount

print(f"Product      : {product}")
print(f"Units Sold   : {units_sold:,}")
print(f"Gross Sales  : ₹{gross_sales:,.2f}")
print(f"Discount ({discount_rate:.0%}): ₹{discount_amount:,.2f}")
print(f"Net Sales    : ₹{net_sales:,.2f}")
```
