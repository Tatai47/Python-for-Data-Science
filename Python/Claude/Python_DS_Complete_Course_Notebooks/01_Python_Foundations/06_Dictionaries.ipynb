# Chapter 6: Dictionaries

### Why This Matters

A **dictionary** stores **key → value** pairs. If a list is "a column of data," a dictionary is closer to "a single labeled record" — which is exactly why, when Pandas needs to represent one row of a DataFrame as a Python object, or when you read a JSON API response, you get a dictionary. Fluency here transfers directly.

### 6.1 Creating and Accessing

```python
employee = {
    "name": "Priya Sharma",
    "department": "Analytics",
    "salary": 1250000,
    "skills": ["Python", "SQL", "Tableau"]      # values can be any type, including other collections
}

print(employee["name"])                          # KeyError if "name" doesn't exist
print(employee.get("location", "Not specified")) # .get() is SAFE — returns a default instead of erroring
print(employee.keys())
print(employee.values())
print(employee.items())     # key-value pairs, used constantly in for-loops
```

💡 **Always prefer `.get()` over `[]` when a key might not exist.** `employee["location"]` raises a `KeyError` and crashes your script if "location" was never set; `employee.get("location", "Not specified")` gracefully falls back to a default. This one habit prevents a large fraction of runtime crashes in real data-processing scripts, where not every record has every field.

### 6.2 Modifying Dictionaries

```python
employee["location"] = "Mumbai"              # add a new key, or overwrite an existing one
employee["salary"] += 100000
employee.update({"role": "Senior Analyst", "active": True})   # merge multiple keys at once
print(employee)

removed = employee.pop("active")             # removes AND returns the value
print("Removed:", removed)
```

### 6.3 Dictionary Comprehensions and Nesting

```python
sales = {"North": 125, "South": 98, "East": 112, "West": 145}
in_thousands = {k: v * 1000 for k, v in sales.items()}
print(in_thousands)

# Nested dictionaries — this is exactly the shape of a typical JSON API response
regions = {
    "North": {"sales": 1250000, "target": 1200000},
    "South": {"sales": 980000, "target": 1000000}
}
print(regions["North"]["sales"])    # chain keys to drill into nested structure
```

Dictionary comprehensions follow the identical pattern as list comprehensions from Chapter 4, just with a `key: value` pair instead of a single expression. Once you're comfortable with one, the other is nearly free.

### Cheat Sheet — Chapter 6

| Operation | Syntax |
|---|---|
| Safe access | `d.get(key, default)` |
| Add/update | `d[key] = value` or `d.update({...})` |
| Remove | `d.pop(key)` |
| Loop key+value | `for k, v in d.items():` |
| Comprehension | `{k: expr for k, v in d.items()}` |
| Nested access | `d["outer"]["inner"]` |

### 🎯 Business Challenge — Chapter 6

*Task:* Build a product dictionary (name, category, price, stock); add a `discount` key; compute discounted price; print a summary.

```python
product = {"name": "Bluetooth Speaker", "category": "Electronics", "price": 2499.0, "stock": 58}
product["discount"] = 0.15

discounted_price = product["price"] * (1 - product["discount"])

print(f"{product['name']} ({product['category']})")
print(f"Original price : ₹{product['price']:,.2f}")
print(f"Discount       : {product['discount']:.0%}")
print(f"Final price    : ₹{discounted_price:,.2f}")
print(f"In stock       : {product['stock']} units")
```
