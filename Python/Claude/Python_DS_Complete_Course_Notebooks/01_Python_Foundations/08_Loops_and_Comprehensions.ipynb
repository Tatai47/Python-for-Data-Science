# Chapter 8: Loops and Comprehensions

### Why This Matters

Loops are how you process a *collection* of things — every transaction, every customer, every row — without writing out the same code 500 times. Understanding loops deeply, including their less commonly-taught features (`else` on a loop, `enumerate`), pays off constantly in data-cleaning scripts.

### 8.1 for Loops

```python
regions = ["North", "South", "East", "West"]

for region in regions:
    print(f"Processing {region} region")

for i, region in enumerate(regions, start=1):    # enumerate gives you (index, value) pairs
    print(i, region)
```

`enumerate()` solves a problem beginners commonly solve badly — manually tracking a counter variable (`i = 0`, then `i += 1` inside the loop). Whenever you need both the position *and* the value while looping, reach for `enumerate()` instead of managing an index by hand.

```python
sales = [125, 98, 112, 145]
total = 0
for s in sales:
    total += s
print("Total:", total)

for i in range(1, 6):        # range(start, stop) — stop is exclusive, exactly like slicing
    print(i, end=" ")
print()
```

### 8.2 while Loops

```python
count = 0
while count < 5:
    print(count)
    count += 1        # if you forget this line, the loop runs forever — always double-check your increment
```

`while` loops run as long as a condition holds, and are the right choice when you don't know in advance *how many* iterations you'll need — e.g., "keep asking the user for input until they type 'quit'," or "keep retrying an API call until it succeeds or you've tried 5 times." When you know the exact collection or count up front, prefer `for`.

### 8.3 break, continue, and the Loop's else

```python
for s in [80, 95, 110, 70, 130]:
    if s < 100:
        continue          # skip the rest of THIS iteration, move to the next value
    if s > 120:
        print("Exceptional:", s)
        break              # exit the loop entirely, right now
    print("Good:", s)
else:
    print("Loop completed without break")     # only runs if the loop was NOT exited via break
```

The `else` clause on a loop is genuinely obscure and most Python developers forget it exists — but it's precisely the right tool for "search for something, and do X only if it wasn't found," without needing a separate `found = False` flag variable.

### 8.4 Comprehensions Revisited: List, Dict, and Set

```python
sales = [85, 120, 95, 140, 110]
high = [s for s in sales if s >= 100]              # list comprehension (Ch 4 recap)
squared = {s: s**2 for s in sales}                  # dict comprehension (Ch 6 recap)
unique_high = {s for s in sales if s >= 100}        # SET comprehension — new here
print(high, squared, unique_high, sep="\n")
```

### Cheat Sheet — Chapter 8

| Tool | Use when |
|---|---|
| `for x in collection:` | Iterating a known, finite sequence |
| `for i, x in enumerate(collection):` | You need both index and value |
| `while condition:` | Unknown number of iterations |
| `break` | Exit the loop immediately |
| `continue` | Skip to the next iteration |
| `for...else` | Run code only if the loop finished without `break` |
| `[expr for x in seq if cond]` | Build a list |
| `{expr for x in seq if cond}` | Build a set |
| `{k: v for x in seq}` | Build a dict |

### 🎯 Business Challenge — Chapter 8

*Task:* Sales `[72000, 85000, 110000, 95000, 130000, 88000]` — total & count above 100k; growth % vs. a prior baseline of 80000; only months that grew >20%.

```python
sales = [72000, 85000, 110000, 95000, 130000, 88000]
baseline = 80000

total = 0
count_above_100k = 0
for s in sales:
    total += s
    if s > 100000:
        count_above_100k += 1

growth_pcts = [round((s - baseline) / baseline * 100, 1) for s in sales]
strong_growth = [s for s, g in zip(sales, growth_pcts) if g > 20]

print(f"Total sales        : ₹{total:,}")
print(f"Months above 100k  : {count_above_100k}")
print(f"Growth % per month : {growth_pcts}")
print(f"Months grew >20%   : {strong_growth}")
```

💡 `zip(sales, growth_pcts)` pairs up two equal-length lists element by element — an extremely common pattern whenever you've computed a derived list and need to filter the original based on the derived values.
