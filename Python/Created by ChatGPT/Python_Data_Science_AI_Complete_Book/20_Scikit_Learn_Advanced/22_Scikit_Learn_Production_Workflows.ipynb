## 1. Setup

```python
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, roc_auc_score
X,y=load_breast_cancer(return_X_y=True)
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=.2,random_state=42,stratify=y)
```

## 2. Core concepts

Scikit-learn Pipelines bundle preprocessing and modeling so that transformations are learned only from training folds, reducing leakage.

## Pipeline

Scale numeric features and train logistic regression.

```python
pipe=Pipeline([('scale',StandardScaler()),('model',LogisticRegression(max_iter=2000))])
pipe.fit(X_train,y_train)
proba=pipe.predict_proba(X_test)[:,1]
print('ROC AUC:',roc_auc_score(y_test,proba))
```

## Cross-validation

Estimate generalization performance across folds.

```python
scores=cross_val_score(pipe,X,y,cv=5,scoring='roc_auc')
print(scores)
print(scores.mean())
```
