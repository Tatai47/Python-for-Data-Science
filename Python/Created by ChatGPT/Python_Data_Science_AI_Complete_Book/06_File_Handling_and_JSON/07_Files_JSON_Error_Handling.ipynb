## 7.1 Reading & Writing Text Files

```python
# Writing a simple report to a file
with open("sales_report.txt", "w") as f:
    f.write("SALES REPORT – August 2025\n")
    f.write("=" * 40 + "\n")
    f.write("North Region : 1,250,000\n")
    f.write("South Region : 980,000\n")
    f.write("East Region  : 1,120,000\n")
    f.write("West Region  : 1,450,000\n")

print("Report written successfully.")
```

```python
# Reading the file
with open("sales_report.txt", "r") as f:
    content = f.read()
    print(content)
```

```python
# Reading line by line
with open("sales_report.txt", "r") as f:
    for line in f:
        print(line.strip())
```

## 7.2 Working with CSV Files (Without Pandas)

Although we will use Pandas later, knowing the built-in `csv` module is useful.

```python
import csv

# Writing CSV
data = [
    ["Region", "Sales", "Target"],
    ["North", 1250000, 1200000],
    ["South", 980000, 1000000],
    ["East", 1120000, 1100000],
    ["West", 1450000, 1400000]
]

with open("regional_sales.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerows(data)

print("CSV created.")
```

```python
# Reading CSV
with open("regional_sales.csv", "r") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

## 7.3 JSON – The Language of APIs & Config Files

Almost every modern system (web APIs, cloud services, config files) speaks JSON.

```python
import json

# Python dict → JSON string
employee = {
    "name": "Priya Sharma",
    "department": "Analytics",
    "role": "MBA Data Analyst",
    "skills": ["Python", "SQL", "Tableau"],
    "salary": 1250000,
    "active": True
}

json_string = json.dumps(employee, indent=4)
print(json_string)
```

```python
# Writing JSON to file
with open("employee.json", "w") as f:
    json.dump(employee, f, indent=4)

# Reading JSON from file
with open("employee.json", "r") as f:
    data = json.load(f)

print(data["name"])
print(data["skills"])
```

## 7.4 Error Handling – try / except

Real business data is messy. Files may be missing, numbers may be text, divisions by zero happen.  
Proper error handling keeps your analysis scripts robust.

```python
# Basic try-except
try:
    result = 100 / 0
except ZeroDivisionError:
    print("Error: Cannot divide by zero.")
```

```python
# Handling file not found
try:
    with open("non_existent_file.csv", "r") as f:
        content = f.read()
except FileNotFoundError:
    print("Error: The file does not exist. Please check the path.")
```

```python
# Multiple exceptions + else + finally
def safe_divide(a, b):
    try:
        result = a / b
    except ZeroDivisionError:
        print("Cannot divide by zero.")
        return None
    except TypeError:
        print("Both inputs must be numbers.")
        return None
    else:
        print("Division successful.")
        return result
    finally:
        print("Execution of safe_divide completed.")

print(safe_divide(100, 5))
print("-" * 30)
print(safe_divide(100, 0))
```

## 7.5 Practical Pattern – Robust Data Loading Function

```python
def load_sales_data(filename):
    """Safely load a CSV of regional sales."""
    try:
        with open(filename, "r") as f:
            reader = csv.DictReader(f)
            data = list(reader)
            print(f"Successfully loaded {len(data)} records from {filename}")
            return data
    except FileNotFoundError:
        print(f"File '{filename}' not found.")
        return []
    except Exception as e:
        print(f"Unexpected error: {e}")
        return []

records = load_sales_data("regional_sales.csv")
for row in records:
    print(row)
```
