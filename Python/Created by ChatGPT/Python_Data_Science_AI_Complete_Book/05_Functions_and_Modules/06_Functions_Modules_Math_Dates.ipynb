## 6.1 Defining Functions

A function packages a piece of logic so you can reuse it with different inputs.

### Why functions matter for MBA Analysts
- Avoid copy-paste of the same calculation
- Make reports consistent
- Easy to test and update business rules

```python
def calculate_profit(revenue, cost):
    """Calculate absolute profit and margin."""
    profit = revenue - cost
    margin = (profit / revenue) * 100 if revenue != 0 else 0
    return profit, margin

# Using the function
p, m = calculate_profit(1500000, 1050000)
print(f"Profit: ₹{p:,.0f} | Margin: {m:.1f}%")
```

```python
# Function with default arguments
def apply_discount(price, discount_rate=0.10, tax_rate=0.18):
    """Apply discount and then tax."""
    discounted = price * (1 - discount_rate)
    final = discounted * (1 + tax_rate)
    return round(final, 2)

print(apply_discount(10000))                # Uses defaults
print(apply_discount(10000, 0.20))          # Custom discount
print(apply_discount(10000, 0.15, 0.12))    # Custom discount + tax
```

```python
# Keyword arguments (order doesn't matter)
print(apply_discount(price=25000, tax_rate=0.05, discount_rate=0.25))
```

## 6.3 Lambda Functions – One-liner Functions

Useful for quick transformations (especially later with Pandas).

```python
square = lambda x: x ** 2
print(square(9))

margin = lambda rev, cost: ((rev - cost) / rev) * 100
print(f"Margin: {margin(1200000, 900000):.1f}%")
```

Python comes with many powerful built-in modules. You also create your own.

```python
import math

print(math.sqrt(144))
print(math.ceil(4.2))
print(math.floor(4.8))
print(math.pi)
print(math.log(100))
```

## 6.5 Working with Dates & Time

Critical for time-series analysis, cohort analysis, and reporting.

```python
from datetime import datetime, date, timedelta

today = date.today()
print("Today:", today)
print("Year:", today.year, "Month:", today.month, "Day:", today.day)

now = datetime.now()
print("Current time:", now)
print(now.strftime("%d-%b-%Y %H:%M"))
```

```python
# Date arithmetic
order_date = date(2025, 8, 10)
delivery_date = order_date + timedelta(days=5)
print("Order Date   :", order_date)
print("Expected Delivery:", delivery_date)

days_passed = (today - order_date).days
print(f"Days since order: {days_passed}")
```

```python
# Parsing dates from strings (common when reading Excel/CSV)
date_str = "15-Aug-2025"
parsed = datetime.strptime(date_str, "%d-%b-%Y").date()
print(parsed)
```

## 6.6 Practical Business Functions Library

```python
def cagr(start_value, end_value, years):
    """Compound Annual Growth Rate."""
    return (end_value / start_value) ** (1 / years) - 1

def moving_average(values, window=3):
    """Simple moving average."""
    result = []
    for i in range(len(values) - window + 1):
        window_avg = sum(values[i:i+window]) / window
        result.append(round(window_avg, 2))
    return result

def contribution_margin(selling_price, variable_cost):
    return selling_price - variable_cost

# Examples
print(f"CAGR: {cagr(1000000, 1850000, 3)*100:.1f}%")
print("3-month MA:", moving_average([120, 135, 150, 140, 160, 175]))
print("Contribution Margin:", contribution_margin(999, 420))
```
