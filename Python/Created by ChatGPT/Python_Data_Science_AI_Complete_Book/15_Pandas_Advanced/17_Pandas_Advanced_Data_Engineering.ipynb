## 2. Core concepts

Analytical datasets are rarely analysis-ready. Reshaping and feature engineering turn transactional tables into structures suitable for reporting and modeling.

## Create a time series

Build weekly sales and a rolling average.

```python
dates=pd.date_range('2026-01-01', periods=12, freq='W')
df=pd.DataFrame({'date':dates,'sales':[120,135,128,160,172,155,180,190,205,198,215,225]})
df['rolling_4w']=df['sales'].rolling(4).mean()
print(df)
```

## Pivot and melt

Move between long and wide representations.

```python
long=pd.DataFrame({'region':['East','East','West','West'],'month':['Jan','Feb','Jan','Feb'],'sales':[100,120,90,110]})
wide=long.pivot_table(index='region',columns='month',values='sales')
print(wide)
print(wide.reset_index().melt(id_vars='region',var_name='month',value_name='sales'))
```

## Pipeline

Compose cleaning operations in a readable chain.

```python
raw=pd.DataFrame({'name':['A','B','C'],'sales':['100','200','150']})
clean=(raw.assign(name=lambda d:d['name'].str.strip().str.upper(),sales=lambda d:pd.to_numeric(d['sales']))
       .assign(sales_growth=lambda d:d['sales'].pct_change()))
print(clean)
```
