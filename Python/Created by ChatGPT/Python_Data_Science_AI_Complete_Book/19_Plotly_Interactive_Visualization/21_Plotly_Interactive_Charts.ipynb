## 2. Core concepts

Plotly creates browser-friendly interactive figures. Interactivity is most useful when users need to explore dimensions rather than consume one fixed narrative.

## Interactive scatter

Explore revenue, ad spend, and channel.

```python
import pandas as pd
import plotly.express as px
df=pd.DataFrame({'ad_spend':[10,20,30,40,50],'revenue':[50,72,90,135,150],'channel':['Search','Search','Social','Social','Email']})
fig=px.scatter(df,x='ad_spend',y='revenue',color='channel',size='revenue',hover_data=['channel'],title='Ad Spend vs Revenue')
fig.show()
```

## 3-D chart

Visualize three business dimensions.

```python
import plotly.express as px
fig=px.scatter_3d(df,x='ad_spend',y='revenue',z=df['revenue']/df['ad_spend'],color='channel')
fig.show()
```
