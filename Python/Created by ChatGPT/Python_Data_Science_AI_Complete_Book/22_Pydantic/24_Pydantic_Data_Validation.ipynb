## 2. Core concepts

Pydantic models act as executable schemas. This is useful at API boundaries, configuration layers, ETL validation points, and structured LLM outputs.

## Basic model

Validate a customer record.

```python
from pydantic import BaseModel, Field
class Customer(BaseModel):
    customer_id:int
    name:str
    age:int=Field(ge=18,le=120)
    segment:str

customer=Customer(customer_id=101,name='Asha',age=29,segment='Premium')
print(customer.model_dump())
```

## Nested model

Validate an order containing line items.

```python
class Item(BaseModel):
    sku:str
    qty:int=Field(gt=0)
    price:float=Field(gt=0)
class Order(BaseModel):
    order_id:str
    items:list[Item]
order=Order(order_id='O-1',items=[{'sku':'P1','qty':2,'price':99.0}])
print(order.model_dump())
```
