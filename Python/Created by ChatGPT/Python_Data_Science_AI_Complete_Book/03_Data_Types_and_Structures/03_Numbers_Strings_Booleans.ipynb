## 3.1 Numbers – int & float

In business analysis you constantly work with:
- Integers → number of units, customers, days
- Floats → revenue, prices, percentages, ratios

```python
# Integers
units_sold = 18500
number_of_stores = 42
days_in_month = 30

# Floats
revenue = 2456789.50
unit_price = 132.75
discount_rate = 0.15

print(type(units_sold))
print(type(revenue))
```

```python
# Arithmetic operations
print(10 + 3)    # Addition
print(10 - 3)    # Subtraction
print(10 * 3)    # Multiplication
print(10 / 3)    # Division (always returns float)
print(10 // 3)   # Floor division
print(10 % 3)    # Modulus (remainder)
print(10 ** 3)   # Exponentiation (10 cubed)
```

```python
# Business calculation example
list_price = 999.00
discount = 0.18
tax_rate = 0.18

discounted_price = list_price * (1 - discount)
final_price = discounted_price * (1 + tax_rate)

print(f"List Price      : ₹{list_price:,.2f}")
print(f"After Discount  : ₹{discounted_price:,.2f}")
print(f"Final Price (incl. tax): ₹{final_price:,.2f}")
```

### Useful Number Functions

```python
print(round(3.14159, 2))     # Round to 2 decimal places
print(abs(-1500))            # Absolute value
print(max(120, 95, 180, 75)) # Maximum
print(min(120, 95, 180, 75)) # Minimum
print(pow(2, 5))             # 2 to the power 5
```

## 3.2 Casting (Type Conversion)

Very common when reading data from Excel/CSV where numbers come as text.

```python
# String to number
revenue_str = "2450000"
revenue = int(revenue_str)
print(revenue, type(revenue))

price_str = "132.75"
price = float(price_str)
print(price, type(price))

# Number to string (useful for reports)
units = 18500
message = "Units sold: " + str(units)
print(message)
```

## 3.3 Strings – Working with Text Data

Product names, customer segments, regions, feedback – all are strings.

```python
product = "Premium Wireless Headphones"
region = 'North'
feedback = """Customer loved the product.
Delivery was fast.
Would recommend."""

print(product)
print(region)
print(feedback)
```

```python
# Common string operations
name = "  MBA Data Analyst  "
print(name.strip())          # Remove leading/trailing spaces
print(name.upper())
print(name.lower())
print(name.title())
print(len(name.strip()))

# Checking content
email = "analyst@company.com"
print(email.startswith("analyst"))
print(email.endswith(".com"))
print("@" in email)
```

```python
# String formatting – f-strings (preferred)
customer = "Rahul Sharma"
order_value = 12450.75
print(f"Customer {customer} placed an order of ₹{order_value:,.2f}")
```

```python
# Splitting and joining (very useful for data cleaning)
regions = "North,South,East,West"
region_list = regions.split(",")
print(region_list)

joined = " | ".join(region_list)
print(joined)
```

## 3.4 Booleans – True / False Logic

Used heavily in filtering data and business rules.

```python
is_profitable = True
has_discount = False
target_achieved = 1250000 > 1000000

print(is_profitable)
print(target_achieved)
print(type(target_achieved))
```

```python
# Comparison operators
a = 100
b = 150

print(a == b)   # Equal
print(a != b)   # Not equal
print(a > b)
print(a < b)
print(a >= 100)
print(a <= 100)
```

```python
# Logical operators
revenue = 1250000
margin = 0.22

print(revenue > 1000000 and margin > 0.20)  # Both conditions true
print(revenue > 2000000 or margin > 0.20)   # At least one true
print(not (margin < 0.15))                  # Negation
```
