## 4.1 Lists – Ordered & Mutable Collection

Lists are the most used data structure for holding sequences of values (sales figures, product names, customer IDs, etc.).

````python
# Creating lists
regions = ["North", "South", "East", "West"]
sales = [1250000, 980000, 1120000, 1450000]
mixed = ["Premium", 45000, True, 3.5]

print(regions)
print(sales)
````

````python
# Accessing elements (0-based indexing)
print(regions[0])      # First element
print(regions[-1])     # Last element
print(sales[1:3])      # Slicing
````

````python
# Modifying lists
regions.append("Central")           # Add at end
regions.insert(1, "North-East")     # Insert at position
print(regions)

sales[1] = 1020000                  # Update value
print(sales)

removed = regions.pop()             # Remove last
print("Removed:", removed)
print(regions)
````

````python
# Useful list methods
sales = [125, 98, 112, 145, 98, 130]
print("Length:", len(sales))
print("Max:", max(sales))
print("Min:", min(sales))
print("Sum:", sum(sales))
print("Average:", sum(sales)/len(sales))
print("Count of 98:", sales.count(98))
print("Index of 145:", sales.index(145))

sales.sort()
print("Sorted:", sales)
sales.reverse()
print("Reversed:", sales)
````

````python
# List comprehension – powerful & concise (very useful later with Pandas)
sales = [125, 98, 112, 145, 98, 130]
high_sales = [s for s in sales if s > 110]
print(high_sales)

growth = [s * 1.1 for s in sales]
print(growth)
````

## 4.2 Tuples – Ordered & Immutable

Use tuples when the data should not change (e.g., coordinates, fixed categories, database records).

````python
point = (12.5, 45.8)
quarters = ("Q1", "Q2", "Q3", "Q4")

print(quarters[2])
print(len(quarters))

# Tuples are immutable – this will fail:
# quarters[0] = "Quarter 1"
````

````python
# Unpacking (very common)
q1, q2, q3, q4 = quarters
print(q1, q3)
````

## 4.3 Sets – Unique & Unordered

Perfect for removing duplicates and set operations (union, intersection).

````python
customers_q1 = {"Rahul", "Priya", "Amit", "Sneha", "Rahul"}
customers_q2 = {"Priya", "Vikram", "Sneha", "Ananya"}

print(customers_q1)  # Duplicates automatically removed

print("Common customers:", customers_q1 & customers_q2)          # Intersection
print("All customers:", customers_q1 | customers_q2)             # Union
print("Only in Q1:", customers_q1 - customers_q2)                # Difference
````

## 4.4 Dictionaries – Key-Value Pairs (Most Important for Analysis)

Dictionaries map keys to values. Think of them as lookup tables or Excel VLOOKUP on steroids.

````python
# Creating a dictionary
region_sales = {
    "North": 1250000,
    "South": 980000,
    "East": 1120000,
    "West": 1450000
}

print(region_sales)
print(region_sales["West"])
````

````python
# Common operations
region_sales["Central"] = 875000          # Add new key
region_sales["South"] = 1020000           # Update existing
print(region_sales)

print(region_sales.keys())
print(region_sales.values())
print(region_sales.items())

print("North" in region_sales)            # Check existence
print(region_sales.get("North-East", 0))  # Safe get with default
````

````python
# Looping through dictionary
for region, sales in region_sales.items():
    print(f"{region:10} → ₹{sales:,}")
````

````python
# Nested dictionary – very common for business data
product_info = {
    "SKU001": {"name": "Wireless Mouse", "price": 799, "stock": 150},
    "SKU002": {"name": "Mechanical Keyboard", "price": 3499, "stock": 45},
    "SKU003": {"name": "USB-C Hub", "price": 1299, "stock": 80}
}

print(product_info["SKU002"]["name"])
print(product_info["SKU002"]["price"])
````

## 4.5 Business Mini-Case: Sales Performance Dashboard (Dictionary + List)

````python
monthly_sales = {
    "Jan": 1850000,
    "Feb": 1720000,
    "Mar": 2100000,
    "Apr": 1950000,
    "May": 2280000,
    "Jun": 2410000
}

total = sum(monthly_sales.values())
average = total / len(monthly_sales)
best_month = max(monthly_sales, key=monthly_sales.get)
worst_month = min(monthly_sales, key=monthly_sales.get)

print("=" * 45)
print("HALF-YEARLY SALES SUMMARY")
print("=" * 45)
for month, sales in monthly_sales.items():
    print(f"{month:5} : ₹{sales:,}")
print("-" * 45)
print(f"Total     : ₹{total:,}")
print(f"Average   : ₹{average:,.0f}")
print(f"Best Month: {best_month} (₹{monthly_sales[best_month]:,})")
print(f"Worst Month: {worst_month} (₹{monthly_sales[worst_month]:,})")
print("=" * 45)
````
