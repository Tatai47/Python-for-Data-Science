## 1. Setup

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf
```

## 2. Core concepts

Statsmodels emphasizes estimation, statistical tests, confidence intervals, and model summaries. It complements scikit-learn rather than replacing it.

## OLS regression

Estimate the effect of advertising spend on sales.

```python
df=pd.DataFrame({'sales':[120,140,155,170,190,210,230],'ad_spend':[10,15,18,22,25,30,35]})
model=smf.ols('sales ~ ad_spend',data=df).fit()
print(model.summary())
```

## Prediction vs inference

Contrast a coefficient with a prediction goal.

```python
print(model.params)
print(model.conf_int())
```
