# Heading 1 : List Usage in Real World

### Output shoul look like : 
Completed
Todo List remaining
- Buy groceries
- Pay the bills
- Schedule Meeting
- Go for a run

```python
lst = ["Buy groceries","Clean the house","Pay the bills"]
#adding the task
lst.append("Schedule Meeting")
lst.append("Go for a run")
## Removing a task
lst.remove("Clean the house")
print("Completed")
##checking if a task is in the list
if "Clean the house" in lst:
    print("Hey don't forget to pay the utility bills.")

print("Todo List remaining")
for i in lst:
    print(f"- {i}")
```
