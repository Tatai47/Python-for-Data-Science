### Data serialization
* Using JSON package to convert Dictionary to JSON
* Then converting JSON to Dictionary to JSON

```python
import json
data = {
    'name':"Chittajit Chakraborty",
    'Age':24
}
jsonStr = json.dumps(data)
print(jsonStr)
print(type(jsonStr))

dictConvert = json.loads(jsonStr)
print(dictConvert)
print(type(dictConvert))
```

### Working with CSV
* Reading / Creating a excel (CSV) file 
* Then fetching all data inside this CSV file 
* Then putting or inserting data into that CSV file

```python
import csv
with open("example.csv", mode='w',newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['name','age'])
    writer.writerow(['Chittajit Chakraborty',24])

with open('example.csv',mode='r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

### Date time
* Fetching today's date + time
* Then calculating previous day time asper today's current time + date

```python
from datetime import datetime, timedelta
now = datetime.now()
print(now)
yesterday = now - timedelta(days=1)
print(yesterday)
```

### Regular expression
* d+ Means digits
* searching the digits in the string
* then using group() print the digit in group

```python
import re
pattern = r'\d+'
text = "There are 123 apples"
match = re.search(pattern,text)
print(match.group())
print(type(match.group()))
for i in match.group():
    print(i)
```
