### Filter Function in python

The filter() function constructs an iterator from elements of an interable for which a function returns true. It is used to filter out items from a list 
or any other iterable based on a condition.


### *Okay run atleast 1 time first function then execute the second one either it will show error.

```python
def even(number):
    if number%2==0:
        return True 
```

```python
lst = [1,2,3,4,5,5,6,67,7,8,98,9,0,9,78,6,3,4,34,7,23,42376,537,37,57568]
list(filter(even,lst))
```

### Create a filter so that onyl numbers will show which is more then 5
`numbers = [1,2,3,4,5,5,6,67,7,8,98,9,0,9,78,6,3,4,34,7,23,42376,537,37,57568]`


```python
def moreThanFifty(x):
    if x > 50 :
        return x
```

### Method 1
Using `filter()`

```python
list(filter(moreThanFifty,lst))
```

### Method 2
Using `lambda`

```python
list(filter(lambda x: x>50, lst))
```

### Filter with a lambda function and multiple conditions

```python
list(filter(lambda x:x>5 and x%2==0, lst))
```

### Filter on disctionaries
Apply filter() to check if the age is greater than 25 or not

```python
ageData = [
   { 'Name':'Chittajit', 'Age':24},
   { 'Name':'Anika', 'Age':23},
   { 'Name':'Romit', 'Age':23},
   { 'Name':'Sayan', 'Age':18}
]
def ageGreaterThan20(person):
    return person['Age']>20

list(filter(ageGreaterThan20,ageData))
```
