The important idea is:

> Markdown explains the program; Python code executes the program.

## 8. Mathematical Formulas

Inline mathematics:

The mean can be represented as $\bar{x}$.

A full equation:

$$
\bar{x} = \frac{1}{n}\sum_{i=1}^{n}x_i
$$

A common machine-learning metric:

$$
R^2 = 1 - \frac{\sum_i (y_i-\hat{y}_i)^2}
{\sum_i (y_i-\bar{y})^2}
$$

LaTeX-style mathematics is especially useful for statistics, data science, physics, engineering, and machine learning.

## 18. Working with Pandas

Pandas is commonly used for tabular data.

The next example creates a DataFrame directly in Python.


``````python
import pandas as pd

df = pd.DataFrame({
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor"],
    "PurchasePrice": [45000, 900, 1800, 12000],
    "SellingPrice": [50000, 1200, 2500, 15000],
    "Quantity": [2, 10, 5, 3]
})

df
``````

## 19. Calculated Columns

The Markdown cell explains the logic:

$$
Profit = SellingPrice - PurchasePrice
$$

$$
TotalProfit = Profit \times Quantity
$$


``````python
df["Profit"] = df["SellingPrice"] - df["PurchasePrice"]
df["TotalProfit"] = df["Profit"] * df["Quantity"]

df
``````

## 20. Basic Pandas Analysis

``````python
print("Rows and columns:", df.shape)
print("Total quantity:", df["Quantity"].sum())
print("Average selling price:", df["SellingPrice"].mean())
print("Minimum selling price:", df["SellingPrice"].min())
print("Maximum selling price:", df["SellingPrice"].max())
``````

## 21. Filtering Data

## 24. Matplotlib Visualization

A Markdown cell can describe what a chart represents.

The following Code cell creates a bar chart of total profit by product.


``````python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.bar(df["Product"], df["TotalProfit"])
plt.title("Total Profit by Product")
plt.xlabel("Product")
plt.ylabel("Total Profit")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
``````

## 29. Missing Values

A standard workflow is:

1. Detect missing values.
2. Understand why they are missing.
3. Drop or replace them appropriately.


``````python
sample = pd.DataFrame({
    "Name": ["A", "B", "C", "D"],
    "Marks": [80, None, 75, 90]
})

print("Missing values:")
print(sample.isna())

print("\nMissing count:")
print(sample.isna().sum())
``````

## 30. Filling Missing Values

Here we replace a missing mark with the average mark.


``````python
mean_marks = sample["Marks"].mean()
sample["Marks"] = sample["Marks"].fillna(mean_marks)

sample
``````

## 39. A Complete Mini Project

This example combines Markdown, Python, Pandas, calculations, filtering, and visualization.

### Business question

> Which product generates the most total profit?


``````python
import pandas as pd
import matplotlib.pyplot as plt

sales = pd.DataFrame({
    "Product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Printer"],
    "PurchasePrice": [45000, 900, 1800, 12000, 8000],
    "SellingPrice": [50000, 1200, 2500, 15000, 10000],
    "Quantity": [2, 10, 5, 3, 4]
})

sales["Profit"] = sales["SellingPrice"] - sales["PurchasePrice"]
sales["TotalProfit"] = sales["Profit"] * sales["Quantity"]

sales
``````

``````python
best_product = sales.loc[sales["TotalProfit"].idxmax(), "Product"]
best_profit = sales["TotalProfit"].max()

print("Best product:", best_product)
print("Highest total profit:", best_profit)
``````

``````python
plt.figure(figsize=(9, 5))
plt.bar(sales["Product"], sales["TotalProfit"])
plt.title("Total Profit by Product")
plt.xlabel("Product")
plt.ylabel("Total Profit")
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
``````

## 42. The Most Important Rule

When writing a Jupyter Notebook:

### Put this in a Markdown cell

```markdown
## Calculate Profit

Profit is calculated as:

$$
Profit = SellingPrice - PurchasePrice
$$
```

### Put this in a Python Code cell

```python
profit = selling_price - purchase_price
print(profit)
```

The Markdown **describes and documents** the program.

The Python Code cell **executes** the program.

That separation is what makes a Jupyter Notebook readable, reproducible, and useful as a technical report.


# Conclusion

You can combine almost all ordinary Markdown documentation with Python programming inside a `.ipynb` file.

A good notebook should:

- Explain the objective with Markdown.
- Keep code in executable Code cells.
- Use headings to organize the analysis.
- Use tables and formulas to communicate results.
- Use Pandas/NumPy for data processing.
- Use Matplotlib for visualization.
- Show outputs immediately below the relevant code.
- Finish with findings and conclusions.

**Think of a Jupyter Notebook as a document in which documentation and executable Python live together.**
