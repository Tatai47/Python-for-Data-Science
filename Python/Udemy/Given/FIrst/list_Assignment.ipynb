### Assignment 7: Nested Lists

Create a nested list representing a 3x3 matrix and print the matrix. Access and print the element at the second row and third column.

### Assignment 8: List of Dictionaries

Create a list of dictionaries where each dictionary represents a student with keys 'name' and 'score'. Sort the list of dictionaries by the 'score' in descending order and print the sorted list.

### Assignment 9: Matrix Transposition

Write a function that takes a 3x3 matrix (nested list) as input and returns its transpose. Print the original and transposed matrices.

### Assignment 10: Flattening a Nested List

Write a function that takes a nested list and flattens it into a single list. Print the original and flattened lists.

### Assignment 11: List Manipulation

Write a function that rotates a list by n positions. Print the original and rotated lists.

### Assignment 15: List Intersection

Write a function that takes two lists and returns a new list containing only the elements that are present in both lists. Print the intersected list.

```python
# Assignment 1: Creating and Accessing Lists
positive_integers = list(range(1, 21))
print("Assignment 1:", positive_integers)

# Assignment 2: Accessing List Elements
first_elem = positive_integers[0]
middle_elem = positive_integers[len(positive_integers) // 2]
last_elem = positive_integers[-1]
print("Assignment 2 - First:", first_elem, "Middle:", middle_elem, "Last:", last_elem)

# Assignment 3: List Slicing
first_five = positive_integers[:5]
last_five = positive_integers[-5:]
sub_list = positive_integers[5:16]
print("Assignment 3 - First 5:", first_five, "| Last 5:", last_five, "| Index 5 to 15:", sub_list)

# Assignment 4: List Comprehensions
squares = [x**2 for x in range(1, 11)]
print("Assignment 4:", squares)

# Assignment 5: Filtering Lists
evens = [x for x in positive_integers if x % 2 == 0]
print("Assignment 5:", evens)

# Assignment 6: List Methods
random_nums = [42, 5, 12, 42, 89, 5, 23, 12]
unique_sorted_asc = sorted(list(set(random_nums)))
unique_sorted_desc = sorted(list(set(random_nums)), reverse=True)
print("Assignment 6 - Ascending:", unique_sorted_asc, "| Descending:", unique_sorted_desc)

# Assignment 7: Nested Lists
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
print("Assignment 7 - Matrix:", matrix)
print("Assignment 7 - Element at row 2, col 3:", matrix[1][2])

# Assignment 8: List of Dictionaries
students = [
    {"name": "Alice", "score": 85},
    {"name": "Bob", "score": 92},
    {"name": "Charlie", "score": 78}
]
sorted_students = sorted(students, key=lambda s: s["score"], reverse=True)
print("Assignment 8:", sorted_students)

# Assignment 9: Matrix Transposition
def transpose_matrix(mat):
    return [[mat[j][i] for j in range(len(mat))] for i in range(len(mat[0]))]

original_matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
transposed = transpose_matrix(original_matrix)
print("Assignment 9 - Original:", original_matrix, "| Transposed:", transposed)

# Assignment 10: Flattening a Nested List
def flatten_list(nested):
    flat = []
    for item in nested:
        if isinstance(item, list):
            flat.extend(flatten_list(item))
        else:
            flat.append(item)
    return flat

nested_data = [1, [2, 3], [4, [5, 6]], 7]
print("Assignment 10 - Original:", nested_data, "| Flattened:", flatten_list(nested_data))

# Assignment 11: List Manipulation
nums_11 = list(range(1, 11))
# Remove indices 2, 4, 6 (values at indices in original list: 3, 5, 7)
for idx in sorted([2, 4, 6], reverse=True):
    nums_11.pop(idx)
nums_11.insert(5, 99)
print("Assignment 11:", nums_11)

# Assignment 12: List Zipping
list_a = ['a', 'b', 'c']
list_b = [1, 2, 3]
zipped_result = list(zip(list_a, list_b))
print("Assignment 12:", zipped_result)

# Assignment 13: List Reversal
def reverse_list(lst):
    return lst[::-1]

sample_list = [10, 20, 30, 40]
print("Assignment 13 - Original:", sample_list, "| Reversed:", reverse_list(sample_list))

# Assignment 14: List Rotation
def rotate_list(lst, n):
    n = n % len(lst) if lst else 0
    return lst[-n:] + lst[:-n]

rot_list = [1, 2, 3, 4, 5]
print("Assignment 14 - Original:", rot_list, "| Rotated by 2:", rotate_list(rot_list, 2))

# Assignment 15: List Intersection
def list_intersection(l1, l2):
    return list(set(l1) & set(l2))

list1 = [1, 2, 3, 4, 5]
list2 = [4, 5, 6, 7, 8]
print("Assignment 15:", list_intersection(list1, list2))
```
