```python
# Question 15: Check if a number is even or odd
num_15 = 7
if num_15 % 2 == 0:
    print(f"Q15: {num_15} is even.")
else:
    print(f"Q15: {num_15} is odd.")

# Question 16: Find the sum of the first n natural numbers
n_16 = 10
total_sum = sum(range(1, n_16 + 1))
print(f"Q16: The sum of the first {n_16} natural numbers is {total_sum}")

# Question 17: Check if a year is a leap year
year_17 = 2024
if (year_17 % 4 == 0 and year_17 % 100 != 0) or (year_17 % 400 == 0):
    print(f"Q17: {year_17} is a leap year.")
else:
    print(f"Q17: {year_17} is not a leap year.")

# Question 18: Reverse a string
text_18 = "Python"
reversed_text = text_18[::-1]
print(f"Q18: The reversed string is: {reversed_text}")

# Question 19: Check if a string is a palindrome
text_19 = "radar"
if text_19 == text_19[::-1]:
    print(f"Q19: '{text_19}' is a palindrome.")
else:
    print(f"Q19: '{text_19}' is not a palindrome.")

# Question 20: Sort a list of numbers in ascending order
numbers_20 = [42, 12, 89, 5, 23]
numbers_20.sort()
print(f"Q20: Sorted list in ascending order: {numbers_20}")
```
