### Assignment 4: Nested Tuples

### Assignment 5: Tuple Concatenation

Concatenate two tuples: (1, 2, 3) and (4, 5, 6). Print the resulting tuple.

### Assignment 6: Tuple Methods

Create a tuple with duplicate elements and count the occurrences of an element. Find the index of the first occurrence of an element in the tuple.

Create a tuple with duplicate elements. Convert it to a set to remove duplicates and print the resulting set.

### Assignment 15: Tuple Functions

Write functions that take a tuple and return the minimum, maximum, and sum of the elements. Print the results for a sample tuple.

```python
# Assignment 1: Creating and Accessing Tuples
positive_integers_tuple = tuple(range(1, 11))
print("Assignment 1:", positive_integers_tuple)

# Assignment 2: Accessing Tuple Elements
first_elem = positive_integers_tuple[0]
middle_elem = positive_integers_tuple[len(positive_integers_tuple) // 2]
last_elem = positive_integers_tuple[-1]
print("Assignment 2 - First:", first_elem, "Middle:", middle_elem, "Last:", last_elem)

# Assignment 3: Tuple Slicing
first_three = positive_integers_tuple[:3]
last_three = positive_integers_tuple[-3:]
sub_tuple = positive_integers_tuple[2:6]
print("Assignment 3 - First 3:", first_three, "| Last 3:", last_three, "| Index 2 to 5:", sub_tuple)

# Assignment 4: Nested Tuples
matrix_tuple = (
    (1, 2, 3),
    (4, 5, 6),
    (7, 8, 9)
)
print("Assignment 4 - Matrix:", matrix_tuple)
print("Assignment 4 - Element at row 2, col 3:", matrix_tuple[1][2])

# Assignment 5: Tuple Concatenation
tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)
concatenated = tuple1 + tuple2
print("Assignment 5:", concatenated)

# Assignment 6: Tuple Methods
dup_tuple = (1, 2, 3, 2, 4, 2, 5)
count_twos = dup_tuple.count(2)
index_of_four = dup_tuple.index(4)
print("Assignment 6 - Count of 2:", count_twos, "| Index of first 4:", index_of_four)

# Assignment 7: Unpacking Tuples
five_tuple = (10, 20, 30, 40, 50)
a, b, c, d, e = five_tuple
print("Assignment 7 - Unpacked variables:", a, b, c, d, e)

# Assignment 8: Tuple Conversion
num_list = [1, 2, 3, 4, 5]
converted_tuple = tuple(num_list)
print("Assignment 8:", converted_tuple)

# Assignment 9: Tuple of Tuples
tuple_of_tuples = ((1, 2, 3), (4, 5, 6), (7, 8, 9))
print("Assignment 9:", tuple_of_tuples)

# Assignment 10: Tuple and List
initial_tuple = (1, 2, 3, 4, 5)
temp_list = list(initial_tuple)
temp_list.append(6)
final_tuple = tuple(temp_list)
print("Assignment 10:", final_tuple)

# Assignment 11: Tuple and String
char_tuple = ('P', 'y', 't', 'h', 'o', 'n')
joined_string = "".join(char_tuple)
print("Assignment 11:", joined_string)

# Assignment 12: Tuple and Dictionary
dict_with_tuple_keys = {(1, 2): 100, (3, 4): 200}
print("Assignment 12:", dict_with_tuple_keys)

# Assignment 13: Nested Tuple Iteration
nested_tup = ((1, 2), (3, 4), (5, 6))
print("Assignment 13:")
for row in nested_tup:
    for item in row:
        print(item, end=" ")
print()

# Assignment 14: Tuple and Set
tup_with_dups = (1, 2, 2, 3, 4, 4, 5)
unique_set = set(tup_with_dups)
print("Assignment 14:", unique_set)

# Assignment 15: Tuple Functions
def tuple_stats(tup):
    return min(tup), max(tup), sum(tup)

sample_tup = (10, 5, 20, 15)
min_val, max_val, sum_val = tuple_stats(sample_tup)
print("Assignment 15 - Min:", min_val, "| Max:", max_val, "| Sum:", sum_val)
```
