## 3. Basic Operators (Arithmetic, Comparison, Logical)

**Question 11:** Write a Python program to perform arithmetic operations: addition, subtraction, multiplication, and division.

```python
# Your code here
# Define two numbers
num1 = 20
num2 = 5

# Perform arithmetic operations
addition = num1 + num2
subtraction = num1 - num2
multiplication = num1 * num2
division = num1 / num2  # Returns a float

# Display the results
print(f"Addition: {num1} + {num2} = {addition}")
print(f"Subtraction: {num1} - {num2} = {subtraction}")
print(f"Multiplication: {num1} * {num2} = {multiplication}")
print(f"Division: {num1} / {num2} = {division}")
```

**Question 12:** Write a Python program to demonstrate comparison operators: equal to, not equal to, greater than, less than.

```python
# Your code here
# Define two numbers for comparison
num1 = 15
num2 = 10

# Demonstrate comparison operators
print(f"Is {num1} equal to {num2}? {num1 == num2}")
print(f"Is {num1} not equal to {num2}? {num1 != num2}")
print(f"Is {num1} greater than {num2}? {num1 > num2}")
print(f"Is {num1} less than {num2}? {num1 < num2}")
```

**Question 13:** Write a Python program to demonstrate logical operators: and, or, not.

```python
# Your code here
# Define boolean variables for demonstration
x = True
y = False

# Demonstrate logical operators
print(f"x and y: {x and y}")  # True only if both are True
print(f"x or y: {x or y}")    # True if at least one is True
print(f"not x: {not x}")      # Reverses the boolean value
```

**Question 17:** Write a Python program to check if a year is a leap year.

```python
# Your code here
# Define the year
year = 2024

# Check leap year conditions
if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"{year} is a leap year.")
else:
    print(f"{year} is not a leap year.")
```

**Question 18:** Write a Python program to reverse a string.

```python
# Your code here
# Define the string
text = "Python"

# Reverse the string using slicing
reversed_text = text[::-1]

# Display the result
print(f"The reversed string is: {reversed_text}")
```

**Question 19:** Write a Python program to check if a string is a palindrome.

```python
# Your code here
# Define the string
text = "radar"

# Check if the string equals its reverse
if text == text[::-1]:
    print(f"'{text}' is a palindrome.")
else:
    print(f"'{text}' is not a palindrome.")
```
