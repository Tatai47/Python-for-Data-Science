### Assignment 3: Set Operations

### Assignment 5: Filtering Sets

Create a set with duplicate elements and remove the duplicates using set methods. Print the modified set.

### Assignment 7: Subsets and Supersets

Create two sets: one with the first 5 positive integers and another with the first 3 positive integers. Check if the second set is a subset of the first set and if the first set is a superset of the second set. Print the results.

### Assignment 8: Frozenset

Create a set and remove elements from it until it is empty. Print the set after each removal.

Create two sets and update the first set with the symmetric difference of the two sets. Print the modified first set.

### Assignment 14: Set Membership Testing

Create a set and test if certain elements are present in the set. Print the results.

### Assignment 15: Set of Tuples

Create a set containing tuples, where each tuple contains two elements. Print the set.

```python
# Assignment 1: Creating and Accessing Sets
set_1 = set(range(1, 11))
print("Assignment 1:", set_1)

# Assignment 2: Adding and Removing Elements
set_2 = set(range(1, 11))
set_2.add(11)
set_2.discard(1)  # Using discard avoids KeyError if 1 is already removed
print("Assignment 2:", set_2)

# Assignment 3: Set Operations
set_a = {1, 2, 3, 4, 5}
set_b = {2, 4, 6, 8, 10}
print("Assignment 3 - Union:", set_a | set_b)
print("Assignment 3 - Intersection:", set_a & set_b)
print("Assignment 3 - Difference (A - B):", set_a - set_b)
print("Assignment 3 - Symmetric Difference:", set_a ^ set_b)

# Assignment 4: Set Comprehensions
squares_set = {x**2 for x in range(1, 11)}
print("Assignment 4:", squares_set)

# Assignment 5: Filtering Sets
even_set = {x for x in set_1 if x % 2 == 0}
print("Assignment 5:", even_set)

# Assignment 6: Set Methods
duplicate_set = {1, 2, 2, 3, 4, 4, 5}
# Sets inherently contain unique elements, converting to set removes duplicates
unique_set = set(duplicate_set)
print("Assignment 6:", unique_set)

# Assignment 7: Subsets and Supersets
set_full = {1, 2, 3, 4, 5}
set_sub = {1, 2, 3}
print("Assignment 7 - Is subset:", set_sub.issubset(set_full))
print("Assignment 7 - Is superset:", set_full.issuperset(set_sub))

# Assignment 8: Frozenset
frozen = frozenset(range(1, 6))
print("Assignment 8:", frozen)

# Assignment 9: Set and List Conversion
original_set = {1, 2, 3, 4, 5}
conv_list = list(original_set)
conv_list.append(6)
final_set = set(conv_list)
print("Assignment 9:", final_set)

# Assignment 10: Set and Dictionary
# Note: Sets are mutable and unhashable, so frozensets must be used as dictionary keys
dict_with_sets = {frozenset([1, 2]): 10, frozenset([3, 4]): 20}
print("Assignment 10:", dict_with_sets)

# Assignment 11: Iterating Over Sets
iter_set = {10, 20, 30, 40}
print("Assignment 11:")
for item in iter_set:
    print(item)

# Assignment 12: Removing Elements from Sets
rem_set = {1, 2, 3}
print("Assignment 12:")
while rem_set:
    removed_element = rem_set.pop()
    print(f"Removed {removed_element}, Set now: {rem_set}")

# Assignment 13: Set Symmetric Difference Update
s1 = {1, 2, 3, 4}
s2 = {3, 4, 5, 6}
s1.symmetric_difference_update(s2)
print("Assignment 13:", s1)

# Assignment 14: Set Membership Testing
member_set = {10, 20, 30, 40, 50}
print("Assignment 14 - Is 30 in set?", 30 in member_set)
print("Assignment 14 - Is 99 in set?", 99 in member_set)

# Assignment 15: Set of Tuples
tuple_set = {(1, 2), (3, 4), (5, 6)}
print("Assignment 15:", tuple_set)
```
