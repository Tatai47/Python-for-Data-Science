## 5.1 Tuples – Immutable Sequences

```python
point = (10, 20)
employee = ("Priya", "Analytics", 1250000)
single = (42,)          # Note the comma

print(employee[0])
print(employee[1:])
name, dept, salary = employee   # Unpacking
print(name, dept, salary)
```

Tuples are faster and safer when data should not change (coordinates, RGB colors, database records).

```python
# Tuple methods
t = (1, 2, 2, 3, 2, 4)
print(t.count(2))
print(t.index(3))
```

## 5.2 Sets – Unique Elements & Fast Membership

```python
regions = {"North", "South", "East", "West", "North"}
print(regions)          # Duplicates removed

regions.add("Central")
regions.discard("East")
print(regions)

print("North" in regions)
```

```python
# Set operations – powerful for business analysis
customers_q1 = {"C001", "C002", "C003", "C004"}
customers_q2 = {"C003", "C004", "C005", "C006"}

print("New customers:", customers_q2 - customers_q1)
print("Common customers:", customers_q1 & customers_q2)
print("All customers:", customers_q1 | customers_q2)
print("Symmetric difference:", customers_q1 ^ customers_q2)
```
