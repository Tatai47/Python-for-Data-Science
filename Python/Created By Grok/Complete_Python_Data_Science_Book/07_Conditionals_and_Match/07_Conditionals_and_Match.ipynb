## 7.1 if / elif / else

```python
sales = 1350000
target = 1200000

if sales > target * 1.1:
    status = "Excellent – significantly above target"
elif sales >= target:
    status = "Met target"
else:
    status = "Below target – action needed"
print(status)
```

## 7.2 Ternary Expression & Nested Conditions

```python
sales = 950000
status = "Above target" if sales >= 1000000 else "Below target"
print(status)

# Nested
score = 78
if score >= 90:
    grade = "A"
elif score >= 75:
    grade = "B"
elif score >= 60:
    grade = "C"
else:
    grade = "D"
print(grade)
```

## 7.3 match-case (Python 3.10+)

```python
region = "West"

match region:
    case "North":
        manager = "Aisha"
    case "South":
        manager = "Rahul"
    case "East" | "West":
        manager = "Shared Team"
    case _:
        manager = "Unassigned"
print(manager)
```
