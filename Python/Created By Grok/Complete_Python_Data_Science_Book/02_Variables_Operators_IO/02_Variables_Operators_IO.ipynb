## 2.1 Variables

A variable is a name that refers to a value. Python is dynamically typed.

```python
company = "UrbanCart"
revenue = 12500000
growth_rate = 0.18
is_profitable = True

print(company)
print(type(company))
print(type(revenue))
print(type(growth_rate))
print(type(is_profitable))
```

### Naming Rules & Best Practices
- Use `snake_case` for variables and functions
- Names should be descriptive (`monthly_revenue` not `mr`)
- Cannot start with a number
- Avoid Python keywords (`if`, `for`, `class`, etc.)

```python
# Good names
customer_count = 15420
average_order_value = 2450.75
region_name = "West"

# Multiple assignment
x, y, z = 10, 20, 30
print(x, y, z)

# Swapping values (Pythonic)
a, b = 5, 9
a, b = b, a
print(a, b)
```

## 2.2 Operators

### Arithmetic Operators

```python
print(15 + 4)    # Addition
print(15 - 4)    # Subtraction
print(15 * 4)    # Multiplication
print(15 / 4)    # Division (always float)
print(15 // 4)   # Floor division
print(15 % 4)    # Modulus (remainder)
print(15 ** 4)   # Exponentiation

# Business example: compound growth
principal = 100000
rate = 0.12
years = 5
future_value = principal * (1 + rate) ** years
print(f"Future value after {years} years: ₹{future_value:,.0f}")
```

### Comparison & Logical Operators

```python
sales = 1250000
target = 1200000

print(sales > target)          # True
print(sales >= target)
print(sales == target)
print(sales != target)

# Logical operators
high_sales = sales > 1000000
met_target = sales >= target
print(high_sales and met_target)
print(high_sales or met_target)
print(not met_target)
```

### Assignment Operators

```python
total = 100
total += 50      # total = total + 50
total *= 2       # total = total * 2
print(total)
```

## 2.3 Input & Output

`print()` is your main output tool. `input()` reads from the user (always returns string).

```python
print("Sales Report", "Q3 2025", sep=" | ", end="\n\n")
print(f"Revenue: ₹{1250000:,}")
print("Growth: {:.1%}".format(0.185))
```
