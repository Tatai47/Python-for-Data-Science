## 13.1 Advanced Comprehensions

```python
matrix = [[1,2,3],[4,5,6],[7,8,9]]
flat = [num for row in matrix for num in row]
print(flat)

pairs = [(x, y) for x in range(3) for y in range(3) if x != y]
print(pairs)
```

## 13.2 Generators

```python
def sales_generator(n):
    for i in range(n):
        yield i * 10000

for s in sales_generator(5):
    print(s)
```

## 13.3 Decorators (Introduction)

```python
def log_calls(func):
    def wrapper(*args, **kwargs):
        print(f"Calling {func.__name__}")
        return func(*args, **kwargs)
    return wrapper

@log_calls
def add(a, b):
    return a + b

print(add(3, 5))
```
