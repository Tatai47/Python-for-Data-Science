## 9.1 Creating DataFrames

```python
import pandas as pd
import numpy as np

# From dictionary
data = {
    "Region": ["North", "South", "East", "West", "North", "South"],
    "Month": ["Jan", "Jan", "Jan", "Jan", "Feb", "Feb"],
    "Sales": [1250000, 980000, 1120000, 1450000, 1320000, 1050000],
    "Target": [1200000, 1000000, 1100000, 1400000, 1250000, 1000000]
}

df = pd.DataFrame(data)
df
```

```python
# Quick exploration
print(df.shape)
print(df.columns.tolist())
print(df.dtypes)
df.head()
```

```python
df.info()
df.describe()
```

---

## 9.2 Selecting & Filtering Data

```python
# Column selection
df["Sales"]
df[["Region", "Sales"]]
```

```python
# Filtering rows (Boolean indexing)
df[df["Sales"] > 1200000]
```

```python
# Multiple conditions
df[(df["Sales"] > 1100000) & (df["Region"] == "West")]
```

```python
# Using .loc and .iloc
df.loc[0:2, ["Region", "Sales"]]     # Label based
df.iloc[0:3, 0:2]                    # Position based
```

---

## 9.3 Creating New Columns (Feature Engineering)

```python
df["Achievement"] = (df["Sales"] / df["Target"]) * 100
df["Variance"] = df["Sales"] - df["Target"]
df["Status"] = np.where(df["Achievement"] >= 100, "Met", "Missed")
df
```

---

## 9.4 Sorting & Ranking

```python
df.sort_values("Sales", ascending=False)
```

```python
df.sort_values(["Region", "Sales"], ascending=[True, False])
```

---

## 9.5 Aggregation & GroupBy (The Heart of Analysis)

```python
# Total sales by region
df.groupby("Region")["Sales"].sum()
```

```python
# Multiple aggregations
df.groupby("Region").agg({
    "Sales": ["sum", "mean", "max"],
    "Target": "sum",
    "Achievement": "mean"
}).round(1)
```

```python
# Pivot table style
pd.pivot_table(df, values="Sales", index="Region", columns="Month", aggfunc="sum", fill_value=0)
```

---

## 9.6 Handling Missing Data

```python
# Create sample with missing values
df2 = df.copy()
df2.loc[1, "Sales"] = np.nan
df2.loc[3, "Target"] = np.nan
df2
```

```python
print(df2.isna().sum())

# Drop rows with any NaN
df2.dropna()

# Fill with mean / median / specific value
df2["Sales"].fillna(df2["Sales"].mean(), inplace=False)
```

## 9.8 Merging DataFrames (Joins)

```python
region_info = pd.DataFrame({
    "Region": ["North", "South", "East", "West"],
    "Manager": ["Anita", "Ravi", "Priya", "Vikram"],
    "Zone": ["A", "B", "A", "B"]
})

merged = pd.merge(df, region_info, on="Region", how="left")
merged
```
