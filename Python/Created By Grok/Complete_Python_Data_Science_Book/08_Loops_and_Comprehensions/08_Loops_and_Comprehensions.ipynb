## 8.1 for Loops

```python
regions = ["North", "South", "East", "West"]
for region in regions:
    print(f"Processing {region} region")

for i, region in enumerate(regions, start=1):
    print(i, region)
```

```python
sales = [125, 98, 112, 145]
total = 0
for s in sales:
    total += s
print("Total:", total)

# range()
for i in range(1, 6):
    print(i, end=" ")
print()
```

## 8.2 while Loops

```python
count = 0
while count < 5:
    print(count)
    count += 1
```

## 8.3 break, continue, else

```python
for s in [80, 95, 110, 70, 130]:
    if s < 100:
        continue
    if s > 120:
        print("Exceptional:", s)
        break
    print("Good:", s)
else:
    print("Loop completed without break")
```

## 8.4 Comprehensions (Lists, Dicts, Sets)

```python
sales = [85, 120, 95, 140, 110]
high = [s for s in sales if s >= 100]
squared = {s: s**2 for s in sales}
unique_high = {s for s in sales if s >= 100}
print(high)
print(squared)
print(unique_high)
```
