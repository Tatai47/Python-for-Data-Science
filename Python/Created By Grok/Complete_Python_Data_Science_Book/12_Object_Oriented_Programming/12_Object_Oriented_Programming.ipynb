## 12.1 Classes and Objects

```python
class Employee:
    def __init__(self, name, department, salary):
        self.name = name
        self.department = department
        self.salary = salary

    def give_raise(self, percent):
        self.salary *= (1 + percent)

    def info(self):
        return f"{self.name} | {self.department} | ₹{self.salary:,.0f}"

emp = Employee("Priya Sharma", "Analytics", 1250000)
print(emp.info())
emp.give_raise(0.10)
print(emp.info())
```

## 12.2 Inheritance

```python
class Manager(Employee):
    def __init__(self, name, department, salary, team_size):
        super().__init__(name, department, salary)
        self.team_size = team_size

    def info(self):
        base = super().info()
        return f"{base} | Team: {self.team_size}"

mgr = Manager("Rahul Verma", "Sales", 1800000, 8)
print(mgr.info())
```
