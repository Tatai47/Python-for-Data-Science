## 17.1 When to Use Deep Learning vs Gradient Boosting

| Data Type | Preferred Approach |
|-----------|--------------------|
| Tabular (sales, CRM, transactions) | XGBoost / LightGBM / CatBoost |
| Images, video | CNN / Vision Transformers (PyTorch / FastAI) |
| Text / documents | Transformers (Hugging Face) |
| Time series (complex patterns) | LSTM / Temporal Fusion Transformer / N-BEATS |
| Multimodal | AutoGluon Multimodal or custom PyTorch |

## 17.4 PyTorch Lightning – Cleaner Engineering

Lightning separates research code from engineering boilerplate (logging, multi-GPU, checkpointing).

````python
lightning_example = '''
import pytorch_lightning as pl

class LitModel(pl.LightningModule):
    def __init__(self, n_features):
        super().__init__()
        self.model = SimpleNet(n_features)   # reuse the PyTorch module
        self.criterion = nn.BCELoss()

    def training_step(self, batch, batch_idx):
        x, y = batch
        pred = self.model(x)
        loss = self.criterion(pred, y)
        self.log("train_loss", loss)
        return loss

    def configure_optimizers(self):
        return torch.optim.Adam(self.parameters(), lr=1e-3)

trainer = pl.Trainer(max_epochs=30, accelerator="auto")
trainer.fit(LitModel(n_features), train_loader)
'''
print("PyTorch Lightning template:")
print(lightning_example)
````

## 17.6 JAX – High-Performance NumPy + Autodiff

JAX is used heavily in research and at Google. It compiles NumPy-like code to run on GPU/TPU and provides automatic differentiation.

```python
import jax
import jax.numpy as jnp

def loss_fn(params, x, y):
    pred = model(params, x)
    return jnp.mean((pred - y)**2)

grad_fn = jax.grad(loss_fn)
```

For most business analysts, start with Keras or FastAI. Move to pure PyTorch or JAX when you need maximum control or custom research.

## Key Takeaways
- Deep learning is powerful for **unstructured** data (images, text, audio).
- For classic tabular business data, gradient boosting is usually faster to develop and equally or more accurate.
- **Keras** and **FastAI** give the quickest path to a working neural network.
