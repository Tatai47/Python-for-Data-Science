## 9.1 Defining and Calling Functions

```python
def calculate_profit(revenue, cost):
    """Return profit and margin."""
    profit = revenue - cost
    margin = (profit / revenue) * 100 if revenue else 0
    return profit, margin

p, m = calculate_profit(1500000, 1100000)
print(f"Profit: ₹{p:,.0f} | Margin: {m:.1f}%")
```

## 9.2 Default Arguments, *args, **kwargs

```python
def greet(name, title="Analyst"):
    return f"Hello {title} {name}"

print(greet("Priya"))
print(greet("Rahul", title="Manager"))

def total_sales(*amounts):
    return sum(amounts)

print(total_sales(100, 200, 300))

def create_report(**kwargs):
    for k, v in kwargs.items():
        print(f"{k}: {v}")

create_report(region="North", sales=1250000, target=1200000)
```

## 9.3 Lambda, map, filter

```python
sales = [85, 120, 95, 140]
squared = list(map(lambda x: x**2, sales))
high = list(filter(lambda x: x >= 100, sales))
print(squared)
print(high)
```
