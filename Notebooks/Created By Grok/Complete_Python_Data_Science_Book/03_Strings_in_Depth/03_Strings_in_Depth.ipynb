## 3.1 Creating & Indexing Strings

```python
name = "Priya Sharma"
feedback = 'Great product, fast delivery!'
multi = """This is a
multi-line
string."""

print(name[0])          # First character
print(name[-1])         # Last character
print(name[0:5])        # Slicing
print(name[6:])         # From index 6 to end
print(name[::-1])       # Reverse
```

## 3.2 Essential String Methods

```python
text = "  UrbanCart Sales Report - Q3 2025  "

print(text.strip())                 # Remove leading/trailing spaces
print(text.lower())
print(text.upper())
print(text.title())
print(text.replace("Q3", "Q4"))
print(text.count("a"))
print("Sales" in text)
print(text.find("Report"))
```

```python
# Splitting and joining – extremely useful
regions = "North,South,East,West"
region_list = regions.split(",")
print(region_list)

print(" | ".join(region_list))

# Cleaning messy customer names
raw_name = "  pRiYa   sHaRmA  "
clean = " ".join(raw_name.strip().title().split())
print(clean)
```

## 3.3 f-strings & Formatting (Modern Way)

```python
product = "Laptop"
units = 145
price = 62500.5

print(f"Product: {product}")
print(f"Units sold: {units:,}")
print(f"Revenue: ₹{units * price:,.2f}")
print(f"Growth: {0.157:.1%}")
```
