## 1.4 Your First Business Calculation

Let’s calculate profit and profit margin — something every MBA does.

```python
revenue = 2_450_000
cost = 1_875_000
profit = revenue - cost
margin = (profit / revenue) * 100

print(f"Revenue     : ₹{revenue:,}")
print(f"Cost        : ₹{cost:,}")
print(f"Profit      : ₹{profit:,}")
print(f"Profit Margin: {margin:.2f}%")
```
