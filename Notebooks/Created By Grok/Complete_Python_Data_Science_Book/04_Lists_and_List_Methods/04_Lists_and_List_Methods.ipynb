## 4.1 Creating & Accessing Lists

```python
sales = [125000, 98000, 112000, 145000]
regions = ["North", "South", "East", "West"]
mixed = ["Laptop", 45, 62500.0, True]

print(sales[0])
print(sales[-1])
print(sales[1:3])
print(len(sales))
print(sum(sales))
print(max(sales), min(sales))
```

## 4.2 Modifying Lists

```python
products = ["Laptop", "Mouse", "Keyboard"]
products.append("Monitor")           # Add at end
products.insert(1, "Headset")        # Insert at position
products.extend(["Webcam", "Stand"]) # Add multiple
print(products)

products.remove("Mouse")
last = products.pop()                # Remove & return last
print(products)
print("Removed:", last)
```

## 4.3 Important List Methods & Operations

```python
sales = [125, 98, 112, 145, 98, 132]

print(sales.count(98))
print(sales.index(145))
sales.sort()                         # In-place sort
print(sales)
sales.sort(reverse=True)
print(sales)

print(sorted(sales))                 # Returns new list
print(sales[::-1])                   # Reverse copy
```

## 4.4 List Comprehensions (Very Important)

```python
sales = [125000, 98000, 112000, 145000, 87000]

# Traditional way
high = []
for s in sales:
    if s > 100000:
        high.append(s)
print(high)

# List comprehension – preferred
high = [s for s in sales if s > 100000]
print(high)

# Transform
in_lakhs = [round(s/100000, 2) for s in sales]
print(in_lakhs)
```
