# Chapter 6: Dictionaries

**Goal**: Dictionaries store key-value pairs — perfect for records, configurations, and frequency counts.

---

## 6.1 Creating and Accessing

```python
employee = {
    "name": "Priya Sharma",
    "department": "Analytics",
    "salary": 1250000,
    "skills": ["Python", "SQL", "Tableau"]
}

print(employee["name"])
print(employee.get("location", "Not specified"))
print(employee.keys())
print(employee.values())
print(employee.items())
```

## 6.2 Modifying Dictionaries

```python
employee["location"] = "Mumbai"
employee["salary"] += 100000
employee.update({"role": "Senior Analyst", "active": True})
print(employee)

removed = employee.pop("active")
print("Removed:", removed)
```

## 6.3 Dictionary Comprehensions & Nesting

```python
sales = {"North": 125, "South": 98, "East": 112, "West": 145}
in_thousands = {k: v*1000 for k, v in sales.items()}
print(in_thousands)

# Nested dictionary – common for business data
regions = {
    "North": {"sales": 1250000, "target": 1200000},
    "South": {"sales": 980000, "target": 1000000}
}
print(regions["North"]["sales"])
```
