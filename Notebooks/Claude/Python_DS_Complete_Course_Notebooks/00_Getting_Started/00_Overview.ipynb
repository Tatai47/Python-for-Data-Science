## How to Actually Use This Course (Study Plan)

Don't just read — **run every code block**. The single biggest reason people finish these books and forget everything in a month is that they read code instead of typing/running it. A realistic pace for someone with your background (MBA, some dev experience, freelance work already in analytics and full-stack):

| Phase | Weeks | Volumes | Outcome |
|---|---|---|---|
| 1. Python Core | 1–2 | Vol 01 | You can write clean scripts, functions, classes |
| 2. Data Wrangling | 2–3 | Vol 02, 03 | You can load, clean, reshape, and visualize any tabular dataset |
| 3. Statistics + Classic ML | 2 | Vol 04, 05 | You can run a regression/classification project end-to-end and defend it statistically |
| 4. Competitive ML | 1–2 | Vol 06, 07 | You can win a Kaggle-style tabular competition and automate model selection |

Given your MBA + analytics + dev background, you can likely compress this to **6–8 weeks** of focused evening/weekend study, since Volumes 01–02 will move fast for you.

### Environment Setup

You only need **one** environment. Recommended: a fresh virtual environment plus VS Code (which you already use) or Jupyter.

```bash
# Create and activate an isolated environment
python -m venv dsvenv
source dsvenv/bin/activate        # Windows: dsvenv\Scripts\activate

# Core stack (install first — everything in Volumes 1–5 needs only this)
pip install numpy pandas polars matplotlib seaborn plotly scipy statsmodels scikit-learn jupyter openpyxl

# Gradient boosting & tuning (Volume 6)
pip install xgboost lightgbm catboost optuna

# AutoML toolkit (Volume 7) — install only the ones you actually try, these are heavy
pip install pycaret flaml
# h2o, auto-sklearn, autogluon each have special install requirements — see Volume 7

# Deep learning (Volume 8) — pick ONE framework to start, add others later
pip install tensorflow            # OR: pip install torch torchvision
pip install pytorch-lightning fastai

# NLP / GenAI (Volume 9)
pip install spacy transformers sentence-transformers langchain langchain-community llama-index chromadb
python -m spacy download en_core_web_sm

# Databases & apps (Volume 10)
pip install mysql-connector-python pymongo streamlit pydantic
```

**Practical advice:** don't install everything on day one. Install the Volume 1–5 stack now; add the rest chapter by chapter as you reach it. Several of these libraries (TensorFlow, PyTorch, H2O, RAPIDS) are multi-GB and version-sensitive against your Python version and OS/GPU — Volume 7's and Volume 8's chapters each start with a "what you need before this chapter" box.

### A Note on the Business Framing

All three of your source books were explicitly written "for MBA students and business professionals." This course keeps that spirit: every technical concept is anchored to something you'd actually do as a business analyst — sales, churn, revenue, customer segments, forecasting — rather than abstract toy examples. Where a concept has a pure-CS use case (e.g., generators, decorators) that rarely comes up in analytics work, this course tells you plainly that it's a "know it exists, revisit when needed" topic rather than pretending it's a daily tool.
