# Chapter 4: H2O AutoML — Enterprise-Scale, Distributed AutoML

### Why This Matters

**H2O** is designed from the ground up for distributed computing — running across multiple machines or many cores on larger datasets than comfortably fit the single-machine tools covered so far — and is popular specifically in larger enterprise data science teams for that reason, alongside a genuinely useful web-based UI (Flow) for less code-comfortable stakeholders to inspect results.

```python
import h2o
from h2o.automl import H2OAutoML

h2o.init()                                            # starts a local (or connects to a remote) H2O cluster
train = h2o.import_file("train.csv")
test = h2o.import_file("test.csv")

aml = H2OAutoML(max_runtime_secs=300, seed=42, sort_metric="AUC")
aml.train(y="churn", training_frame=train)

print(aml.leaderboard)                                 # every model H2O tried, ranked
best_model = aml.leader
preds = best_model.predict(test)
```

The workflow shape is familiar by now (initialize → train with a time budget → inspect a leaderboard → predict), but note `h2o.init()` — this actually starts a background Java-based compute cluster (H2O's engine runs on the JVM, not pure Python, which is part of what enables its distributed scaling), a meaningfully heavier setup step than any other tool in this volume. `aml.leaderboard` returns every model H2O tried (often including automatically-built **stacked ensembles**, similar in spirit to AutoGluon) ranked by your chosen metric, letting you inspect the full space of what was tried, not just the single winner.

💡 **When H2O is the right choice specifically:** you're working with a genuinely large dataset (multiple millions of rows) where single-machine tools become slow or memory-constrained, or you're on a team that already has H2O infrastructure and conventions in place. For the dataset sizes a typical business analyst works with day to day, PyCaret, FLAML, or AutoGluon will usually be a lighter-weight, equally effective choice.

### Cheat Sheet — H2O AutoML

| Task | Code |
|---|---|
| Start engine | `h2o.init()` |
| Load data | `h2o.import_file(path)` |
| Run AutoML | `H2OAutoML(max_runtime_secs=300).train(y="target", training_frame=train)` |
| Leaderboard | `aml.leaderboard` |
| Best model | `aml.leader` |
