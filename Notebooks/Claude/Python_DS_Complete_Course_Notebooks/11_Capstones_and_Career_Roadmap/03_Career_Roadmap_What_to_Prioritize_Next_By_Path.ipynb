### Project Ideas to Build a Portfolio

Rather than only completing this course's own exercises, build 2–3 complete, original projects end to end — this is what actually demonstrates capability to an employer or client, far more than exercise completion:

- **Data analysis:** an interactive dashboard (Volume 10's Streamlit) analyzing a public dataset relevant to a domain you care about
- **Classical ML:** a churn, fraud-detection, or price-prediction model on a real (e.g., Kaggle) dataset, taken all the way through Volume 06's gradient boosting and Volume 07's AutoML comparison
- **Deep learning:** an image classifier using FastAI's transfer learning (Volume 08) on a domain-specific image set
- **GenAI:** a RAG chatbot (Volume 09) over a real set of documents — your own notes, a company's public policies, or a textbook

### Staying Current

This field moves quickly — several libraries in this course (Polars, LangChain, RAPIDS) evolve their APIs meaningfully between versions. Three durable habits matter more than memorizing any specific library's current syntax:

- **Read the official documentation directly** rather than relying solely on tutorials, which go stale faster than reference docs
- **Practice on real, messy data** (Kaggle competitions, your own work data where appropriate) rather than only clean textbook datasets
- **Build a public portfolio** (GitHub, a blog, Kaggle notebooks) — this compounds over time and is frequently what actually gets noticed in hiring, far more than any certificate
