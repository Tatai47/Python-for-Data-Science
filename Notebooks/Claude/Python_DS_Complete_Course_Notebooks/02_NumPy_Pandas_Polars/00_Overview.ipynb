## How These Three Libraries Relate

Before diving in, it's worth seeing the big picture, because it makes the rest of this volume click into place:

- **NumPy** gives you the `ndarray` — a fast, fixed-type, multi-dimensional array. It's the computational bedrock: nearly every numeric library in Python (Pandas, scikit-learn, TensorFlow, PyTorch) either is built directly on NumPy arrays or borrows its indexing/broadcasting philosophy.
- **Pandas** builds on top of NumPy to give you the `DataFrame` — a labeled, spreadsheet-like table with named columns and a flexible index, plus hundreds of methods for exactly the kind of messy, heterogeneous, real-world business data NumPy alone handles awkwardly (NumPy arrays want one uniform data type; a business dataset has text, dates, categories, and numbers side by side).
- **Polars** is a newer, independent DataFrame library (not built on Pandas or NumPy) written in Rust, offering a similar but stricter API with dramatically better performance on large datasets through parallelism and *lazy* (query-planned) execution.

**Practical guidance for you:** learn Pandas thoroughly first — it's still the default in almost every job posting, every Stack Overflow answer, and every other library's documentation examples. Learn Polars once you've hit a genuine performance wall (a dataset that's slow or memory-heavy in Pandas) — it's a "reach for this when you need it" tool, not a full replacement for daily use, unless a team you join has already standardized on it.
