## The Single Most Important Decision in This Volume: Do You Even Need Deep Learning?

Volume 06 already told you that gradient boosting usually wins on tabular business data. This volume opens by reinforcing, not contradicting, that lesson — deep learning's real strength is **unstructured data**: images, audio, free text, and complex sequential patterns, where the "features" aren't already neatly organized into meaningful columns the way a sales spreadsheet is.

| Data Type | Preferred Approach |
|---|---|
| Tabular (sales, CRM, transactions) | XGBoost / LightGBM / CatBoost (Volume 06) |
| Images, video | CNNs / Vision Transformers (PyTorch, FastAI) |
| Text / documents | Transformers (Hugging Face — Volume 09) |
| Complex time series | LSTM / specialized forecasting architectures |
| Multiple data types combined | AutoGluon Multimodal (Volume 07) or custom PyTorch |

**Before writing a single line of neural network code for a tabular business problem, always train a LightGBM/XGBoost baseline first (Volume 06) and require the neural network to beat it by a meaningful margin to justify its extra complexity, training time, and reduced interpretability.** This isn't a hedge — it reflects genuine, repeated experience across the data science industry with tabular business data specifically.

### The One Concept Underlying Every Framework in This Volume

A neural network is, structurally, a sequence of layers, each performing a linear transformation (multiply inputs by learned weights, add a learned bias) followed by a **non-linear activation function** (like ReLU or sigmoid). Stacking many such layers lets the network approximate very complex functions. Training means: run data through the network (**forward pass**), measure how wrong the output was (the **loss**), then use **backpropagation** (an efficient application of calculus's chain rule) to compute how much each individual weight contributed to that error, and nudge every weight slightly in the direction that reduces it (**gradient descent**). Every framework below — TensorFlow, PyTorch, JAX — is fundamentally a tool for defining that layer structure and automating exactly this gradient computation (called **automatic differentiation**, or "autodiff") so you never have to derive calculus by hand.
