## Why This Volume Matters More Than It Looks

A perfectly correct analysis that's presented as a wall of numbers gets ignored. The same analysis as a clean chart with an obvious takeaway gets acted on. As an MBA-trained analyst, this is arguably where your dual background pays off most directly — you understand *what a decision-maker needs to see* in five seconds, and this volume gives you the technical tools to build exactly that.

**The three libraries, and when to reach for each:**

- **Matplotlib** — the foundational plotting library everything else in the Python visualization world is built on top of. Lower-level, more verbose, but gives you complete control over every pixel. You'll use it directly for custom, publication-quality static charts, and indirectly every time you use Seaborn (which is built on it).
- **Seaborn** — a higher-level library built on Matplotlib, specialized for *statistical* graphics (distributions, relationships between variables, category comparisons) with far less code and better default aesthetics.
- **Plotly** — builds *interactive*, browser-based charts (hover tooltips, zoom, pan) rather than static images — ideal for dashboards and exploratory work where the viewer benefits from poking at the data themselves.

**Practical workflow:** most analysts default to Seaborn for quick statistical exploration, drop to raw Matplotlib when they need a specific custom layout or annotation Seaborn doesn't expose, and reach for Plotly specifically when the output is a dashboard or something a stakeholder will interact with directly (e.g., in Streamlit — see Volume 10).
