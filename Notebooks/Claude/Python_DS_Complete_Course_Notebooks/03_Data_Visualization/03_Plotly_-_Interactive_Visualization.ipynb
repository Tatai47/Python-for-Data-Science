# Chapter 3: Plotly — Interactive Visualization

### Why This Matters

Matplotlib and Seaborn produce *static images* — perfect for a PDF report or a slide deck, but the viewer can't hover for exact values, zoom into a busy time period, or toggle a series on and off. **Plotly** builds charts as interactive, browser-rendered objects, which is exactly what a web dashboard (Volume 10's Streamlit chapter) or a shared exploratory notebook benefits from — letting the *viewer* explore the data on their own terms rather than you pre-deciding every view they'll need.

### 3.1 Plotly Express — The Fast, High-Level API

```python
import plotly.express as px

df = pd.DataFrame({
    'ad_spend': [10, 20, 30, 40, 50],
    'revenue':  [50, 72, 90, 135, 150],
    'channel':  ['Search', 'Search', 'Social', 'Social', 'Email']
})

fig = px.scatter(df, x='ad_spend', y='revenue', color='channel', size='revenue',
                  hover_data=['channel'], title='Ad Spend vs Revenue')
fig.show()
```

`plotly.express` (conventionally imported as `px`) mirrors Seaborn's philosophy — point it at a DataFrame and column names, and it handles the rest. The result, though, is a fully interactive figure: hovering over any point shows its exact values (further customized by `hover_data`), and the legend (built automatically from `color='channel'`) is clickable to show/hide series. `size='revenue'` maps a third numeric variable to marker size (a "bubble chart"), letting you encode three dimensions of data — x-position, y-position, and size — in a single 2D chart.

```python
# 3D scatter — genuinely useful for exploring three numeric business dimensions at once
fig = px.scatter_3d(df, x='ad_spend', y='revenue', z=df['revenue']/df['ad_spend'], color='channel')
fig.show()
```

⚠️ **Professional caution on 3D charts:** they look impressive but are, in most real analysis, *harder to read accurately* than two well-chosen 2D charts, because humans are poor at judging depth and relative position on a 2D screen. Reach for `scatter_3d` mainly for exploratory poking-around in a notebook (where you can freely rotate the interactive view) rather than for a static report or a chart a stakeholder will view once and need to interpret correctly without you in the room.

### 3.2 When to Choose Plotly Over Matplotlib/Seaborn

| Situation | Best choice |
|---|---|
| A chart for a PDF report or PowerPoint slide | Matplotlib / Seaborn (static, prints reliably) |
| A chart embedded in a web dashboard (Streamlit, etc.) | Plotly (interactive, renders in-browser) |
| Quick statistical exploration in a notebook | Seaborn |
| A stakeholder needs to drill into specific data points themselves | Plotly |
| Precise, publication-grade custom layout | Matplotlib (most granular control) |

### Cheat Sheet — Plotly

| Task | Code |
|---|---|
| Scatter (with color/size/hover) | `px.scatter(df, x, y, color=, size=, hover_data=[...])` |
| Line chart | `px.line(df, x, y, color=)` |
| Bar chart | `px.bar(df, x, y, color=)` |
| 3D scatter | `px.scatter_3d(df, x, y, z, color=)` |
| Facets (small multiples) | `px.scatter(df, x, y, facet_col="category")` |
| Show the figure | `fig.show()` |

### 🎯 Practice — Plotly

```python
import plotly.express as px

regional = pd.DataFrame({
    "region": ["North","South","East","West"]*4,
    "month": sorted(["Jan","Feb","Mar","Apr"]*4),
    "sales": np.random.randint(80, 200, 16)
})

fig = px.line(regional, x="month", y="sales", color="region", markers=True,
              title="Regional Sales Trend (Interactive)")
fig.show()

fig2 = px.bar(regional, x="region", y="sales", color="month", barmode="group",
              title="Sales by Region and Month")
fig2.show()
```
