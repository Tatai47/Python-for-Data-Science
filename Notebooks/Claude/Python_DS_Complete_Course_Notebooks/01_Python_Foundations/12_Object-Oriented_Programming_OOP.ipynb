# Chapter 12: Object-Oriented Programming (OOP)

### Why This Matters

Everything in Python — strings, lists, DataFrames, ML models — is an **object**: a bundle of data (**attributes**) plus behavior (**methods**) that act on that data. You've been *using* objects since Chapter 1 (`"text".upper()` is calling a method on a string object). This chapter teaches you to *create your own* — the natural way to model real business entities (an Employee, a Product, a Customer) that have both properties and behaviors.

### 12.1 Classes and Objects

```python
class Employee:
    def __init__(self, name, department, salary):    # the CONSTRUCTOR — runs when you create a new Employee
        self.name = name              # self.X stores X as an attribute ON THIS SPECIFIC object
        self.department = department
        self.salary = salary

    def give_raise(self, percent):
        self.salary *= (1 + percent)

    def info(self):
        return f"{self.name} | {self.department} | ₹{self.salary:,.0f}"

emp = Employee("Priya Sharma", "Analytics", 1250000)     # creates an INSTANCE of the Employee class
print(emp.info())
emp.give_raise(0.10)
print(emp.info())
```

A **class** is a blueprint; an **object** (or **instance**) is a specific thing built from that blueprint. `Employee` the class describes "any employee has a name, department, and salary, and can receive a raise." `emp` is one *particular* employee built from that blueprint. You can create as many independent `Employee` objects as you like from the same class, each with its own separate `name`/`department`/`salary`.

**`self`** is simply "the specific object this method is being called on" — every method you define inside a class takes `self` as its first parameter (Python passes it automatically; you never type it yourself when *calling* the method — `emp.give_raise(0.10)`, not `emp.give_raise(emp, 0.10)`). If this feels abstract, an analogy: `self` plays the same role as "this" or "myself" would in a sentence — `give_raise` is an instruction that says "give **myself** a raise," and `self` is how the method knows which specific employee "myself" refers to.

`__init__` is a **dunder method** (short for "double underscore") — a special method Python calls automatically at a specific moment, here, the moment you create a new object with `Employee(...)`. You'll meet several more dunder methods as you go deeper (`__str__`, `__len__`, `__eq__`), but `__init__` is the one you'll write in nearly every class you ever create.

### 12.2 Inheritance

```python
class Manager(Employee):        # Manager INHERITS everything Employee has, then extends it
    def __init__(self, name, department, salary, team_size):
        super().__init__(name, department, salary)     # call the PARENT's __init__ to set up the shared fields
        self.team_size = team_size

    def info(self):                                      # OVERRIDING the parent's info() method
        base = super().info()                             # but still reusing the parent's version inside it
        return f"{base} | Team: {self.team_size}"

mgr = Manager("Rahul Verma", "Sales", 1800000, 8)
print(mgr.info())     # Rahul Verma | Sales | ₹18,00,000 | Team: 8
print(mgr.give_raise(0.05))    # inherited from Employee, unchanged — Manager didn't need to redefine it
```

**Inheritance** models an "is-a" relationship: a Manager *is an* Employee, plus something extra (a team). Rather than copy-pasting all of `Employee`'s code into a separate `Manager` class, `Manager(Employee)` reuses it directly. `super()` is how a subclass reaches back into its parent's version of a method — critical when you want to *extend* behavior (as `info()` does here, adding the team size on top of the base info) rather than replace it entirely.

💡 This exact pattern — a base class with shared behavior, and subclasses that extend it — is precisely how scikit-learn's model classes are structured underneath (every classifier "is an" `Estimator`), and how Pydantic models (Volume 10) and PyTorch neural network layers (Volume 08) work. Understanding inheritance now means several "why does this library work this way" moments later will resolve instantly instead of feeling mysterious.

### ⚠️ Common Mistakes

- Forgetting `self` as the first parameter of an instance method — Python will raise a confusing `TypeError` about argument counts.
- Forgetting to call `super().__init__(...)` in a subclass's constructor, leaving the inherited attributes never actually set.
- Confusing a **class** (the blueprint, e.g. `Employee`) with an **instance** (a specific object, e.g. `emp`) — attributes set with `self.x = ...` belong to the instance, not the class itself.

### 🎯 Business Challenge — Chapter 12

```python
class Product:
    def __init__(self, name, price, stock):
        self.name = name
        self.price = price
        self.stock = stock

    def apply_discount(self, percent):
        self.price *= (1 - percent)

    def sell(self, units):
        if units > self.stock:
            print(f"Cannot sell {units} units — only {self.stock} in stock.")
            return
        self.stock -= units
        print(f"Sold {units} units of {self.name}. Remaining stock: {self.stock}")

laptop = Product("Laptop", 62500.0, 40)
mouse = Product("Wireless Mouse", 799.0, 200)

laptop.apply_discount(0.10)
laptop.sell(5)
mouse.sell(250)     # deliberately more than stock, to demonstrate the guard clause
```
