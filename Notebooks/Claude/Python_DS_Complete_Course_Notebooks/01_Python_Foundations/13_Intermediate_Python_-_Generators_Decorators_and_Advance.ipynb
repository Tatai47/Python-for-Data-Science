### 13.1 Advanced Comprehensions

```python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flat = [num for row in matrix for num in row]         # nested loops inside a comprehension, left-to-right order
print(flat)     # [1, 2, 3, 4, 5, 6, 7, 8, 9]

pairs = [(x, y) for x in range(3) for y in range(3) if x != y]
print(pairs)
```

Read a nested comprehension the same way you'd nest ordinary `for` loops, just written on one line, outer loop first: `for row in matrix` is the outer loop, `for num in row` is the inner loop, and `num` (the innermost expression) is what ends up in the resulting list.

### 13.2 Generators — Lazy, Memory-Efficient Sequences

```python
def sales_generator(n):
    for i in range(n):
        yield i * 10000        # yield PAUSES the function and hands back one value, instead of returning and exiting

for s in sales_generator(5):
    print(s)
```

A regular function with `return` computes its entire result and hands it all back at once. A **generator function** (any function containing `yield`) instead produces values **one at a time, on demand**, pausing its execution between each one. The practical benefit: if you needed a sequence of 100 million values, a list comprehension would try to build and hold all 100 million in memory simultaneously; a generator produces them one at a time, using almost no memory, no matter how large the sequence conceptually is. This matters directly once you start working with large datasets that don't comfortably fit in RAM (Volume 02's discussion of chunked CSV reading, and Volume 08's data loaders for deep learning, both lean on exactly this idea).

### 13.3 Decorators (Introduction)

```python
def log_calls(func):                 # a decorator: a function that takes a function and returns a new function
    def wrapper(*args, **kwargs):    # remember *args/**kwargs from Chapter 9 — this is exactly why they exist
        print(f"Calling {func.__name__}")
        return func(*args, **kwargs)
    return wrapper

@log_calls          # this line is shorthand for: add = log_calls(add)
def add(a, b):
    return a + b

print(add(3, 5))    # prints "Calling add" then 8
```

A **decorator** wraps a function with extra behavior — here, printing a log message — without modifying the original function's own code at all. The `@log_calls` syntax right above `def add(...)` is pure syntactic sugar for `add = log_calls(add)`; both do exactly the same thing, but the `@` form is dramatically more readable once you're used to it. 💡 You'll meet decorators constantly once you reach Volume 07's AutoML and Volume 08's deep-learning chapters — `@st.cache_data` in Streamlit, `@app.route` in web frameworks, and various training-loop decorators in PyTorch Lightning all follow this exact pattern.

### 🎯 Business Challenge — Chapter 13

```python
def fibonacci(n):
    a, b = 0, 1
    for _ in range(n):
        yield a
        a, b = b, a + b

first_15 = list(fibonacci(15))
even_fibs = [f for f in first_15 if f % 2 == 0]

print("First 15 Fibonacci numbers:", first_15)
print("Even ones:", even_fibs)
```
