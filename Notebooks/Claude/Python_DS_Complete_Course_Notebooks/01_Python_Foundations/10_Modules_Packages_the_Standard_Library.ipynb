# Chapter 10: Modules, Packages & the Standard Library

### Why This Matters

You will never write everything from scratch. Python ships with a **standard library** — dozens of modules covering dates, math, file paths, randomness, and more — installed with Python itself, no `pip install` required. Knowing what's already there prevents you from reinventing (usually worse) versions of well-tested tools.

### 10.1 Importing Modules

```python
import math
import statistics
from datetime import datetime, timedelta
from collections import Counter

print(math.sqrt(144))                                    # 12.0
print(statistics.mean([10, 20, 30, 40]))                 # 25
print(datetime.now().strftime("%Y-%m-%d"))                # today's date as a string
print(Counter(["North", "South", "North", "East", "North"]))   # Counter({'North': 3, 'South': 1, 'East': 1})
```

Three import styles, and when to use each:

```python
import math                     # access everything as math.sqrt(), math.pi, etc.
from math import sqrt           # import just sqrt directly -> call it as sqrt(144)
import numpy as np              # import with an alias -> the near-universal convention for NumPy/Pandas
```

💡 `import numpy as np` and `import pandas as pd` are conventions so universal across the entire data science world (every tutorial, every Stack Overflow answer, every colleague's notebook) that deviating from them just makes your code harder for others — and future you — to read. Adopt them from day one in Volume 02.

`Counter` deserves special attention: it's a dictionary subclass built specifically for counting occurrences, and it comes with useful extras like `.most_common(n)` — genuinely one of the most useful "hidden" standard-library tools for a quick exploratory look at categorical data before you even reach Pandas' `.value_counts()` (which does the same job, more powerfully, once you have a DataFrame).

### 10.2 The Standard Library Tools You'll Actually Use as an Analyst

| Module | What it's for |
|---|---|
| `datetime` | Dates, times, durations (`timedelta`) |
| `collections` | `Counter` (frequency counts), `defaultdict` (dict with automatic default values) |
| `pathlib` | Modern, cross-platform file path handling |
| `json` | Reading/writing JSON |
| `csv` | Basic CSV reading/writing (Pandas will mostly replace this from Volume 02 on) |
| `random` | Simulations, sampling, shuffling |
| `math` / `statistics` | Mathematical functions and basic stats |

```python
from pathlib import Path
print(Path.cwd())                    # current working directory, as a proper Path object
print(Path("data") / "sales.csv")    # the / operator JOINS path segments — works on Windows AND Mac/Linux
```

💡 `pathlib`'s `/` operator for joining paths is worth adopting over manually concatenating strings with `"data" + "/" + "sales.csv"` — the latter silently breaks on Windows, where the path separator is `\`, not `/`. `pathlib` handles that difference for you invisibly.
