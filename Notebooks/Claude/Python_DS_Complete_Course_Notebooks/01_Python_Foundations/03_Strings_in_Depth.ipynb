### 3.1 Creating and Indexing Strings

```python
name = "Priya Sharma"
feedback = 'Great product, fast delivery!'          # single or double quotes both work
multi = """This is a
multi-line
string."""                                          # triple quotes span lines — handy for SQL/paragraphs

print(name[0])      # 'P'   -> indexing starts at 0, not 1
print(name[-1])      # 'a'   -> negative indices count from the end
print(name[0:5])     # 'Priya' -> slicing [start:stop), stop is EXCLUSIVE
print(name[6:])      # 'Sharma' -> omit stop to go to the end
print(name[::-1])    # 'amrahS ayirP' -> step of -1 reverses the string
```

The **slice syntax `[start:stop:step]`** is one of Python's most powerful and most reused ideas — you will see the *exact same syntax* apply to lists (Chapter 4), NumPy arrays, and Pandas Series/DataFrames later. Learn it once here, get it for free everywhere else. The critical detail people forget: `stop` is **exclusive** — `name[0:5]` gives you characters at positions 0,1,2,3,4, not position 5.

### 3.2 Essential String Methods

```python
text = "  UrbanCart Sales Report - Q3 2025  "

print(text.strip())        # removes leading/trailing whitespace only (not interior)
print(text.lower())
print(text.upper())
print(text.title())        # "Urbancart Sales Report - Q3 2025" (capitalizes each word)
print(text.replace("Q3", "Q4"))
print(text.count("a"))     # counts occurrences (case-sensitive)
print("Sales" in text)     # membership test -> True
print(text.find("Report")) # returns the starting index, or -1 if not found
```

💡 **A crucial distinction:** strings are **immutable** in Python. `text.strip()` does not modify `text` in place — it *returns a new string*. If you don't reassign it (`text = text.strip()`), the original `text` variable is unchanged. This trips up nearly every beginner at least once, and understanding it now will save you real debugging time in Volume 02, where Pandas has an analogous (and even more consequential) distinction around in-place operations.

```python
# Splitting and joining — the single most useful string pattern for parsing raw data
regions = "North,South,East,West"
region_list = regions.split(",")
print(region_list)              # ['North', 'South', 'East', 'West']
print(" | ".join(region_list))  # 'North | South | East | West'

# Cleaning a genuinely messy customer name — combining several methods in one line
raw_name = "  pRiYa   sHaRmA  "
clean = " ".join(raw_name.strip().title().split())
print(clean)   # 'Priya Sharma'
```

Walk through that last line carefully, because chaining methods like this is standard professional style: `raw_name.strip()` removes the outer whitespace → `.title()` fixes the capitalization → `.split()` breaks on *any* run of whitespace (collapsing the double space between "pRiYa" and "sHaRmA" into separate words) → `" ".join(...)` reassembles them with single spaces. Four operations, one line, and it handles a genuinely messy real-world input.

### 3.3 f-strings and Formatting (the Modern Way)

```python
product, units, price = "Laptop", 145, 62500.5
print(f"Product: {product}")
print(f"Units sold: {units:,}")
print(f"Revenue: ₹{units * price:,.2f}")     # you can put full expressions inside the braces
print(f"Growth: {0.157:.1%}")
```

### ⚠️ Common Mistakes

- Forgetting strings are immutable and expecting `text.strip()` alone to "clean" the variable.
- Using `.replace()` expecting it to be case-insensitive — it isn't; `"Sales".replace("sales", "Revenue")` does nothing because the case doesn't match.
- Confusing `.find()` (returns -1 on failure, never raises an error) with indexing (`text[100]` on a 10-character string *does* raise an `IndexError`).

### Cheat Sheet — Chapter 3

| Method | Purpose |
|---|---|
| `.strip() / .lstrip() / .rstrip()` | Remove whitespace (or specified chars) |
| `.lower() / .upper() / .title()` | Change case |
| `.replace(old, new)` | Substitute text |
| `.split(sep)` | String → list |
| `sep.join(list)` | List → string |
| `.find(x)` / `x in text` | Search |
| `.count(x)` | Count occurrences |
| `text[a:b:c]` | Slice |

### 🎯 Business Challenge — Chapter 3

*Task:* Clean the messy feedback `"  the DELIVERY was LATE and the product quality was POOR!!!  "`, check whether "late" or "poor" appear, and build a summary sentence.

```python
raw_feedback = "  the DELIVERY was LATE and the product quality was POOR!!!  "

clean_feedback = raw_feedback.strip().lower()
has_late = "late" in clean_feedback
has_poor = "poor" in clean_feedback

flags = []
if has_late:
    flags.append("late delivery")
if has_poor:
    flags.append("poor quality")

summary = f"Feedback flagged for: {', '.join(flags)}." if flags else "No issues detected."
print(summary)
```
