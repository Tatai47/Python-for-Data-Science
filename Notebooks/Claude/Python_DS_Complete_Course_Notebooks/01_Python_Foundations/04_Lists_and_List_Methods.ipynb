# Chapter 4: Lists and List Methods

### Why This Matters

A **list** is Python's default container for "a bunch of things in order" — monthly sales figures, a roster of customer IDs, product names. Before Pandas gives you the DataFrame, the list is your entire toolkit for sequences, and understanding it deeply makes everything downstream (NumPy arrays, Pandas Series) feel familiar rather than foreign, since they all borrow the same indexing/slicing vocabulary from Chapter 3.

### 4.1 Creating and Accessing Lists

```python
sales = [125000, 98000, 112000, 145000]
regions = ["North", "South", "East", "West"]
mixed = ["Laptop", 45, 62500.0, True]     # lists can hold mixed types — Python doesn't enforce one type

print(sales[0])       # 125000 — first element
print(sales[-1])      # 145000 — last element
print(sales[1:3])     # [98000, 112000] — same slicing rules as strings
print(len(sales))     # 4
print(sum(sales))     # 480000
print(max(sales), min(sales))
```

### 4.2 Modifying Lists — Lists Are Mutable

This is the flip side of Chapter 3's immutability lesson: **lists change in place**.

```python
products = ["Laptop", "Mouse", "Keyboard"]
products.append("Monitor")             # adds to the end
products.insert(1, "Headset")          # inserts at a specific position, shifting others right
products.extend(["Webcam", "Stand"])   # adds multiple items at once
print(products)

products.remove("Mouse")               # removes the FIRST matching value (errors if not found)
last = products.pop()                  # removes AND returns the last item
print(products)
print("Removed:", last)
```

⚠️ `.append()` vs `.extend()` is a genuinely common bug: `products.append(["Webcam", "Stand"])` would add *one new item that is itself a list* `["Webcam", "Stand"]`, not two new items. `.extend()` unpacks the iterable you give it; `.append()` never does.

### 4.3 Useful List Operations

```python
sales = [125, 98, 112, 145, 98, 132]

print(sales.count(98))     # 2 — how many times 98 appears
print(sales.index(145))    # 3 — position of the first 145

sales.sort()                # sorts IN PLACE — mutates sales, returns None
print(sales)
sales.sort(reverse=True)
print(sales)

print(sorted(sales))        # returns a NEW sorted list, leaves the original untouched
print(sales[::-1])          # reverse via slicing (also returns a new list/copy)
```

⚠️ **The single most common list bug in beginner code:** `x = sales.sort()` — this sets `x` to `None`, not to a sorted list, because `.sort()` mutates in place and returns nothing. If you want a new sorted list without touching the original, use `sorted(sales)` instead.

### 4.4 List Comprehensions — Learn This Well

A **list comprehension** is a compact way to build a new list by transforming or filtering an existing one. It replaces the classic "make an empty list, loop, append" pattern:

```python
sales = [125000, 98000, 112000, 145000, 87000]

# The traditional way
high = []
for s in sales:
    if s > 100000:
        high.append(s)

# The comprehension — does exactly the same thing, one line
high = [s for s in sales if s > 100000]
print(high)

# Comprehensions can transform, not just filter
in_lakhs = [round(s / 100000, 2) for s in sales]
print(in_lakhs)
```

Read `[s for s in sales if s > 100000]` left to right as: "give me `s`, for every `s` in `sales`, but only where `s > 100000`." Once this reading becomes automatic, you'll find yourself reaching for comprehensions constantly — and, more importantly, you'll be able to read the vectorized Pandas/NumPy operations in Volume 02 far more comfortably, because they express the exact same "transform every element" idea, just faster and without an explicit loop.

💡 **Why comprehensions matter beyond style:** they're measurably faster than an equivalent `for`-loop-with-`.append()` in CPython, because the loop machinery is implemented in C rather than interpreted Python bytecode. For small lists this is irrelevant; for anything you do millions of times, it adds up.

### Cheat Sheet — Chapter 4

| Operation | Method | Mutates original? |
|---|---|---|
| Add to end | `.append(x)` | Yes |
| Add many | `.extend([x, y])` | Yes |
| Insert at position | `.insert(i, x)` | Yes |
| Remove by value | `.remove(x)` | Yes |
| Remove by position | `.pop(i)` | Yes (returns removed item) |
| Sort in place | `.sort()` | Yes (returns `None`) |
| Sort, new list | `sorted(list)` | No |
| Reverse copy | `list[::-1]` | No |
| Filter/transform | `[expr for x in list if cond]` | No (builds new list) |

### 🎯 Business Challenge — Chapter 4

*Task:* Given monthly sales `[85000, 92000, 110000, 105000, 98000, 125000]`: total and average; months exceeding 100000; append a projected month (135000); sort descending.

```python
monthly_sales = [85000, 92000, 110000, 105000, 98000, 125000]

total = sum(monthly_sales)
average = total / len(monthly_sales)
strong_months = [s for s in monthly_sales if s > 100000]

monthly_sales.append(135000)
monthly_sales.sort(reverse=True)

print(f"Total sales      : ₹{total:,}")
print(f"Average sales    : ₹{average:,.0f}")
print(f"Months > 100k    : {strong_months}")
print(f"Updated (sorted) : {monthly_sales}")
```
