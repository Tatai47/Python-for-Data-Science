## Machine Learning, Framed for a Business Analyst

Machine learning is the practice of letting a computer find patterns in data and use them to predict or decide, rather than you hand-writing every rule. For your work specifically, four questions cover the overwhelming majority of real business ML use cases:

| Business Question | ML Type | Example |
|---|---|---|
| How much will we sell next month? | **Regression** | Sales/demand forecasting |
| Will this customer churn / default / convert? | **Classification** | Churn prediction, credit risk |
| Which customers behave similarly? | **Clustering** | Customer segmentation |
| Which factors actually drive the outcome? | Regression / Classification + feature importance | Marketing attribution |

Scikit-learn is the industry-standard library for all of the "classical" (non-deep-learning) techniques covering these questions, and virtually every model in it shares the exact same four-step interface — learn this pattern once, and you can use nearly any scikit-learn model with almost no new syntax:

```python
model = SomeModel(hyperparameters...)   # 1. choose and configure a model
model.fit(X_train, y_train)             # 2. train it on labeled data
model.predict(X_test)                   # 3. generate predictions on new data
model.score(X_test, y_test)             # 4. evaluate how good those predictions were
```
