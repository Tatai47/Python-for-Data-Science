## 2. Core concepts

CatBoost is designed for gradient boosting on decision trees with strong handling of categorical features and ordered boosting strategies.

## Categorical classification

Train on a tiny synthetic customer dataset.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
df=pd.DataFrame({'channel':['Email','Search','Social','Email','Search','Social','Email','Social'], 'region':['East','East','West','West','South','South','East','West'],'spend':[10,30,20,15,40,25,12,33],'churn':[0,0,1,0,0,1,0,1]})
X=df.drop(columns='churn'); y=df['churn']
cat=['channel','region']
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.25,random_state=42,stratify=y)
model=CatBoostClassifier(iterations=100,depth=4,learning_rate=.05,verbose=False)
model.fit(Xtr,ytr,cat_features=cat)
print(accuracy_score(yte,model.predict(Xte)))
```
