## Tokenization

Inspect tokens without requiring a language model.

```python
import spacy
nlp=spacy.blank('en')
doc=nlp('OpenAI launched a new model in San Francisco.')
print([(t.text,t.pos_) for t in doc])
```

## Entity extraction

Use a small rule-based pattern instead of assuming a model is installed.

```python
from spacy.matcher import PhraseMatcher
matcher=PhraseMatcher(nlp.vocab)
patterns=[nlp.make_doc('San Francisco'),nlp.make_doc('OpenAI')]
matcher.add('ORG_OR_PLACE',patterns)
doc=nlp('OpenAI launched an office in San Francisco.')
print([(doc[s:e].text, m) for m,s,e in matcher(doc)])
```
