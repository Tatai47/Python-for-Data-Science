## Regression example

Train a simple model if XGBoost is available.

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
rng=np.random.default_rng(7)
X=rng.normal(size=(1000,5))
y=3*X[:,0]-2*X[:,1]+rng.normal(0,.5,1000)
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.2,random_state=42)
model=xgb.XGBRegressor(n_estimators=200,max_depth=4,learning_rate=.05,subsample=.9,colsample_bytree=.9)
model.fit(Xtr,ytr)
pred=model.predict(Xte)
print('RMSE:',mean_squared_error(yte,pred)**0.5)
```
