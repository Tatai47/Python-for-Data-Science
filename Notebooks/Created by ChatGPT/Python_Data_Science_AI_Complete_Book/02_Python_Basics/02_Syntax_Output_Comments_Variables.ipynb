## 2.1 Python Syntax Rules (Very Important)

Python is **indentation-based**. Unlike Excel formulas or SQL, the amount of space at the beginning of a line matters.

### Key Rules
1. Use **4 spaces** (or 1 Tab) for indentation.
2. Statements usually end at the end of the line (no need for `;`).
3. Python is **case-sensitive** (`Revenue` ≠ `revenue`).
4. Comments start with `#`.

### Good vs Bad Indentation

```python
# Correct
if True:
    print("This is correctly indented")
    print("Both lines belong to the if block")
```

```python
# Incorrect – this will raise an IndentationError
# if True:
# print("This will fail")
```

## 2.2 Output – The `print()` Function

`print()` is how you display results to the user or yourself while analyzing data.

```python
print("Sales Report – Q3 2025")
print("Total Revenue:", 2450000)
print("Units Sold:", 18500)
print("Average Selling Price:", round(2450000 / 18500, 2))
```

```python
# Multiple values and separators
print("Region", "North", "South", "East", sep=" | ")
print("Revenue", 850000, 720000, 910000, sep=" → ")
```

```python
# Formatted output (f-strings – modern & recommended)
region = "West"
revenue = 980000
print(f"The {region} region generated ₹{revenue:,} in revenue.")
```

## 2.4 Variables – Storing Business Data

A variable is a named container that holds a value.

### Rules for Variable Names
- Must start with a letter or underscore `_`
- Can contain letters, numbers, underscores
- Cannot be a Python keyword (`if`, `for`, `class`, etc.)
- Case-sensitive
- Use **snake_case** (recommended for Python)

```python
# Good variable names for business analysis
total_revenue = 2450000
units_sold = 18500
avg_selling_price = total_revenue / units_sold
customer_segment = "Premium"
is_profitable = True

print(total_revenue)
print(avg_selling_price)
print(customer_segment)
```

```python
# Multiple assignment
north, south, east, west = 850000, 720000, 910000, 980000
print("North:", north, "| West:", west)
```

```python
# Re-assigning variables (common while cleaning data)
revenue = 1000000
print("Initial revenue:", revenue)

revenue = revenue * 1.1   # Apply 10% growth
print("After growth:", revenue)
```

## 2.5 Checking Data Type – `type()`

Knowing the type of your data is essential when reading files or databases.

```python
print(type(2450000))          # int
print(type(3.14))             # float
print(type("North Region"))   # str
print(type(True))             # bool
print(type([1, 2, 3]))        # list
```

## 2.6 Business Mini-Case: Monthly Performance Tracker

You are tracking performance of four regions.

```python
# Region-wise revenue (in ₹)
north_revenue = 1250000
south_revenue = 980000
east_revenue  = 1120000
west_revenue  = 1450000

total_revenue = north_revenue + south_revenue + east_revenue + west_revenue
average_revenue = total_revenue / 4

print("=" * 40)
print("MONTHLY REGIONAL PERFORMANCE")
print("=" * 40)
print(f"North   : ₹{north_revenue:,}")
print(f"South   : ₹{south_revenue:,}")
print(f"East    : ₹{east_revenue:,}")
print(f"West    : ₹{west_revenue:,}")
print("-" * 40)
print(f"Total   : ₹{total_revenue:,}")
print(f"Average : ₹{average_revenue:,.0f}")
print("=" * 40)
```
