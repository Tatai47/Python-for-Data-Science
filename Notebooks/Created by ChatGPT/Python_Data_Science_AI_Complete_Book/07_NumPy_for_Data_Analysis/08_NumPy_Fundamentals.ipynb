## 8.1 Creating Arrays

````python
import numpy as np

# From list
sales = np.array([125, 98, 112, 145, 130, 160])
print(sales)
print(type(sales))
print(sales.dtype)
````

````python
# Multi-dimensional (matrix)
region_month = np.array([
    [125, 130, 140],   # North
    [98, 105, 110],    # South
    [112, 115, 125],   # East
    [145, 150, 160]    # West
])
print(region_month)
print("Shape:", region_month.shape)   # (rows, columns)
````

````python
# Special arrays
print(np.zeros((3, 4)))
print(np.ones((2, 5)))
print(np.arange(0, 11, 2))
print(np.linspace(0, 1, 5))          # 5 evenly spaced numbers between 0 and 1
print(np.random.rand(3, 3))          # Random numbers [0,1)
print(np.random.randint(50, 150, size=(3, 4)))  # Random integers
````

---

## 8.2 Array Operations (Vectorization – The Superpower)

````python
sales = np.array([125, 98, 112, 145, 130, 160])
targets = np.array([120, 100, 110, 140, 125, 150])

# Element-wise operations (no loops needed!)
print("Difference :", sales - targets)
print("Achievement %:", (sales / targets) * 100)
print("Growth +10% :", sales * 1.1)
````

````python
# Aggregations
print("Total sales :", sales.sum())
print("Average     :", sales.mean())
print("Max         :", sales.max())
print("Min         :", sales.min())
print("Std Dev     :", sales.std())
print("Variance    :", sales.var())
````

---

## 8.3 Indexing, Slicing & Boolean Masking

````python
print(sales[0])
print(sales[1:4])
print(sales[-3:])

# Boolean masking – extremely powerful
print(sales > 120)
print(sales[sales > 120])           # Only high performers
print(sales[(sales > 100) & (sales < 140)])
````

````python
# 2D indexing
print(region_month[0])          # First row (North)
print(region_month[:, 1])       # Second column (all regions, month 2)
print(region_month[1:3, 0:2])   # Sub-matrix
````

---

## 8.4 Reshaping & Useful Functions

````python
arr = np.arange(12)
print(arr)
print(arr.reshape(3, 4))
print(arr.reshape(2, 6))
````

````python
# Statistical functions along axes
print("Row-wise mean (each region):", region_month.mean(axis=1))
print("Column-wise mean (each month):", region_month.mean(axis=0))
````

## 8.5 Business Mini-Case: Portfolio Return Simulation

````python
# Simulate 5 years of monthly returns for a portfolio
np.random.seed(42)
monthly_returns = np.random.normal(loc=0.01, scale=0.04, size=60)  # 5 years

cumulative = np.cumprod(1 + monthly_returns)
final_value = 1000000 * cumulative[-1]

print(f"Starting Capital : ₹10,00,000")
print(f"Final Value      : ₹{final_value:,.0f}")
print(f"Total Return     : {(final_value/1000000 - 1)*100:.1f}%")
print(f"Best Month       : {monthly_returns.max()*100:.1f}%")
print(f"Worst Month      : {monthly_returns.min()*100:.1f}%")
print(f"Average Monthly  : {monthly_returns.mean()*100:.2f}%")
````
