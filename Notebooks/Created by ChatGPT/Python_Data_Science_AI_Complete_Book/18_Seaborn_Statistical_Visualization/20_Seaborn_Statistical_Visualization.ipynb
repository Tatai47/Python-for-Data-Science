## 1. Setup

```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
np.random.seed(42)
```

## 2. Core concepts

Seaborn provides high-level statistical graphics built on Matplotlib. Good charting combines visual encoding with an explicit business question.

## Distribution

Compare customer spending by segment.

```python
df=pd.DataFrame({'spend':np.random.gamma(4,50,600),'segment':np.random.choice(['Value','Core','Premium'],600)})
sns.histplot(data=df,x='spend',hue='segment',kde=True)
plt.show()
```

## Relationship

Inspect a numeric relationship with a regression line.

```python
df['visits']=df['spend']/20 + np.random.normal(0,3,len(df))
sns.regplot(data=df,x='visits',y='spend')
plt.show()
```

## Categorical comparison

Compare distributions across segments.

```python
sns.boxplot(data=df,x='segment',y='spend')
plt.show()
```
