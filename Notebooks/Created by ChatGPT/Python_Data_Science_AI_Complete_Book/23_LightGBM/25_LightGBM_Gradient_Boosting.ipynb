## 2. Core concepts

Gradient boosting builds an additive model of weak learners. LightGBM emphasizes efficiency and uses histogram-based algorithms and leaf-wise growth.

## Classification example

Use the breast-cancer dataset when LightGBM is installed.

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.metrics import roc_auc_score
X,y=load_breast_cancer(return_X_y=True)
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.2,random_state=42,stratify=y)
model=lgb.LGBMClassifier(n_estimators=150,learning_rate=.05,num_leaves=31,verbose=-1)
model.fit(Xtr,ytr)
print(roc_auc_score(yte,model.predict_proba(Xte)[:,1]))
```
