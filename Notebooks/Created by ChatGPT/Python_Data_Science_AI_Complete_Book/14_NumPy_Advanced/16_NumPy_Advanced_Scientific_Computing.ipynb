## 2. Core concepts

NumPy stores homogeneous numeric data in compact arrays. The main performance advantage comes from vectorized operations implemented in compiled code. Broadcasting lets compatible shapes interact without explicit loops.

## Broadcasting and vectorization

Compare array arithmetic with a manual loop.

```python
a=np.array([10,20,30,40],dtype=float)
b=np.array([1,2,3,4],dtype=float)
print(a/b)
print(a*1.15)
```

## Linear algebra

Use matrix multiplication, determinants, and solving a linear system.

```python
A=np.array([[3,1],[1,2]],float)
y=np.array([9,8],float)
x=np.linalg.solve(A,y)
print('solution:',x)
print('det:',np.linalg.det(A))
```

## Simulation

Estimate the probability that the mean of 50 standard-normal observations exceeds 0.3.

```python
samples=rng.normal(size=(10000,50))
means=samples.mean(axis=1)
print((means>0.3).mean())
```
