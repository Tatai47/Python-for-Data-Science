## 2. Core concepts

Polars uses a columnar execution model and an expression system. Lazy mode allows query optimization before execution.

## Eager DataFrame

Create a small Polars dataset and aggregate sales.

```python
import polars as pl
df=pl.DataFrame({'region':['East','West','East','South'],'sales':[100,120,180,90]})
print(df.group_by('region').agg(pl.col('sales').sum().alias('total_sales')))
```

## Lazy query

Build a lazy pipeline and inspect its plan.

```python
lf=df.lazy().filter(pl.col('sales')>100).group_by('region').agg(pl.col('sales').mean().alias('avg_sales'))
print(lf.explain())
print(lf.collect())
```
