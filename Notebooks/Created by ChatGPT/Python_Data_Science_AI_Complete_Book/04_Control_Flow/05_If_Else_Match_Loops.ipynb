## 5.1 If – Elif – Else (Decision Making)

Business rules are full of conditions:  
- If sales > target → bonus  
- If margin < 15% → review pricing  
- If customer is Premium → special offer

````python
sales = 1350000
target = 1200000

if sales > target:
    print("Target Achieved! Congratulations.")
elif sales == target:
    print("Exactly met the target.")
else:
    print("Target missed. Action required.")
````

````python
# Nested conditions & business grading
score = 87

if score >= 90:
    grade = "Outstanding"
elif score >= 75:
    grade = "Excellent"
elif score >= 60:
    grade = "Good"
elif score >= 40:
    grade = "Average"
else:
    grade = "Needs Improvement"

print(f"Performance Grade: {grade}")
````

````python
# Multiple conditions
revenue = 1850000
margin = 0.18
region = "West"

if revenue > 1500000 and margin > 0.15:
    print("High performing region – eligible for expansion budget")
elif revenue > 1000000 or margin > 0.20:
    print("Solid performance – maintain current strategy")
else:
    print("Underperforming – deep dive required")
````

## 5.2 Match-Case (Python 3.10+) – Cleaner Multi-way Branching

Useful when you have many discrete categories (regions, product lines, customer tiers).

````python
region = "West"

match region:
    case "North":
        manager = "Anita Sharma"
        target = 1500000
    case "South":
        manager = "Ravi Kumar"
        target = 1200000
    case "East":
        manager = "Priya Patel"
        target = 1100000
    case "West":
        manager = "Vikram Singh"
        target = 1600000
    case _:
        manager = "Unassigned"
        target = 0

print(f"Region: {region} | Manager: {manager} | Target: ₹{target:,}")
````

## 5.3 While Loops – Repeat Until Condition is False

Less common in data analysis than `for` loops, but useful for simulations or waiting for a condition.

````python
# Simple while example – compound interest simulation
principal = 100000
rate = 0.08
years = 0
target = 200000

while principal < target:
    principal *= (1 + rate)
    years += 1

print(f"It will take {years} years to double the money at 8% interest.")
print(f"Final amount: ₹{principal:,.0f}")
````

## 5.4 For Loops – The Workhorse of Data Analysis

You will use `for` loops constantly until you switch fully to Pandas vectorized operations.

````python
regions = ["North", "South", "East", "West"]
sales = [1250000, 980000, 1120000, 1450000]

for region in regions:
    print(f"Processing {region} region...")
````

````python
# Looping with index
for i in range(len(regions)):
    print(f"{regions[i]:8} → ₹{sales[i]:,}")
````

````python
# zip() – elegant way to loop over multiple lists together
for region, amount in zip(regions, sales):
    print(f"{region:8} → ₹{amount:,}")
````

````python
# range() function
print(list(range(5)))          # 0 to 4
print(list(range(1, 6)))       # 1 to 5
print(list(range(0, 11, 2)))   # Even numbers
````

````python
# Looping over dictionary
region_sales = {"North": 1250000, "South": 980000, "East": 1120000, "West": 1450000}

for region, amount in region_sales.items():
    status = "Above Target" if amount > 1100000 else "Below Target"
    print(f"{region:8} | ₹{amount:,} | {status}")
````

## 5.5 Break & Continue

````python
# break – exit the loop early
sales = [850, 920, 1100, 780, 1350, 990]

for s in sales:
    if s > 1200:
        print(f"Found high performer: {s}. Stopping search.")
        break
    print(f"Checking {s}...")
````

````python
# continue – skip current iteration
for s in sales:
    if s < 900:
        continue          # Skip low values
    print(f"Good sales: {s}")
````

## 5.6 Business Mini-Case: Commission Calculator

````python
sales_team = {
    "Anita": 1450000,
    "Ravi": 980000,
    "Priya": 1620000,
    "Vikram": 875000,
    "Sneha": 2100000
}

print("SALES COMMISSION REPORT")
print("=" * 50)

for name, amount in sales_team.items():
    if amount >= 2000000:
        rate = 0.08
    elif amount >= 1500000:
        rate = 0.06
    elif amount >= 1000000:
        rate = 0.04
    else:
        rate = 0.02
    
    commission = amount * rate
    print(f"{name:10} | Sales: ₹{amount:>10,} | Rate: {rate*100:.0f}% | Commission: ₹{commission:>9,.0f}")

print("=" * 50)
````
