## 2. Core concepts

Streamlit maps Python functions and widgets to a reactive app. It is excellent for internal analytics prototypes, demos, and lightweight data products.

## Minimal app

A complete starter app.

```python
# app.py
import streamlit as st
import pandas as pd

st.title('Sales Explorer')
region=st.selectbox('Region',['All','East','West','South'])
df=pd.DataFrame({'region':['East','West','South'],'sales':[120,150,90]})
if region!='All':
    df=df[df['region']==region]
st.metric('Total Sales',f"{df['sales'].sum():,.0f}")
st.dataframe(df,use_container_width=True)
```
