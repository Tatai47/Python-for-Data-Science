## 1. Setup

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## 2. Core concepts

Matplotlib is an object-oriented plotting library. The Figure contains Axes, and each Axes contains the graphical elements used to communicate a metric.

## Subplots

Create two linked business views.

```python
months=pd.date_range('2026-01-01', periods=6, freq='MS')
sales=[120,135,150,142,168,180]
margin=[18,20,21,19,23,24]
fig,ax=plt.subplots(2,1,figsize=(8,6),sharex=True)
ax[0].plot(months,sales,marker='o')
ax[0].set_ylabel('Sales')
ax[1].bar(months,margin)
ax[1].set_ylabel('Margin %')
plt.tight_layout()
plt.show()
```

## Annotations

Highlight the best month.

```python
best=int(np.argmax(sales))
fig,ax=plt.subplots(figsize=(8,4))
ax.plot(months,sales,marker='o')
ax.annotate('Peak',xy=(months[best],sales[best]),xytext=(0.6,0.85),textcoords='axes fraction',arrowprops={'arrowstyle':'->'})
plt.show()
```
