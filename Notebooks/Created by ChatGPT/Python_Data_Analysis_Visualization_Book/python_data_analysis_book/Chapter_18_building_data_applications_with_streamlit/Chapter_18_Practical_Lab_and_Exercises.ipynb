## Worked reference solution

The following is one valid approach. It is not the only possible solution. Compare your work with it, and explain any differences in assumptions, handling, or output.

```python
app_code = r'''import streamlit as st
import pandas as pd
from pathlib import Path
st.set_page_config(page_title="Sales Analysis",layout="wide")
st.title("Sales Analysis Dashboard")
DATA=Path(__file__).resolve().parents[1]/"data"/"sales.csv"
@st.cache_data(ttl=300)
def load_data(): return pd.read_csv(DATA,parse_dates=["order_date"])
sales=load_data()
regions=st.sidebar.multiselect("Region",sorted(sales.region.unique()),default=sorted(sales.region.unique()))
products=st.sidebar.multiselect("Product",sorted(sales["product"].unique()),default=sorted(sales["product"].unique()))
df=sales[sales.region.isin(regions)&sales["product"].isin(products)]
a,b,c=st.columns(3); a.metric("Orders",len(df)); b.metric("Revenue",f"{df.revenue.sum():,.0f}"); c.metric("Profit",f"{df.profit.sum():,.0f}")
left,right=st.columns(2); left.subheader("Revenue by region"); left.bar_chart(df.groupby("region").revenue.sum()); right.subheader("Daily revenue"); right.line_chart(df.groupby("order_date").revenue.sum())
st.subheader("Filtered orders"); st.dataframe(df,use_container_width=True)
st.download_button("Download filtered CSV",df.to_csv(index=False).encode("utf-8"),"filtered_sales.csv","text/csv")
'''
from pathlib import Path
Path("app.py").write_text(app_code,encoding="utf-8"); print("Wrote app.py. Run: streamlit run app.py")
```
