### Step 1 — Guided 1 — Define the app layout

Use page configuration before other Streamlit commands. Keep imports and data-loading functions at the top of the script.

```python
app_code = """import streamlit as st
st.set_page_config(page_title='Sales Explorer', layout='wide')
st.title('Sales Explorer')
st.write('A small interactive sales dashboard')
"""
print(app_code)
```

### Step 2 — Guided 2 — Add user controls

Multiselect returns a list of selected values. A widget key is useful when multiple controls have similar labels.

```python
controls = """regions = st.sidebar.multiselect('Region', options=sorted(sales.region.unique()), default=sorted(sales.region.unique()), key='region_filter')
filtered = sales[sales.region.isin(regions)]
"""
print(controls)
```

### Step 3 — Guided 3 — Add KPIs and charts

Metrics should have explicit units and definitions. Use Plotly for richer interaction or built-in charts for simple cases.

```python
display_code = """a,b,c = st.columns(3)
a.metric('Orders', len(filtered))
b.metric('Revenue', f'{filtered.revenue.sum():,.0f}')
c.metric('Profit', f'{filtered.profit.sum():,.0f}')
st.bar_chart(filtered.groupby('region').revenue.sum())
"""
print(display_code)
```

Save the script as `app.py`, then run `streamlit run app.py` in a terminal. Test empty filters, all filters, missing values, and a fresh browser session.
