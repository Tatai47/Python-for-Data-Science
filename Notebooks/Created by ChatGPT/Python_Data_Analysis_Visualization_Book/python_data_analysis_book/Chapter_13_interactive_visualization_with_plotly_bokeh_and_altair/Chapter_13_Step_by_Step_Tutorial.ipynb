### Step 1 — Guided 1 — Create a Plotly figure

Plotly Express maps table columns to visual encodings. Keep units in axis titles and include useful hover values.

```python
from pathlib import Path
ROOT=Path.cwd()
if not (ROOT/"data").exists() and (ROOT.parent/"data").exists(): ROOT=ROOT.parent
DATA=ROOT/"data"
print("Data directory:", DATA.resolve())
import pandas as pd, plotly.express as px
sales=pd.read_csv(DATA/"sales.csv")
fig=px.bar(sales.groupby("region",as_index=False).revenue.sum(),x="region",y="revenue",title="Revenue by region"); fig.show()
```

### Step 2 — Guided 2 — Customize a trace

`update_layout` changes figure-level properties; `update_traces` changes properties shared by selected traces.

```python
fig.update_layout(xaxis_title="Region",yaxis_title="Total revenue",template="plotly_white"); fig.update_traces(hovertemplate="Region=%{x}<br>Revenue=%{y:,.0f}<extra></extra>"); fig.show()
```

### Step 3 — Guided 3 — Add a Bokeh bar glyph

Bokeh separates plot model construction from rendering. Install it to run the example.

```python
try:
 from bokeh.plotting import figure,show
 region=sales.groupby("region",as_index=False).revenue.sum()
 p=figure(x_range=region.region.tolist(),height=320,title="Bokeh revenue by region"); p.vbar(x=region.region.tolist(),top=region.revenue.tolist(),width=.7); show(p)
except ImportError: print("pip install bokeh")
```

### Step 4 — Guided 4 — Declare a chart with Altair

The type suffix tells Altair how to interpret a field. `N` means nominal, `Q` quantitative, and `T` temporal.

```python
try:
 import altair as alt
 display(alt.Chart(sales).mark_point().encode(x="units:Q",y="revenue:Q",color="region:N",tooltip=["order_id:N","product:N"]))
except ImportError: print("pip install altair")
```
