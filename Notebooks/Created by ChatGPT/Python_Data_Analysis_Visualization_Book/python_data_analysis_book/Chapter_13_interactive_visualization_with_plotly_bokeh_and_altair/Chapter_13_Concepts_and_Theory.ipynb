### Plotly Express and Graph Objects
Plotly Express (`px`) is a high-level API: `px.scatter(data_frame,x,y,color,size,hover_name,facet_col,...)`; other functions include `line`, `bar`, `histogram`, and `box`. Graph Objects (`go`) gives lower-level control with `go.Figure`, traces (`go.Scatter`, `go.Bar`), `update_layout`, and `update_traces`. Use Express for common charts and Graph Objects for trace-level control.

#### Demonstration 1 — Plotly Express bar chart

Run the code and inspect both the output and the assumptions behind it.

```python
from pathlib import Path
ROOT=Path.cwd()
if not (ROOT/"data").exists() and (ROOT.parent/"data").exists(): ROOT=ROOT.parent
DATA=ROOT/"data"
print("Data directory:", DATA.resolve())
import pandas as pd, plotly.express as px
sales=pd.read_csv(DATA/"sales.csv")
region=sales.groupby("region",as_index=False).revenue.sum()
fig=px.bar(region,x="region",y="revenue",title="Revenue by region"); fig.update_layout(xaxis_title="Region",yaxis_title="Revenue"); fig.show()
```

### Layout and interaction
A Plotly Figure has `data` traces and a `layout`. Use `update_layout(title,xaxis_title,yaxis_title,template)` and `update_traces(hovertemplate=...)`. `fig.show()` renders; `write_html(path,include_plotlyjs=...)` exports HTML. Give the default chart a useful view and hover fields; excessive interaction can add friction.

#### Demonstration 2 — Plotly scatter with hover fields

Run the code and inspect both the output and the assumptions behind it.

```python
fig=px.scatter(sales,x="units",y="revenue",color="region",hover_data=["product","order_id"],title="Units vs revenue"); fig.show()
```

### Bokeh model
`figure()` creates a plot. Glyphs such as `line`, `circle`, `scatter`, and `vbar` add marks. `ColumnDataSource` provides named columns for linked selection and hover tools. Tools include pan, wheel zoom, box zoom, reset, save, and hover. Use `output_file`/`save` for HTML output or notebook output for inline rendering.

#### Demonstration 3 — Graph Objects trace

Run the code and inspect both the output and the assumptions behind it.

```python
import plotly.graph_objects as go
product=sales.groupby("product",as_index=False).profit.sum().sort_values("profit",ascending=False)
fig=go.Figure(data=[go.Bar(x=product["product"],y=product.profit,name="Profit")]); fig.update_layout(title="Profit by product",xaxis_title="Product",yaxis_title="Profit"); fig.show()
```

### Altair and chart grammar
Altair is declarative: `alt.Chart(data).mark_point().encode(x="units:Q",y="revenue:Q",color="region:N")`. Types: `Q` quantitative, `N` nominal, `O` ordinal, `T` temporal. Transforms include `transform_filter`, `transform_aggregate`, and `transform_calculate`. Parameters/selections drive interactive filters. Large datasets may encounter row limits.

#### Demonstration 4 — Bokeh glyph example (optional)

Run the code and inspect both the output and the assumptions behind it.

```python
try:
 from bokeh.plotting import figure,show
 p=figure(title="Profit by product",x_range=product["product"].tolist(),height=350,width=650,tools="pan,wheel_zoom,reset,save")
 p.vbar(x=product["product"].tolist(),top=product.profit.tolist(),width=.7); p.xaxis.axis_label="Product"; p.yaxis.axis_label="Profit"; show(p)
except ImportError: print("Install Bokeh: pip install bokeh")
```

### Library selection
Plotly has broad chart coverage and integrates with dashboard frameworks; Bokeh offers browser-oriented plot models and linked interactions; Altair provides concise declarative specifications. Evaluate deployment, accessibility, performance, maintainability, team skill, and offline behavior.

#### Demonstration 5 — Altair declarative chart (optional)

Run the code and inspect both the output and the assumptions behind it.

```python
try:
 import altair as alt
 chart=alt.Chart(sales).mark_circle(size=65,opacity=.6).encode(x=alt.X("units:Q"),y=alt.Y("revenue:Q"),color="region:N",tooltip=["order_id:N","product:N","revenue:Q"]).properties(title="Units vs revenue",width=600,height=350)
 display(chart)
except ImportError: print("Install Altair: pip install altair")
```
