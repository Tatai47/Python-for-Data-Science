## Worked reference solution

```python
import pandas as pd, plotly.express as px
sales=pd.read_csv(DATA/"sales.csv"); region=sales.groupby("region",as_index=False).agg(revenue=("revenue","sum"),profit=("profit","sum"),orders=("order_id","count"))
fig=px.scatter(region,x="orders",y="revenue",size="profit",color="region",hover_name="region",title="Regional performance"); fig.update_layout(xaxis_title="Order count",yaxis_title="Revenue"); fig.show()
print("Equivalent Bokeh and Altair examples are in the tutorial notebook.")
```
