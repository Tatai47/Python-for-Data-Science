### Step 1 — Guided 1 — Ask an analytical question

Define the metric and observational grain before loading data. Here each row is one order record and revenue is the sum of `revenue`.

```python
from pathlib import Path
ROOT=Path.cwd()
if not (ROOT/"data").exists() and (ROOT.parent/"data").exists(): ROOT=ROOT.parent
DATA=ROOT/"data"
print("Data directory:", DATA.resolve())
import pandas as pd
sales=pd.read_csv(DATA/"sales.csv",parse_dates=["order_date"])
print("Question: which region generated the highest total revenue?")
print("Metric: sum of revenue, grouped by region")
```

### Step 2 — Guided 2 — Inspect the result

`head(n)` previews rows; `info()` prints non-null counts and dtypes; `describe()` computes descriptive statistics.

```python
sales.info()
display(sales.head(3)); display(sales[["units","revenue","profit"]].describe())
```

### Step 3 — Guided 3 — Summarize and rank

Group with `groupby`; aggregate with named `.agg`; sort the resulting summary, not the original raw observations.

```python
summary=(sales.groupby("region",as_index=False).agg(revenue=("revenue","sum"),orders=("order_id","count")).sort_values("revenue",ascending=False))
display(summary)
```

### Step 4 — Guided 4 — Write a defensible conclusion

Separate the descriptive finding from a causal explanation. Ranking does not explain why a region performed better.

```python
leader=summary.iloc[0]
print(f"Top region: {leader.region}; total revenue: {leader.revenue:,.2f}")
print("This is a descriptive ranking, not evidence of a causal driver.")
```
