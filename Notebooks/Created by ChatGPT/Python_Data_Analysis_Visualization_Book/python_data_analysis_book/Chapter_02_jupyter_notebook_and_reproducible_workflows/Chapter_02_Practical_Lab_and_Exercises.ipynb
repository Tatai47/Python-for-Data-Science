```python
import pandas as pd
sales=pd.read_csv(DATA/"sales.csv",parse_dates=["order_date"])
checks={"rows":len(sales),"duplicate_order_ids":int(sales.order_id.duplicated().sum()),"missing_cells":int(sales.isna().sum().sum()),"start":str(sales.order_date.min().date()),"end":str(sales.order_date.max().date())}
summary=sales.groupby("region",as_index=False).revenue.sum().sort_values("revenue",ascending=False)
print(checks); display(summary)
```
