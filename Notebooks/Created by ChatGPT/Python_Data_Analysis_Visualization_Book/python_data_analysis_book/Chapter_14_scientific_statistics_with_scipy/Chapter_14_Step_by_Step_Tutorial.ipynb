### Step 1 — Guided 1 — Describe before testing

Plot and summarize the variables first. Check the sample size, spread, skewness, and missingness before choosing a test.

```python
from pathlib import Path
ROOT=Path.cwd()
if not (ROOT/"data").exists() and (ROOT.parent/"data").exists(): ROOT=ROOT.parent
DATA=ROOT/"data"
print("Data directory:", DATA.resolve())
import pandas as pd
from scipy import stats
sales=pd.read_csv(DATA/"sales.csv"); x=sales.profit.dropna().to_numpy()
print("n, mean, median, sd:",len(x),x.mean(),stats.tmean(x),x.std(ddof=1))
print("skewness:",stats.skew(x,bias=False))
```

### Step 2 — Guided 2 — Match a test to the design

Use Welch independent t-test when comparing two independent group means without assuming equal variances. Paired measurements require a paired design.

```python
north=sales.loc[sales.region=="North","profit"]; west=sales.loc[sales.region=="West","profit"]
print(stats.ttest_ind(north,west,equal_var=False))
```

### Step 3 — Guided 3 — Interpret probability and p-values

The p-value is conditional on the null model and assumptions. It does not measure effect magnitude or the probability the null is true.

```python
test=stats.ttest_ind(north,west,equal_var=False); print("p-value:",test.pvalue); print("mean difference:",north.mean()-west.mean())
```

### Step 4 — Guided 4 — Report an interval

Report a confidence interval together with the point estimate and units. This interval describes uncertainty in the mean, not the range containing 95% of individual orders.

```python
n=len(north); ci=stats.t.interval(.95,n-1,loc=north.mean(),scale=stats.sem(north)); print("North mean:",north.mean(),"95% CI:",ci)
```
