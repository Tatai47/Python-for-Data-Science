## Worked reference solution

The following is one valid approach. It is not the only possible solution. Compare your work with it, and explain any differences in assumptions, handling, or output.

```python
import pandas as pd
from scipy import stats
sales=pd.read_csv(DATA/"sales.csv")
summary=sales.groupby("region").profit.agg(n="count",mean="mean",median="median",sd="std"); display(summary)
north=sales.loc[sales.region=="North","profit"]; south=sales.loc[sales.region=="South","profit"]
test=stats.ttest_ind(north,south,equal_var=False)
ci=stats.t.interval(.95,len(north)-1,loc=north.mean(),scale=stats.sem(north))
print("Welch test:",test); print("North mean 95% CI:",ci); print("Mean difference North-South:",north.mean()-south.mean())
print("Correlation:",stats.pearsonr(sales.units,sales.revenue))
```
