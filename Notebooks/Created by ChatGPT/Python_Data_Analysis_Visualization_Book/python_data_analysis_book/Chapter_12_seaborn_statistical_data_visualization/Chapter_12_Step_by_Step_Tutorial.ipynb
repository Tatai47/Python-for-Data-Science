### Step 1 — Guided 1 — Compare distributions

Use density when comparing distribution shape across groups with different sample sizes.

```python
from pathlib import Path
ROOT=Path.cwd()
if not (ROOT/"data").exists() and (ROOT.parent/"data").exists(): ROOT=ROOT.parent
DATA=ROOT/"data"
print("Data directory:", DATA.resolve())
import pandas as pd,seaborn as sns,matplotlib.pyplot as plt
sales=pd.read_csv(DATA/"sales.csv"); sns.set_theme(style="whitegrid")
ax=sns.histplot(data=sales,x="revenue",hue="region",stat="density",common_norm=False,bins=20); plt.show()
```

### Step 2 — Guided 2 — Examine numeric association

Scatter plots reveal form, spread, clusters, and potential outliers. A correlation coefficient alone hides these features.

```python
ax=sns.scatterplot(data=sales,x="units",y="revenue",hue="product",alpha=.7); plt.show()
```

### Step 3 — Guided 3 — Compare group means and spread

A barplot estimates a summary statistic. A boxplot shows distributional spread. Do not confuse a barplot with a countplot.

```python
fig,axes=plt.subplots(1,2,figsize=(10,4)); sns.barplot(data=sales,x="region",y="profit",errorbar=None,color="steelblue",ax=axes[0]); sns.boxplot(data=sales,x="region",y="profit",ax=axes[1]); plt.show()
```

### Step 4 — Guided 4 — Check correlation

Correlation can be driven by scale, outliers, shared components, or confounding. It is not a causal measure.

```python
display(sales.select_dtypes(include="number").corr().round(2))
```
