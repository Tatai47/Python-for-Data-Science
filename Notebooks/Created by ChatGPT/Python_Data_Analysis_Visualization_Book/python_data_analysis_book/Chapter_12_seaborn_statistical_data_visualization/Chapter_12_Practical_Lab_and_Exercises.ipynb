## Worked reference solution

The following is one valid approach. It is not the only possible solution. Compare your work with it, and explain any differences in assumptions, handling, or output.

```python
import pandas as pd,seaborn as sns,matplotlib.pyplot as plt
sales=pd.read_csv(DATA/"sales.csv"); sns.set_theme(style="whitegrid")
fig,axes=plt.subplots(2,2,figsize=(13,9),layout="constrained")
sns.histplot(data=sales,x="revenue",hue="region",bins=20,stat="density",common_norm=False,ax=axes[0,0])
sns.scatterplot(data=sales,x="units",y="revenue",hue="product",alpha=.7,ax=axes[0,1])
sns.boxplot(data=sales,x="region",y="profit",ax=axes[1,0])
sns.heatmap(sales.select_dtypes(include="number").corr(),cmap="vlag",center=0,annot=True,fmt=".2f",ax=axes[1,1])
axes[0,0].set_title("Revenue distribution"); axes[0,1].set_title("Units and revenue"); axes[1,0].set_title("Profit spread"); axes[1,1].set_title("Correlation"); plt.show()
```
