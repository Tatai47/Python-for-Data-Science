### Step 1 — Guided 1 — Establish the shape

Check the number of rows and columns before assuming the import is complete.

```python
from pathlib import Path
ROOT=Path.cwd()
if not (ROOT/"data").exists() and (ROOT.parent/"data").exists(): ROOT=ROOT.parent
DATA=ROOT/"data"
print("Data directory:", DATA.resolve())
import pandas as pd
sales=pd.read_csv(DATA/"sales.csv",parse_dates=["order_date"])
print("shape:",sales.shape); display(sales.head())
```

### Step 2 — Guided 2 — Filter a region

Use `.loc` to make the row condition and selected columns explicit.

```python
north=sales.loc[sales.region.eq("North"),["order_id","order_date","product","revenue"]]
display(north.head())
```

### Step 3 — Guided 3 — Summarize numeric columns

`describe().T` rotates the summary so each variable appears on a row.

```python
display(sales[["units","unit_price","revenue","profit"]].describe().T)
```

### Step 4 — Guided 4 — Compute profit margin

The ratio is defined per order. For overall margin, divide total profit by total revenue rather than averaging ratios without weighting.

```python
sales["profit_margin"]=sales.profit/sales.revenue
print("overall margin:",sales.profit.sum()/sales.revenue.sum())
display(sales[["revenue","profit","profit_margin"]].head())
```
