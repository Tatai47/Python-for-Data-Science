## Worked reference solution

The following is one valid approach. It is not the only possible solution. Compare your work with it, and explain any differences in assumptions, handling, or output.

```python
import pandas as pd
sales=pd.read_csv(DATA/"sales.csv",parse_dates=["order_date"],dtype={"order_id":"string"})
profile={"rows":len(sales),"columns":sales.shape[1],"duplicate_ids":int(sales.order_id.duplicated().sum()),"missing":int(sales.isna().sum().sum())}
sales["profit_margin"]=sales.profit/sales.revenue
north=sales.loc[sales.region.eq("North"),["order_id","order_date","product","revenue","profit_margin"]]
print(profile,"North orders:",len(north),"weighted margin:",round(sales.profit.sum()/sales.revenue.sum(),4)); display(north.head())
```
