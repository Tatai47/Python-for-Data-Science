```python
import numpy as np
marks=np.array([[78,81,74],[91,89,93],[64,70,78],[88,92,85],[55,61,58]],dtype=float)
mu=marks.mean(axis=0); sd=marks.std(axis=0); passed=(marks>=40).all(axis=1); z=(marks-mu)/sd
print("means:",mu.round(2),"sample? population SD:",sd.round(2)); print("passed all:",passed); print("z means:",z.mean(axis=0).round(10))
```
