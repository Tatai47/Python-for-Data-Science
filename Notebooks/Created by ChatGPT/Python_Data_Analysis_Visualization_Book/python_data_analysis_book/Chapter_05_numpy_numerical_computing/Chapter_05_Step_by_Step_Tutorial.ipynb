### Step 1 — Guided 1 — Create a mark matrix

Represent observations as rows and features as columns. Verify shape and dtype before calculations.

```python
import numpy as np
marks=np.array([[78,81,74],[91,89,93],[64,70,78],[88,92,85],[55,61,58]],dtype=float)
print(marks.shape,marks.dtype); print(marks[0])
```

### Step 2 — Guided 2 — Calculate along the correct axis

`axis=0` returns one result per subject; `axis=1` returns one result per student.

```python
print("subject means:",marks.mean(axis=0))
print("student totals:",marks.sum(axis=1))
```

### Step 3 — Guided 3 — Use boolean masks

Boolean arrays can be reduced with `.all` and `.any` to test rows or columns.

```python
passed_every_subject=(marks>=40).all(axis=1)
high_scores=marks[marks>=90]
print(passed_every_subject,high_scores)
```

### Step 4 — Guided 4 — Standardize columns

Z-scores use `(value - mean) / standard deviation`; keep dimensions for clear broadcasting.

```python
mu=marks.mean(axis=0,keepdims=True); sigma=marks.std(axis=0,keepdims=True)
z=(marks-mu)/sigma
print(np.round(z,2)); print("column mean:",z.mean(axis=0))
```
