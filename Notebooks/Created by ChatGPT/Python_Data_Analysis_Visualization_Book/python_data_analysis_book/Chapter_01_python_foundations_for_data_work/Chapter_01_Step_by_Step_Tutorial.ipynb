### Step 1 — Guided 1 — Summarize a list

Python lists can hold values of mixed types, but numeric analyses should validate that inputs are numeric.

```python
marks=[78,91,64,88,73,95]
print({"count":len(marks),"total":sum(marks),"average":sum(marks)/len(marks),"min":min(marks),"max":max(marks)})
```

### Step 2 — Guided 2 — Pair sequences

`enumerate(iterable, start=0)` yields index-value pairs. `zip(*iterables)` combines positions and stops at the shortest iterable unless `strict=True` (Python 3.10+).

```python
names=["Asha","Kabir","Diya"]; marks=[89,74,96]
for position,(name,mark) in enumerate(zip(names,marks),start=1): print(position,name,mark)
```

### Step 3 — Guided 3 — Validate inputs

Use a clear exception to make invalid inputs visible at the boundary of a function.

```python
def pass_rate(scores, pass_mark=40):
    if not scores: raise ValueError("scores cannot be empty")
    if not all(isinstance(x,(int,float)) and not isinstance(x,bool) for x in scores): raise TypeError("scores must be numeric")
    return sum(x>=pass_mark for x in scores)/len(scores)*100
print(f"Pass rate: {pass_rate([35,60,72,40]):.1f}%")
```

### Step 4 — Guided 4 — Work safely with files

`Path.read_text` and `write_text` are concise for small text files. Use `open` with a context manager to stream larger files.

```python
from pathlib import Path
p=Path("report.txt"); p.write_text("Class report\nPass rate: 75%\n",encoding="utf-8")
print(p.read_text(encoding="utf-8"))
```
