## Worked reference solution

The following is one valid approach. It is not the only possible solution. Compare your work with it, and explain any differences in assumptions, handling, or output.

```python
def summarise_marks(marks, pass_mark=40):
    if len(marks)==0: raise ValueError("marks must not be empty")
    if not all(isinstance(x,(int,float)) and not isinstance(x,bool) for x in marks): raise TypeError("marks must be numeric")
    return {"count":len(marks),"average":sum(marks)/len(marks),"highest":max(marks),"lowest":min(marks),"pass_count":sum(x>=pass_mark for x in marks),"top_three":sorted(marks,reverse=True)[:3]}
class_marks=[78,91,64,88,73,95,38]
summarise_marks(class_marks)
```
