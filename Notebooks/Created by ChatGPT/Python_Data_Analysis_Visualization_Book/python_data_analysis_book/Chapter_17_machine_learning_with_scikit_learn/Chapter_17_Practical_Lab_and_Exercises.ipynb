## Worked reference solution

The following is one valid approach. It is not the only possible solution. Compare your work with it, and explain any differences in assumptions, handling, or output.

```python
import pandas as pd
from sklearn.model_selection import train_test_split,cross_val_score
from sklearn.dummy import DummyRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
sales=pd.read_csv(DATA/"sales.csv"); X=sales[["units","discount"]]; y=sales.revenue
Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.25,random_state=42)
for name,est in [("Baseline",DummyRegressor()),("Linear",LinearRegression())]:
 est.fit(Xtr,ytr); p=est.predict(Xte); print(name,{"MAE":mean_absolute_error(yte,p),"RMSE":mean_squared_error(yte,p)**.5,"R2":r2_score(yte,p)})
cv=cross_val_score(LinearRegression(),Xtr,ytr,cv=5,scoring="neg_mean_absolute_error"); print("CV MAE:",-cv.mean())
```
