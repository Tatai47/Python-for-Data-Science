## Worked reference solution

The following is one valid approach. It is not the only possible solution. Compare your work with it, and explain any differences in assumptions, handling, or output.

```python
import pandas as pd
raw=pd.read_csv(DATA/"messy_sales.csv"); clean=raw.copy()
clean["order_date"]=pd.to_datetime(clean.order_date,errors="coerce"); clean["units"]=pd.to_numeric(clean.units,errors="coerce")
for col in ["region","product"]: clean[col]=clean[col].astype("string").str.strip().str.title()
report={"rows":len(raw),"columns":len(raw.columns),"duplicate_order_ids":int(raw.order_id.duplicated().sum()),"missing_rate":raw.isna().mean().round(3).to_dict(),"invalid_dates":int(clean.order_date.isna().sum()),"missing_units_after_parse":int(clean.units.isna().sum())}
print(report); display(clean.groupby("region",dropna=False).revenue.agg(["count","mean","median","sum"]))
```
