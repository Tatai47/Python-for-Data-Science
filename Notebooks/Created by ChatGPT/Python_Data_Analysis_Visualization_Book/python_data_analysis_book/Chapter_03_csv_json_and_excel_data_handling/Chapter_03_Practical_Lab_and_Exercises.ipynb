## Worked reference solution

The following is one valid approach. It is not the only possible solution. Compare your work with it, and explain any differences in assumptions, handling, or output.

```python
import json, pandas as pd
sales=pd.read_csv(DATA/"sales.csv",dtype={"order_id":"string"},parse_dates=["order_date"])
required={"order_id","order_date","region","revenue"}; missing=required-set(sales.columns)
if missing: raise ValueError(f"Missing columns: {sorted(missing)}")
if not sales.order_id.is_unique: raise ValueError("order_id must be unique")
weather=pd.DataFrame(json.loads((DATA/"weather.json").read_text(encoding="utf-8")))
report=sales.groupby("region",as_index=False).agg(orders=("order_id","count"),revenue=("revenue","sum"))
report.to_csv("region_report.csv",index=False); report.to_json("region_report.json",orient="records",indent=2)
with pd.ExcelWriter("region_report.xlsx",engine="openpyxl") as writer:
 report.to_excel(writer,sheet_name="Regions",index=False); weather.to_excel(writer,sheet_name="Weather",index=False)
display(report)
```
