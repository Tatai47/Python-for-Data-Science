## Worked reference solution

The following is one valid approach. It is not the only possible solution. Compare your work with it, and explain any differences in assumptions, handling, or output.

```python
import pandas as pd, matplotlib.pyplot as plt
sales=pd.read_csv(DATA/"sales.csv",parse_dates=["order_date"]); region=sales.groupby("region").revenue.sum().sort_values(ascending=False); daily=sales.groupby("order_date").revenue.sum().sort_index()
fig,axes=plt.subplots(1,3,figsize=(15,4),layout="constrained")
axes[0].bar(region.index,region.values); axes[0].set(title="Revenue by region",xlabel="Region",ylabel="Revenue")
axes[1].plot(daily.index,daily.values); axes[1].set(title="Daily revenue",xlabel="Date",ylabel="Revenue")
axes[2].hist(sales.profit,bins=20); axes[2].set(title="Profit distribution",xlabel="Profit/order",ylabel="Count")
fig.savefig("sales_snapshot.png",dpi=150,bbox_inches="tight"); fig.savefig("sales_snapshot.svg",bbox_inches="tight"); plt.show()
```
