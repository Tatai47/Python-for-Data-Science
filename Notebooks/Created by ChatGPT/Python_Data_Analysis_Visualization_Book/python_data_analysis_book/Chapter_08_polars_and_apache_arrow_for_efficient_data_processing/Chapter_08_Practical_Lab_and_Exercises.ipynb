## Worked reference solution

The following is one valid approach. It is not the only possible solution. Compare your work with it, and explain any differences in assumptions, handling, or output.

```python
try:
 import polars as pl
 query=(pl.scan_csv(DATA/"sales.csv").with_columns((pl.col("revenue")-pl.col("cost")).alias("profit_check")).filter(pl.col("revenue")>10000).group_by("region").agg(pl.col("profit_check").sum().alias("profit_total"),pl.len().alias("orders")).sort("profit_total",descending=True))
 display(query.collect())
except ImportError: print("Install optional packages: pip install polars pyarrow")
```
