## Worked reference solution

The following is one valid approach. It is not the only possible solution. Compare your work with it, and explain any differences in assumptions, handling, or output.

```python
import pandas as pd
raw=pd.read_csv(DATA/"messy_sales.csv",parse_dates=["order_date"]); clean=raw.copy(); n0=len(clean)
clean["product"]=clean["product"].astype("string").str.strip().str.title()
clean["region"]=clean.region.astype("string").str.strip().str.title().fillna("Unknown")
clean=clean.drop_duplicates(subset="order_id",keep="first"); clean["units"]=clean.units.fillna(clean.units.median())
regional=clean.groupby("region",as_index=False).agg(orders=("order_id","count"),revenue=("revenue","sum"),profit=("profit","sum"))
orders=pd.read_csv(DATA/"orders.csv"); customers=pd.read_csv(DATA/"customers.csv")
joined=orders.merge(customers,on="customer_id",how="left",validate="many_to_one",indicator=True)
print({"input_rows":n0,"clean_rows":len(clean),"unmatched_customers":int((joined._merge!="both").sum())}); display(regional)
```
