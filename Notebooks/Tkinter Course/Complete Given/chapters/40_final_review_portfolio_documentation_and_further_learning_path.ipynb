## 1. The complete end-to-end mental model

A mature desktop analytics application has several coordinated layers: user interaction, application orchestration, business calculations, data-source adapters, and durable storage. The UI captures intent and communicates state; services define analytical semantics; adapters normalize source differences; storage enforces its own constraints; and tests verify that totals and filters remain correct. A chart is a view of a summary, not the source of the truth.

The hardest failures often occur at boundaries: file formats to DataFrames, database types to Python types, raw records to normalized schema, business definition to aggregation, worker result to UI state, and internal errors to user messages. Good engineering makes these boundaries visible and testable. This structure allows a small CSV tool to grow into a workbook/MySQL/MongoDB-enabled workbench without rewriting every screen.

## 2. Portfolio evidence and documentation

A portfolio-ready project should include a concise README, setup instructions, dependency list, architecture diagram, supported source formats, sample data, data dictionary, metric definitions, known limitations, and screenshots. A short demo should show a realistic workflow, not just the welcome screen: import data, inspect quality, filter, compare a metric, drill down, export, and recover from a deliberate error.

Include test results and sample output files generated from synthetic data. Do not include credentials, private database dumps, real student/customer records, or unredacted exception traces.
