## 1. Accessibility is part of correctness

An analytics application can calculate accurate results and still fail if users cannot find the filter, interpret the chart, or tell whether an export succeeded. Labels should be clear and associated spatially with their inputs; units and date ranges should be visible; keyboard focus should be predictable; and tables should show sensible headings. Do not rely on placeholder text alone to identify an input because it disappears as the user types.

Use readable fonts and avoid dense control clusters. Important values should not be truncated without a path to inspect them. Status text should report meaningful states—“Loading 2,400 rows,” “Loaded 2,400 rows,” “No rows match current filters,” or “Export failed”—rather than generic “Done.” Keep color as a secondary signal, not the only indicator.

### Worked example 1: Create a status vocabulary

Use a controlled vocabulary so the same state means the same thing across tabs and workflows. A GUI can map these strings to status labels and button states.

```python
STATUS_MESSAGES = {
    "idle": "Ready to load a dataset",
    "loading": "Loading dataset…",
    "loaded": "Dataset loaded successfully",
    "empty": "No records match the current filters",
    "saving": "Saving export…",
    "saved": "Export saved successfully",
    "error": "The operation failed; review the message and logs",
}
for state, message in STATUS_MESSAGES.items():
    print(f"{state:8} {message}")
```

## 2. Modal dialogs and interruption cost

A modal dialog blocks interaction with the parent window until the user responds. Use it when the choice is consequential, an error needs immediate attention, or confirmation is required before destructive work. Do not use modal dialogs for every successful minor action. An inline status message is often better for an operation that completes normally.

Confirmations should be specific: name the output file, indicate whether an existing file will be replaced, or state how many records will be deleted. For a multi-step process, keep the user’s input when validation fails, explain the next action, and avoid forcing the user to restart the form. Technical detail belongs in logs or an expandable detail view rather than a wall of exception text.

### Worked example 2: A scenario-based usability checklist

Treat this list as the start of a repeatable test protocol. Record outcomes, confusion points, and concrete changes, not just a pass/fail score.

```python
usability_scenarios = [
    "Open a valid CSV and identify the loaded row count.",
    "Cancel the Open dialog and verify that the current data remains loaded.",
    "Apply a filter that returns zero rows and understand why the table is empty.",
    "Trigger a validation error and recover without retyping valid fields.",
    "Export the current filtered result and verify the file contents.",
    "Navigate the main workflow using only the keyboard.",
    "Resize the window to a narrow width and find all important controls.",
]
for number, scenario in enumerate(usability_scenarios, start=1):
    print(f"{number}. {scenario}")
```

## 3. Usability tests as reproducible scenarios

Testing a GUI should include a script of what the user does and what they should observe. For example: launch the app, open a CSV, choose a region, observe the row count, select a record, export filtered data, cancel an overwrite prompt, then exit. The expected result should be measurable—correct status text, a known number of rows, or an output file with the expected columns.

Test keyboard-only behavior, narrow windows, empty datasets, delayed connections, cancelled dialogs, and corrupted imports. A usability review is not just asking whether the developer likes the screen. It is checking whether a new user can complete high-value tasks accurately, recover from errors, and understand the scope of the displayed results.
