## 1. From notebook prototype to package

A notebook is a learning and analysis environment, not a deployment boundary. Move stable functions and classes into `.py` modules, add a small entry point, declare dependencies, and keep sample data separate from user data. Use relative imports in a package and test from outside the project directory so accidental reliance on the current working directory is exposed.

Keep optional database integrations modular. A CSV-only user should not need a configured MySQL server or MongoDB service just to open the application. Validate configuration at startup, display which connectors are available, and provide useful guidance when an optional driver or connection is missing.

### Worked example 1: Resolve bundled resources deliberately

This helper belongs in a Python module where `__file__` is available. Keep mutable output and configuration in a user-writable location, not inside the bundled asset root.

```python
from pathlib import Path
import sys


def resource_path(relative_path: str) -> Path:
    """Return a path to a bundled read-only asset in source or frozen mode."""
    bundle_root = Path(getattr(sys, "_MEIPASS", Path(__file__).resolve().parent))
    return bundle_root / relative_path

# Example: logo_path = resource_path("assets/logo.png")
print("In a notebook, __file__ may not exist; call this helper from a .py module.")
```

## 2. Freezing and resource paths

Tools such as PyInstaller can package a Python application into an executable or bundled folder. Packaging a Tkinter and Matplotlib app requires testing Tcl/Tk resources, Matplotlib data files, image assets, and any optional database driver dependencies. Do not assume an app that runs from a source checkout will behave identically when frozen.

The current working directory can vary when a user launches from a desktop shortcut. Use an intentional resource-path helper for files shipped with the app, and a separate user-data/config directory for mutable files. Never write user reports or logs inside an executable bundle directory that may be read-only or replaced during upgrades.

## 3. Distribution and release checks

Build and test on the target operating system when possible. A binary created on one OS is not a universal binary for every platform. Verify startup, Tk theme behavior, DPI scaling, file dialogs, Excel write support, chart rendering, shutdown, and error logging on a clean test machine or virtual environment.

Database credentials should not be baked into the packaged executable. Provide a setup flow, use environment/configuration outside the bundle, or connect to a trusted service that authenticates users. Document Python/runtime requirements, database server requirements, supported file types, backup/upgrade behavior, and known limitations in a user-facing README.
