## 1. Tkinter in the Python ecosystem

Tkinter is Python’s standard interface to the Tcl/Tk GUI toolkit. It provides a compact way to create native desktop applications: forms, menus, dialogs, tables, charts, and file workflows. It is not a web framework and it does not turn a notebook into a browser application. A Tkinter window lives in a desktop windowing system; hosted notebooks, remote servers, containers, and some IDE kernels may not have a display server available. Keep that distinction visible throughout this course.

For analytics students, Tkinter is the presentation layer around an analysis workflow. pandas and NumPy prepare data; SQL or MongoDB stores and retrieves it; Matplotlib renders plots; Tkinter makes the process accessible to a user who should not have to edit code. A useful separation is **interface → application/service logic → data access → storage**. This prevents widget callbacks from becoming tangled with SQL, cleaning rules, and business calculations.

## 1. A widget tree, not a drawing surface

A Tkinter application is a tree of widgets. The root window is normally the top-level parent; frames group related children; controls such as labels, buttons, and entries sit inside those containers. A widget’s parent determines its ownership and affects geometry management. Organizing the tree around the user’s task—filters, results, status, and actions—is easier to maintain than placing every control directly on the root window.

Creating a widget and displaying it are separate operations. The constructor creates a Tcl/Tk object and registers its Python wrapper. A geometry manager allocates space. Configuration changes properties such as text or padding. The event loop dispatches user actions, redraws windows, and runs scheduled callbacks.

### Worked example 1: Inspect a minimal widget tree

Run it as a single cell. Close the window to return to the notebook. The printed children show that hierarchy is inspectable and can support debugging.

```python
import tkinter as tk
from tkinter import ttk

root = tk.Tk()
root.title("Widget tree")
root.geometry("420x180")

container = ttk.Frame(root, padding=16)
container.pack(fill="both", expand=True)
title = ttk.Label(container, text="Sales overview", font=("TkDefaultFont", 14, "bold"))
title.pack(anchor="w")
status = ttk.Label(container, text="Ready", name="status")
status.pack(anchor="w", pady=(8, 0))

print("Root children:", root.winfo_children())
print("Frame children:", container.winfo_children())
root.mainloop()
```

## 2. The event loop and responsiveness

`mainloop()` enters Tk’s event loop. It processes clicks, key presses, window-system notifications, redraws, and scheduled callbacks. In the usual single-threaded GUI model, a callback must return promptly. A 20-second SQL query, `time.sleep(20)`, or a large pandas transformation in a button callback prevents other events from being processed. The window then looks frozen even if Python is still working.

The design goal is not “make everything asynchronous.” It is to decide which operations are short enough for the UI thread and which need a worker, progress indication, cancellation policy, and safe communication back to the UI thread. Tk widget operations should generally occur on the thread that owns the Tk interpreter.

### Worked example 2: Change a widget option after creation

The button updates application state (`message`) rather than locating a label and editing its text imperatively. That pattern scales when several views need to reflect the same value.

```python
import tkinter as tk
from tkinter import ttk

root = tk.Tk()
message = tk.StringVar(value="Waiting")
label = ttk.Label(root, textvariable=message, padding=12)
button = ttk.Button(root, text="Update status", command=lambda: message.set("Updated"))
label.pack()
button.pack(pady=(0, 12))
root.mainloop()
```

## 3. Widget options, variables, and ownership

Widgets expose options through constructors and `.configure()`. Use `.cget("text")` or `.configure("text")` when you need the current option. Tk variables such as `StringVar`, `IntVar`, `DoubleVar`, and `BooleanVar` synchronize supported controls with Python-managed state. They are not identical to ordinary variables: they are backed by Tcl variables and should be associated with the correct Tk interpreter.

State must have a clear owner. A filter selected in the UI may be represented by a Tk variable; the normalized filter criteria should be plain Python values in the service layer; a database connection belongs to the data-access layer. Avoid having a repository query reach into `entry.get()` or having a button callback directly mutate every widget in the application.

## 1. The application lifecycle

A conventional desktop app creates one `Tk()` root, constructs widgets, lays them out, connects callbacks, performs any lightweight initialization, and calls `mainloop()`. Multiple independent `Tk()` instances in one process are usually a mistake because each owns an interpreter and event-loop resources. Use `Toplevel` for secondary windows such as preferences or record details.

In a notebook, GUI examples need extra discipline: run only the cell that opens the window, then close it before moving on. In a packaged application, a normal `if __name__ == "__main__":` guard prevents the GUI from launching just because another module imports the application class.

### Worked example 1: A clean entry point

This design separates application construction from launching the event loop. It can be imported for testing without opening a window.

```python
import tkinter as tk
from tkinter import ttk


def build_app(root: tk.Tk) -> None:
    root.title("First analytics app")
    root.geometry("520x260")
    root.minsize(420, 220)

    frame = ttk.Frame(root, padding=24)
    frame.pack(fill="both", expand=True)
    ttk.Label(frame, text="Tkinter for data analytics", font=("TkDefaultFont", 16, "bold")).pack(anchor="w")
    ttk.Label(frame, text="Start with a small, responsive window.").pack(anchor="w", pady=(8, 16))
    ttk.Button(frame, text="Close", command=root.destroy).pack(anchor="e")


def main() -> None:
    root = tk.Tk()
    build_app(root)
    root.mainloop()


if __name__ == "__main__":
    main()
```

## 2. Geometry, resizing, and the window manager

`title()`, `geometry()`, `minsize()`, and `resizable()` communicate presentation and resize policy. Geometry strings use a form like `800x500`; they are not a substitute for responsive layout. Prefer a geometry manager plus frames, `fill`, `expand`, and grid weights so the application can adapt to different window sizes and font scaling.

Do not assume a fixed pixel layout will work on every platform or display scale. Use meaningful padding, test a narrow window, test keyboard navigation, and avoid truncating important values in analytics tables. The window manager also sends a close request; applications may bind `WM_DELETE_WINDOW` to a cleanup method when they own files, database connections, or worker threads.

### Worked example 2: Handle the window close protocol

This example demonstrates the protocol hook. In production, keep the close handler brief and put deterministic resource cleanup into a method that is safe to call once.

```python
import tkinter as tk
from tkinter import messagebox, ttk

root = tk.Tk()
root.title("Close protocol")
root.geometry("330x140")


def request_close() -> None:
    if messagebox.askyesno("Exit", "Close this window?"):
        root.destroy()

root.protocol("WM_DELETE_WINDOW", request_close)
ttk.Label(root, text="Try the window close button.", padding=18).pack()
root.mainloop()
```

## 3. Functions and classes as application boundaries

A tiny demo can use module-level code, but a chapter-sized application benefits from a `main()` function. It provides one visible entry point and makes imports safe. A class becomes helpful when widgets share state and methods. Do not use classes merely to wrap a few lines; use them when they represent a window, controller, view, or object with a lifecycle.

Shutdown should be explicit: stop scheduled work, signal workers if applicable, close database resources, and destroy the root window. Avoid forcibly killing worker threads. A cancellation event or task-specific cleanup is safer than assuming that closing the window instantly stops all Python work.

## 1. Choosing the right variable

Use `StringVar` for text selections and entry content, `IntVar` for integer-like options, `DoubleVar` for numeric scale values, and `BooleanVar` for toggles. A variable’s `get()` returns a Python value while `set()` updates the Tcl-backed value. Conversion can fail if a field contains invalid input, so variable type does not replace validation.

A Tk variable can be observed with `trace_add`, but traces should be small and predictable. They are useful for synchronizing dependent controls or showing a live preview; they are poor places to do expensive cleaning or issue a database query after every keystroke. Debounce repeated work and keep derived business logic in a normal Python function.

### Worked example 1: Bind controls through StringVar and BooleanVar

The callback extracts plain values. In a real app, pass those values to an analysis service instead of printing them.

```python
import tkinter as tk
from tkinter import ttk

root = tk.Tk()
root.title("State and controls")
region = tk.StringVar(value="All")
include_cancelled = tk.BooleanVar(value=False)

frame = ttk.Frame(root, padding=16)
frame.pack(fill="both", expand=True)
ttk.Label(frame, text="Region").grid(row=0, column=0, sticky="w", padx=(0, 12), pady=6)
ttk.Combobox(frame, textvariable=region, values=("All", "North", "South", "East", "West"), state="readonly").grid(row=0, column=1, sticky="ew")
ttk.Checkbutton(frame, text="Include cancelled", variable=include_cancelled).grid(row=1, column=0, columnspan=2, sticky="w", pady=6)

def show_state() -> None:
    print({"region": region.get(), "include_cancelled": include_cancelled.get()})

ttk.Button(frame, text="Print state", command=show_state).grid(row=2, column=1, sticky="e", pady=(12, 0))
frame.columnconfigure(1, weight=1)
root.mainloop()
```

## 2. Callbacks and closure behavior

A button’s `command` expects a callable, not the result of calling that callable. `command=save_record` passes the function; `command=save_record()` runs it immediately during widget creation and passes its return value instead. Use `lambda` or `functools.partial` to supply arguments. Remember that callback functions usually run later, after the function that created the widget has returned.

Python closures capture variables, not a fresh value for each loop iteration by default. A common bug is `command=lambda: choose(i)` in a loop; each callback may observe the final value of `i`. Bind the current value with a default parameter (`lambda value=i: choose(value)`) or use `partial`. This matters when creating a dashboard with many category buttons.

### Worked example 2: Avoid callback late binding

The default argument captures the current loop value at callback creation time. `functools.partial(report, choice)` is an equally valid alternative.

```python
import tkinter as tk
from tkinter import ttk

root = tk.Tk()
frame = ttk.Frame(root, padding=12)
frame.pack()


def report(choice: str) -> None:
    print("Selected:", choice)

for choice in ("Revenue", "Profit", "Quantity"):
    ttk.Button(frame, text=choice, command=lambda value=choice: report(value)).pack(fill="x", pady=3)

root.mainloop()
```

## 3. Separate UI state from analytical state

A control can produce a simple filter payload such as `{"region": "North", "minimum_sales": 1000}`. A service function takes that payload and a DataFrame and returns a filtered result. The UI reads controls, calls the service, and renders the result. This separation makes filtering testable with ordinary Python unit tests and reduces the risk of having SQL conditions, `Entry.get()` calls, and Treeview updates in a single callback.

Use explicit state transitions: initial, loading, ready, empty, error. These states can drive button availability and status messages. A disabled Export button when no rows are selected is more understandable than allowing a click that silently produces an empty file.

## 1. Pack for linear composition

`pack()` is convenient for top-to-bottom or left-to-right groups. `side`, `fill`, `expand`, and padding describe allocation behavior. Use `fill="x"` for controls that should stretch horizontally; use `fill="both", expand=True` for content regions such as a table or plot. `expand` allocates extra space to the widget’s parcel, while `fill` controls whether the widget fills that parcel.

Packing many unrelated controls into one parent can make order-sensitive code hard to maintain. Use a frame per logical region—toolbar, content, footer—so each region has a limited, understandable layout policy.

### Worked example 1: Responsive grid form

Resize the window and inspect how the second column behaves. The weight belongs to the frame that manages the column, not the root unless the form itself is expected to resize.

```python
import tkinter as tk
from tkinter import ttk

root = tk.Tk()
root.title("Responsive form")
root.geometry("520x280")
root.columnconfigure(0, weight=1)
root.rowconfigure(0, weight=1)

form = ttk.Frame(root, padding=20)
form.grid(row=0, column=0, sticky="nsew")
form.columnconfigure(1, weight=1)

for row, label_text in enumerate(("Dataset name", "Category", "Owner")):
    ttk.Label(form, text=label_text).grid(row=row, column=0, sticky="w", padx=(0, 14), pady=7)
    ttk.Entry(form).grid(row=row, column=1, sticky="ew", pady=7)

form.rowconfigure(3, weight=1)
ttk.Button(form, text="Save draft").grid(row=4, column=1, sticky="e", pady=(14, 0))
root.mainloop()
```

## 2. Grid for forms and tabular controls

`grid()` lays out widgets by row and column, making it a natural choice for forms, filter panels, and label-value pairs. Use `sticky="w"`, `"e"`, `"nsew"`, and related combinations to control alignment. Configure `columnconfigure(index, weight=1)` and `rowconfigure(index, weight=1)` on the parent that owns the grid if a region should receive extra space as the window grows.

Do not use `pack` and `grid` for children of the same parent. It is safe to use one geometry manager in an inner frame and another in a different frame. A typical dashboard uses `grid` on the root for large regions and `pack` inside each region, or consistently uses `grid` where alignment matters.

### Worked example 2: Nested frame layout

Each parent has one coherent layout. This is more robust than trying to make each child independently position itself on the root window.

```python
import tkinter as tk
from tkinter import ttk

root = tk.Tk()
root.geometry("650x360")
root.rowconfigure(1, weight=1)
root.columnconfigure(0, weight=1)

toolbar = ttk.Frame(root, padding=8)
toolbar.grid(row=0, column=0, sticky="ew")
content = ttk.Frame(root, padding=8)
content.grid(row=1, column=0, sticky="nsew")
footer = ttk.Frame(root, padding=8)
footer.grid(row=2, column=0, sticky="ew")
content.rowconfigure(0, weight=1)
content.columnconfigure(0, weight=1)

for text in ("Open", "Refresh", "Export"):
    ttk.Button(toolbar, text=text).pack(side="left", padx=3)
results = tk.Text(content, wrap="word")
results.grid(row=0, column=0, sticky="nsew")
results.insert("1.0", "This area expands while the toolbar and footer stay compact.")
ttk.Label(footer, text="Ready").pack(anchor="w")
root.mainloop()
```

## 3. Why place is usually the exception

`place()` positions widgets with explicit coordinates or relative positions. It can be useful for overlays, carefully controlled drawing interfaces, or a small decorative badge. It is brittle for general application layouts because font metrics, scaling, translated labels, window resizing, and platform differences invalidate assumptions about fixed coordinates.

Responsive Tkinter layout is not automatic, but it can be predictable. Start with a hierarchy of frames, use grid weights on the correct parent, let text and controls size naturally, and test both a narrow and wide window. A scrollable container may be necessary when a form has more fields than can fit vertically.

## 1. Entry and Text serve different tasks

`ttk.Entry` is suitable for compact single-line values: names, amounts, IDs, paths, and search terms. `tk.Text` handles multiline notes, SQL previews, logs, and report text. The Text widget uses special indices such as `"1.0"` for line 1, character 0 and `"end-1c"` to avoid including the widget’s implicit trailing newline when reading text.

A form should specify which fields are required, expected formats, allowed ranges, and normalization rules. A blank optional text field, an invalid integer, and a valid zero are three different states. Avoid treating a truthiness check as validation: `0` and `""` can have different meanings in analytics.

### Worked example 1: Read and validate a form

The form collects and validates first, then performs an action only on success. Real applications should call a service or repository after validation and handle its errors separately.

```python
import tkinter as tk
from tkinter import ttk, messagebox

root = tk.Tk()
root.title("Validated input")
root.geometry("430x210")
name_var = tk.StringVar()
amount_var = tk.StringVar()

form = ttk.Frame(root, padding=18)
form.pack(fill="both", expand=True)
for row, (label, var) in enumerate((("Product", name_var), ("Unit price", amount_var))):
    ttk.Label(form, text=label).grid(row=row, column=0, sticky="w", padx=(0, 12), pady=6)
    ttk.Entry(form, textvariable=var).grid(row=row, column=1, sticky="ew", pady=6)
form.columnconfigure(1, weight=1)


def submit() -> None:
    name = name_var.get().strip()
    raw_amount = amount_var.get().strip()
    errors = []
    if not name:
        errors.append("Product is required.")
    try:
        amount = float(raw_amount)
        if amount < 0:
            errors.append("Unit price cannot be negative.")
    except ValueError:
        errors.append("Unit price must be a number, such as 12.50.")
    if errors:
        messagebox.showerror("Check the form", "\n".join(errors), parent=root)
        return
    print({"product": name, "unit_price": amount})
    messagebox.showinfo("Validated", "The values are ready to save.", parent=root)

ttk.Button(form, text="Validate", command=submit).grid(row=2, column=1, sticky="e", pady=10)
root.mainloop()
```

## 2. Validate at the appropriate boundary

Keystroke-level validation can prevent a user from entering a forbidden character, but it can also make legitimate edits difficult (for example, temporarily deleting a field while replacing a number). Submission-time validation is better for most multi-field business rules: read all values, normalize them, validate the complete record, then either show actionable errors or call the service layer.

Always validate at more than one boundary. GUI validation improves feedback, but imported CSV/Excel data and direct service calls can bypass the form. The data-cleaning layer should independently verify types, nullability, uniqueness, ranges, and domain constraints. A GUI’s validation is usability support, not a security boundary.

### Worked example 2: Get text and remove the implicit trailing newline

Use Text for multiline content, not a large Entry. The `repr` output makes newlines and trailing characters visible during debugging.

```python
import tkinter as tk
from tkinter import ttk

root = tk.Tk()
text = tk.Text(root, width=48, height=6, wrap="word")
text.pack(padx=12, pady=12)
text.insert("1.0", "A multiline note.\nAdd another line.")


def show_value() -> None:
    value = text.get("1.0", "end-1c")
    print(repr(value))

ttk.Button(root, text="Read note", command=show_value).pack(pady=(0, 12))
root.mainloop()
```

## 3. Error messages that help users recover

A message such as “Invalid value” is not enough. Identify the field, explain the accepted format, preserve the user’s entered value when practical, and move focus to the first invalid field. Avoid showing a modal dialog for every small validation error during typing; an adjacent message or status region is often better. Use dialogs for errors that require attention or may lead to data loss.

Normalize human-entered values at the boundary: trim surrounding whitespace, use a defined decimal separator policy, parse dates explicitly, and keep raw and parsed values separate if auditability matters. Do not silently turn a malformed amount into zero or a failed date into today’s date.

## 1. Choice versus free text

Use `ttk.Combobox` when a user should select one value from a known set; set `state="readonly"` when arbitrary text is not allowed. Use radio buttons for a small mutually exclusive group that benefits from seeing all options. Use checkbuttons for independent Boolean options. A Scale is useful for a numeric range where interactive adjustment matters; a Spinbox is often preferable for discrete integer values.

The widget should embody the meaning of the data, not just look convenient. A region filter may include “All regions,” which is an application sentinel rather than a literal region in the database. A category filter needs a defined meaning for “Uncategorized.” Keep that mapping explicit in the query or service layer.

### Worked example 1: Collect multiple filter options

This pattern creates a plain filter dictionary. The service layer can map “All regions” to no region constraint and use the metric name to select a computation.

```python
import tkinter as tk
from tkinter import ttk

root = tk.Tk()
root.title("Filter controls")
root.geometry("440x250")
region = tk.StringVar(value="All regions")
metric = tk.StringVar(value="Revenue")
show_zero = tk.BooleanVar(value=False)
minimum = tk.IntVar(value=0)

frame = ttk.Frame(root, padding=16)
frame.pack(fill="both", expand=True)
ttk.Label(frame, text="Region").grid(row=0, column=0, sticky="w", pady=5)
ttk.Combobox(frame, textvariable=region, state="readonly", values=("All regions", "North", "South", "East", "West")).grid(row=0, column=1, sticky="ew")
ttk.Label(frame, text="Metric").grid(row=1, column=0, sticky="w", pady=5)
for col, value in enumerate(("Revenue", "Profit", "Units"), start=1):
    ttk.Radiobutton(frame, text=value, value=value, variable=metric).grid(row=1, column=col, sticky="w")
ttk.Checkbutton(frame, text="Include zero-sales rows", variable=show_zero).grid(row=2, column=0, columnspan=3, sticky="w", pady=6)
ttk.Label(frame, text="Minimum revenue").grid(row=3, column=0, sticky="w", pady=5)
ttk.Spinbox(frame, from_=0, to=1000000, increment=500, textvariable=minimum, width=12).grid(row=3, column=1, sticky="w")
frame.columnconfigure(1, weight=1)

def apply() -> None:
    print({"region": region.get(), "metric": metric.get(), "show_zero": show_zero.get(), "minimum": minimum.get()})
ttk.Button(frame, text="Apply", command=apply).grid(row=4, column=1, sticky="e", pady=10)
root.mainloop()
```

## 2. Defaults, empties, and range semantics

A selected default affects results. Starting with an arbitrary region can make a dashboard look like it reports all data when it does not. Prefer a clearly labeled `All` option where appropriate and display the effective filter state in the interface. For dates and numeric ranges, define whether endpoints are inclusive and whether an empty value means “no constraint” or “invalid input.”

Selection controls should be revalidated in the service layer because values can also arrive from saved configuration or tests. If a Combobox value is used in a SQL query, pass it as a parameter; do not concatenate user-selected text into SQL. UI options are not a reason to skip parameterized queries.

### Worked example 2: Schedule a debounced search

The 300 ms delay is a debounce, not a database worker. Replace `perform_search` with a fast local operation or dispatch slow work to a worker thread.

```python
import tkinter as tk
from tkinter import ttk

root = tk.Tk()
query = tk.StringVar()
pending = None
result = tk.StringVar(value="Type to search after a short pause")


def schedule_search(*_args) -> None:
    global pending
    if pending is not None:
        root.after_cancel(pending)
    pending = root.after(300, perform_search)


def perform_search() -> None:
    global pending
    pending = None
    result.set(f"Ready to search for: {query.get().strip()!r}")

query.trace_add("write", schedule_search)
ttk.Entry(root, textvariable=query, width=40).pack(padx=12, pady=12)
ttk.Label(root, textvariable=result, padding=12).pack()
root.mainloop()
```

## 3. Dependent controls and update policy

Some options depend on one another: selecting a country may narrow the city list; selecting a dataset may populate columns; choosing a chart type may change available encodings. Update dependent controls deliberately and keep the selected value valid after the option list changes. Do not perform a full database request on every tiny interaction if a clear Apply button is more appropriate.

For live search, consider a short debounce: schedule the search with `after`, cancel the previously scheduled callback when the user types again, then query after the input has settled. Debouncing reduces unnecessary work; it does not replace background execution when the operation itself is slow.

## 1. Treeview’s data model

`ttk.Treeview` displays hierarchical items and columns. For a flat data table, the tree column (`#0`) is often hidden and named columns represent the DataFrame’s selected fields. Define `columns`, `show="headings"`, column heading text, width, minimum width, and anchor. Use a vertical scrollbar and, for wide records, a horizontal scrollbar.

Treeview stores display values, not a pandas DataFrame. Keep the canonical DataFrame or record model separate. A reliable table layer maps a displayed item ID to a record identifier or row index, rather than assuming the display order is the source order. Rebuilding the entire table on every keystroke is expensive; debounce search and update only when needed.

### Worked example 1: Display DataFrame rows in Treeview

The example is appropriate for a tiny demonstration. A real view should format dates and currency, handle nulls, store stable row identifiers, and cap how many records are inserted at once.

```python
import tkinter as tk
from tkinter import ttk
import pandas as pd

sample = pd.DataFrame({
    "order_id": [101, 102, 103],
    "region": ["North", "West", "South"],
    "revenue": [1250.50, 980.00, 2030.75],
})

root = tk.Tk()
root.title("Records")
root.geometry("600x260")
columns = list(sample.columns)
table = ttk.Treeview(root, columns=columns, show="headings", height=8)
for column in columns:
    table.heading(column, text=column.replace("_", " ").title())
    table.column(column, width=150, anchor="w")
scrollbar = ttk.Scrollbar(root, orient="vertical", command=table.yview)
table.configure(yscrollcommand=scrollbar.set)
table.pack(side="left", fill="both", expand=True, padx=(8, 0), pady=8)
scrollbar.pack(side="right", fill="y", padx=8, pady=8)
for record in sample.itertuples(index=False, name=None):
    table.insert("", "end", values=record)
root.mainloop()
```

## 2. Sort values, not formatted strings

A simplistic sort can compare the visible text. That fails when numeric values have commas, currency symbols, or different number formats: lexicographic order can place “100” before “20.” Sort using the underlying Python values, with an explicit missing-value policy. For stable sorting, preserve original order among equal keys. For large datasets, sort the DataFrame or database result and redraw a bounded page.

Treeview item IDs are strings; they are not automatically domain primary keys. Store a stable `record_id` mapping or use the source primary key as the item identifier when uniqueness and character rules permit. When a selected record is edited or deleted, resolve it through the model/repository, then refresh the appropriate rows.

### Worked example 2: Sort a numeric DataFrame column correctly

Keep raw numeric types for computation and sorting. Add formatting only at the last presentation boundary.

```python
import pandas as pd

frame = pd.DataFrame({"product": ["A", "B", "C"], "revenue": [100, 20, 3]})
# Wrong conceptual approach: sorting the display strings gives lexical order.
frame["revenue_display"] = frame["revenue"].map(lambda value: f"{value:,}")
print("Underlying numeric sort:")
print(frame.sort_values("revenue", ascending=True)[["product", "revenue"]])
print("Display strings are for presentation only:")
print(frame[["product", "revenue_display"]])
```

## 3. Pagination and the data-to-view boundary

A widget is not a replacement for a database query engine. Showing 100,000 rows as Treeview items is usually a poor interaction pattern even if the machine has enough memory. Use server-side filters and pagination for database-backed data, or a bounded view for an in-memory DataFrame. Report the total number of matches and which page is visible.

Keep three counts distinct: raw records loaded, rows matching current filters, and rows displayed in the current page. This prevents users from assuming the table is the whole dataset. When exporting, document whether the export means “current page,” “all filtered rows,” or “all source rows.” A deliberate label and confirmation prevent accidental partial exports.

## 1. Commands and event bindings

A widget option such as `command=` is suited to a semantic action like activating a button. `.bind()` attaches a handler to an event pattern such as `<Return>`, `<Escape>`, or `<Control-s>`. The callback receives an event object containing information about the event and widget. Use the simplest mechanism that matches the interaction; adding global bindings when a local command is enough complicates debugging.

Bindings can be attached to a widget, a widget class, the toplevel, or the `all` bind tag. A shortcut such as Ctrl+S should usually be managed at the application level, while a Return key within a form may trigger submit only when the focused control and validation state make that action appropriate.

### Worked example 1: Add a keyboard shortcut for saving

The callback accepts an optional event, so it works with both `.bind()` and a button command. The return value is used intentionally to stop key processing in this example.

```python
import tkinter as tk
from tkinter import ttk, messagebox

root = tk.Tk()
root.title("Keyboard bindings")
root.geometry("430x200")
name = tk.StringVar()
ttk.Label(root, text="Report name", padding=(12, 12, 12, 4)).pack(anchor="w")
entry = ttk.Entry(root, textvariable=name, width=36)
entry.pack(padx=12, anchor="w")
status = tk.StringVar(value="Press Ctrl+S to validate and save")
ttk.Label(root, textvariable=status, padding=12).pack(anchor="w")


def save(_event=None):
    value = name.get().strip()
    if not value:
        status.set("Enter a report name first")
        entry.focus_set()
        return "break"
    status.set(f"Ready to save {value!r}")
    return "break"

root.bind("<Control-s>", save)
ttk.Button(root, text="Save", command=save).pack(padx=12, anchor="e")
root.after(100, entry.focus_set)
root.mainloop()
```

## 2. Focus and predictable navigation

Focus is the recipient of keyboard input. Users should be able to reach interactive controls using Tab and Shift+Tab, identify the focus location, activate controls with conventional keys, and escape from a dialog without losing data unexpectedly. Do not steal focus when a user is typing in a different field. After a validation failure, moving focus to the first invalid field can reduce recovery effort.

Keyboard shortcuts should be discoverable through menu labels and tooltips where appropriate. Avoid hijacking common keys that users expect to work inside an Entry or Text widget. For instance, Ctrl+A should select all text in an input when focus is there; it should not unexpectedly select every record in a table unless the focus context and interaction design clearly imply that action.

### Worked example 2: Bind Escape to close a simple dialog

For a modal dialog, Escape should usually cancel the dialog rather than destroy the whole application. Make the action match the context and confirm if unsaved work may be lost.

```python
import tkinter as tk
from tkinter import ttk

root = tk.Tk()
root.title("Escape handling")
root.geometry("320x140")
root.bind("<Escape>", lambda _event: root.destroy())
ttk.Label(root, text="Press Escape or close the window.", padding=20).pack()
root.mainloop()
```

## 3. Event propagation and handler return values

Tk uses bind tags to determine which binding scripts receive an event. A handler can return `"break"` to stop further binding processing in certain Tkinter event patterns; use it sparingly because it can interrupt default widget behavior. Event propagation is not the same as web DOM bubbling, so consult Tkinter’s binding model rather than assuming browser semantics.

Keep event handlers short. They should interpret the interaction, call application logic, and update state or schedule longer work. Avoid embedding a multi-step import pipeline directly inside a keyboard binding. Test mouse, keyboard, focus traversal, and dialog cancellation as separate interaction paths.

### Worked example 1: Create a menu bar

A menu entry’s command should call a shared application method in a real application. The example uses status updates to make menu routing visible.

```python
import tkinter as tk
from tkinter import ttk, messagebox

root = tk.Tk()
root.title("Menu example")
root.geometry("500x260")
status = tk.StringVar(value="Ready")
ttk.Label(root, textvariable=status, padding=20).pack(anchor="w")
menu_bar = tk.Menu(root)
file_menu = tk.Menu(menu_bar, tearoff=False)
file_menu.add_command(label="Open…", accelerator="Ctrl+O", command=lambda: status.set("Open selected"))
file_menu.add_command(label="Export…", accelerator="Ctrl+E", command=lambda: status.set("Export selected"))
file_menu.add_separator()
file_menu.add_command(label="Exit", command=root.destroy)
menu_bar.add_cascade(label="File", menu=file_menu)
help_menu = tk.Menu(menu_bar, tearoff=False)
help_menu.add_command(label="About", command=lambda: messagebox.showinfo("About", "Tkinter course demo", parent=root))
menu_bar.add_cascade(label="Help", menu=help_menu)
root.configure(menu=menu_bar)
root.mainloop()
```

## 2. Dialogs return values and may be cancelled

Avoid stacking modal dialogs or showing technical stack traces to ordinary users. Present a short explanation and provide a log entry for details that help debugging.

### Worked example 2: Open a CSV with cancellation and error handling

`withdraw()` hides the root window but does not remove it. Production code should keep the root and dialog lifecycle inside the application and show friendly messages while recording diagnostic context.

```python
from pathlib import Path
import tkinter as tk
from tkinter import filedialog, messagebox
import pandas as pd

root = tk.Tk()
root.withdraw()  # Hide the root while asking for a file.
selected = filedialog.askopenfilename(
    title="Open a CSV dataset",
    filetypes=[("CSV files", "*.csv"), ("All files", "*.*")],
)
if selected:
    path = Path(selected)
    try:
        frame = pd.read_csv(path)
        print(f"Loaded {len(frame):,} rows from {path.name}")
    except (OSError, UnicodeError, pd.errors.ParserError) as exc:
        messagebox.showerror("Could not open file", str(exc), parent=root)
else:
    print("Open cancelled")
root.destroy()
```

A save workflow should define overwrite behavior. For Excel workflows, an unsuccessful export should not leave the user thinking a previous file was updated.

## 1. Canvas as a coordinate-based surface

`Canvas` provides items such as lines, rectangles, ovals, text, polygons, and image objects. Every created item has an item ID and can receive tags that group behavior. Canvas is useful for custom diagrams, simple process flows, selection overlays, or a sparkline where a full plotting library would be excessive.

Coordinate-based drawing means you must define scaling behavior. Use the actual Canvas width and height, bind to a configure event when dimensions change, and redraw from model values rather than manually dragging every item into place. Avoid using Canvas to recreate complex tables or statistical charts; specialized widgets and Matplotlib provide better semantics for those tasks.

### Worked example 1: Draw a simple bar chart on Canvas

The example redraws from source values on resize. It demonstrates coordinates rather than best practices for statistical graphics; use Matplotlib for richer charts.

```python
import tkinter as tk
from tkinter import ttk

root = tk.Tk()
root.title("Canvas bars")
root.geometry("520x300")
canvas = tk.Canvas(root, background="white", height=220)
canvas.pack(fill="both", expand=True, padx=12, pady=12)
values = [42, 65, 28, 83]
labels = ["A", "B", "C", "D"]


def draw(_event=None):
    canvas.delete("all")
    width = max(canvas.winfo_width(), 200)
    height = max(canvas.winfo_height(), 120)
    margin = 32
    usable_h = height - 2 * margin
    bar_w = (width - 2 * margin) / len(values) * 0.62
    gap = (width - 2 * margin) / len(values)
    maximum = max(values) or 1
    canvas.create_line(margin, height - margin, width - margin, height - margin, fill="#555555")
    for i, value in enumerate(values):
        x0 = margin + i * gap + (gap - bar_w) / 2
        y0 = height - margin - (value / maximum) * usable_h
        canvas.create_rectangle(x0, y0, x0 + bar_w, height - margin, fill="#4C78A8", outline="")
        canvas.create_text(x0 + bar_w / 2, y0 - 10, text=str(value))
        canvas.create_text(x0 + bar_w / 2, height - margin + 12, text=labels[i])

canvas.bind("<Configure>", draw)
root.mainloop()
```

## 2. Image lifetime and file paths

Tk images must retain a Python reference while they are displayed. A common failure is creating `PhotoImage` in a short-lived local variable and watching the image disappear when garbage collection releases it. Store image objects on the application or view instance. Use pathlib to locate packaged assets and document supported image formats; formats vary by Tk build and image handler availability.

When loading user-selected assets, validate that the file exists and can be opened, and handle errors gracefully. Large images should be resized for the target widget before loading when practical. For report imagery, include an alternative text label or nearby caption so meaning is not conveyed only through pixels.

### Worked example 2: Load an image and keep a reference

Do not expect this example to run unless the supplied path exists and the Tk build supports the image format. For JPEG and more formats, Pillow (`PIL.ImageTk`) is commonly used as an optional dependency.

```python
import tkinter as tk
from tkinter import ttk
from pathlib import Path

root = tk.Tk()
root.title("Image reference pattern")
asset_path = Path("assets") / "logo.png"  # Replace with an existing image.
try:
    logo = tk.PhotoImage(file=str(asset_path))
except tk.TclError as exc:
    ttk.Label(root, text=f"Could not load image: {exc}", padding=16).pack()
else:
    root.logo_image = logo  # Keep a long-lived Python reference.
    ttk.Label(root, image=root.logo_image).pack(padx=16, pady=16)
root.mainloop()
```

## 3. Custom drawings versus analytical plots

For data analytics, Canvas is appropriate for lightweight schematic visualizations such as a progress indicator, a small hand-drawn distribution sketch, or an annotated process diagram. For axes, scales, legends, multiple data series, statistical transforms, and publication-quality figures, use Matplotlib or another charting library and embed its rendering in the Tk application.

Keep the data transformation separate from drawing code. Canvas should receive prepared coordinates or summaries, not query a database or clean a DataFrame on every redraw. This makes it easier to resize the view and to test the calculations independently of the GUI.

# Chapter 15 — Application architecture: MVC/MVP, services, repositories, and navigation

**Book role:** Application design  
**Suggested book space:** about 10 pages after adding diagrams, annotated screenshots, worked examples, and review exercises.  
**How to study:** read the explanations, run one code cell at a time, change the inputs, then complete the lab without copying the solution first.

> **Execution note:** Tkinter opens native desktop windows. Run GUI cells in a local Python/Jupyter session with a working display, not a headless notebook server. Close each window before moving to the next GUI example. Do not use **Run All** on the complete book: some cells open windows, read or write files, or demonstrate database operations.

## Learning outcomes

- Separate widgets, business rules, and persistence code.
- Choose practical controller/service/repository responsibilities.
- Pass plain data structures across boundaries.
- Refactor a callback-heavy prototype into testable modules.

## 1. Why one giant callback does not scale

A prototype often starts with a button callback that reads fields, loads a file, cleans columns, executes a query, calculates metrics, updates five widgets, and writes an export. That callback quickly becomes difficult to test and change. The solution is not necessarily a textbook Model–View–Controller framework; it is a clear boundary between responsibilities.

A **view** builds widgets and renders state. A **controller** interprets user actions and chooses the next operation. A **service** owns a use case such as “summarize current sales” or “validate imported transactions.” A **repository/adapter** knows how to read and write CSV, Excel, MySQL, or MongoDB. Business rules should not require a running Tk window, and database queries should not depend on the layout of a form.

### Worked example 1: Use a service function without Tkinter

The service accepts plain inputs and returns data; it never calls `.get()` on a widget. This boundary also makes it straightforward to unit-test.

```python
from dataclasses import dataclass
import pandas as pd

@dataclass(frozen=True)
class SalesCriteria:
    region: str | None = None
    category: str | None = None
    include_cancelled: bool = False


def filter_sales(frame: pd.DataFrame, criteria: SalesCriteria) -> pd.DataFrame:
    result = frame.copy()
    if not criteria.include_cancelled and "status" in result.columns:
        result = result[result["status"].ne("Cancelled")]
    if criteria.region and criteria.region != "All":
        result = result[result["region"].eq(criteria.region)]
    if criteria.category and criteria.category != "All":
        result = result[result["category"].eq(criteria.category)]
    return result

# This service can be tested in a regular Python process.
sample = pd.DataFrame({"region": ["North", "South"], "category": ["Tech", "Food"], "status": ["Paid", "Cancelled"]})
print(filter_sales(sample, SalesCriteria(region="North")))
```

## 2. Define service contracts around plain Python values

A service method might accept a `FilterCriteria` dataclass and return a `SummaryResult` containing row count, sales, profit, and a DataFrame. It should not reach into Tk variables or set label text. This allows the same service to be used by a notebook, a command-line job, a test, and the GUI. Data contracts also make errors easier to interpret: invalid criteria should be different from a database outage or a missing file.

Use exceptions at boundaries deliberately. A repository can raise a specific exception or return a typed result; a controller catches expected failures and maps them to user-facing messages while the logger preserves context. Avoid catching every exception and ignoring it; silent failure makes a dashboard untrustworthy.

### Worked example 2: A thin controller callback

This is a design sketch, not a standalone cell: it belongs inside a controller with `view`, state variables, and a DataFrame. The important point is that the callback coordinates operations rather than implementing every rule itself.

```python
def on_apply_filters(self):
    """Illustrative method inside a controller class."""
    criteria = SalesCriteria(
        region=self.region_var.get(),
        category=self.category_var.get(),
        include_cancelled=self.include_cancelled_var.get(),
    )
    try:
        filtered = filter_sales(self.sales_df, criteria)
    except (KeyError, ValueError) as exc:
        self.show_error(f"Could not apply filters: {exc}")
        return
    self.view.render_records(filtered)
    self.view.set_status(f"Showing {len(filtered):,} matching rows")
```

## 3. Dependency direction and future changes

The view should depend on a service interface, and the service should depend on repository behavior—not the reverse. The UI should not know whether records came from Excel, MySQL, or MongoDB. An adapter can normalize the source into a canonical DataFrame schema, after which analysis logic behaves consistently.

This is especially valuable for data analytics because storage and analysis tools change independently. A student can prototype with CSV, later connect to a database, and preserve the metrics and visualization code if the contract stays stable. Avoid premature abstraction, though: introduce an interface where there are multiple concrete sources, testability needs, or a genuine reason to isolate dependencies.

## Learning outcomes

- Move blocking I/O or expensive calculations out of the UI event loop.

## 1. after is scheduling, not parallelism

`root.after(milliseconds, callback)` schedules a callback in the Tk event loop. It is useful for periodic refresh, blinking indicators, debounced search, incremental rendering, and polling a worker queue. It does not make the callback run concurrently. A callback that takes several seconds still blocks all other UI processing when it executes.

Never call `time.sleep()` on the UI thread to animate or wait for work. Schedule the next step with `after`. Store the returned job identifier if you may need to cancel it, and cancel outstanding scheduled callbacks before destroying the root. A repeating task should schedule its next occurrence only after its current work has completed, otherwise slow work can create overlapping jobs.

### Worked example 1: Incremental UI work with after

Each step is short, so the UI remains responsive. This is a simulated progression; real progress should be tied to completed work.

```python
import tkinter as tk
from tkinter import ttk

root = tk.Tk()
root.title("Scheduled progress")
root.geometry("420x170")
progress = ttk.Progressbar(root, maximum=100, length=340, mode="determinate")
progress.pack(padx=20, pady=(24, 8))
status = tk.StringVar(value="Ready")
ttk.Label(root, textvariable=status).pack()


def step(value=0):
    progress["value"] = value
    status.set(f"Progress: {value}%")
    if value < 100:
        root.after(40, step, value + 5)

root.after(200, step)
root.mainloop()
```

## 2. Threads and the Tk boundary

A worker thread can perform blocking file or database I/O and some pure-Python orchestration without blocking event handling. However, Tkinter is not generally thread-safe: worker threads should not directly update widgets. Instead, send a result or status message through `queue.Queue`, then poll the queue from the UI thread using `after()` and apply updates there.

For CPU-bound Python work, threads may not improve throughput due to the Global Interpreter Lock. Consider chunked processing, vectorized NumPy/pandas operations, a process pool, or a background service when appropriate. Start with measurement; do not add threading to every button because coordination, exception propagation, and cancellation require additional design.

### Worked example 2: Worker thread + queue + UI polling

The worker only puts messages into a queue; the main Tk thread changes widgets. This is a teaching example. Production code should keep a thread reference, use cooperative cancellation, catch and log failures, and shut down deliberately.

```python
import threading
import queue
import time
import tkinter as tk
from tkinter import ttk

root = tk.Tk()
root.title("Background task")
root.geometry("460x190")
messages = queue.Queue()
status = tk.StringVar(value="Ready")
button = ttk.Button(root, text="Start work")
button.pack(pady=16)
ttk.Label(root, textvariable=status).pack()


def worker():
    try:
        for step_number in range(1, 6):
            time.sleep(0.3)  # Simulate blocking work; never sleep in the UI callback.
            messages.put(("progress", step_number * 20))
        messages.put(("done", "Completed"))
    except Exception as exc:
        messages.put(("error", str(exc)))


def poll_queue():
    try:
        while True:
            kind, payload = messages.get_nowait()
            if kind == "progress":
                status.set(f"Progress: {payload}%")
            elif kind == "done":
                status.set(payload)
                button.configure(state="normal")
            elif kind == "error":
                status.set(f"Failed: {payload}")
                button.configure(state="normal")
    except queue.Empty:
        pass
    if root.winfo_exists():
        root.after(80, poll_queue)


def start():
    button.configure(state="disabled")
    status.set("Working…")
    threading.Thread(target=worker, daemon=True).start()

button.configure(command=start)
root.after(80, poll_queue)
root.mainloop()
```

## 3. Cancellation and stale results

Cancellation is a request, not a guarantee that a blocking driver call will stop immediately. A `threading.Event` can tell cooperative code to stop between steps; database drivers may have their own timeout and cancellation capabilities. On window close, signal cancellation, disable actions that must not repeat, and avoid destroying a view while a worker is trying to update it.

When users change filters while a query is running, the old result can arrive after a new one and overwrite it. Assign each request a monotonically increasing job ID; render a result only if it belongs to the current job. For long exports, communicate stage, rows processed, outcome, and errors. Progress bars should reflect measurable work rather than an arbitrary animation that reaches 100% before the operation finishes.

## 1. A DataFrame is a labeled analytical table

A pandas DataFrame is a two-dimensional labeled data structure with row and column axes. Columns can have different dtypes; the index labels rows and need not be a unique integer range. ### Worked example 1: Create measures and grouped summaries

The calculation is expressed at the transaction-line level, then grouped by region. Confirm whether one row represents an order or an order line before choosing the denominator for any order-level metric.

```python
import pandas as pd

sales = pd.DataFrame({
    "order_id": [1, 2, 3, 4],
    "region": ["North", "South", "North", "West"],
    "quantity": [2, 1, 3, 2],
    "unit_price": [100.0, 250.0, 100.0, 120.0],
    "unit_cost": [60.0, 150.0, 60.0, 75.0],
})
sales = sales.assign(
    revenue=sales["quantity"] * sales["unit_price"],
    cost=sales["quantity"] * sales["unit_cost"],
)
sales["profit"] = sales["revenue"] - sales["cost"]
summary = (sales.groupby("region", as_index=False)
           .agg(orders=("order_id", "nunique"), revenue=("revenue", "sum"), profit=("profit", "sum"))
           .sort_values("revenue", ascending=False))
print(summary)
```

## 2. Vectorization, masks, and grouped analysis

Vectorized pandas operations compute on a Series or DataFrame without an explicit Python loop per row. Boolean masks filter records; `.assign()` creates derived columns; `.groupby()` produces grouped summaries; `.agg()` lets you name multiple measures; and `.sort_values()` prepares results for a table or chart. NumPy adds efficient numerical arrays, broadcasting, and mathematical functions for numerical analysis.

A Tkinter filter panel should translate user choices into masks and grouping parameters. For example, a category selection yields a boolean mask, then a service calculates revenue and profit. ### Worked example 2: Build a testable filter and KPI service

The returned object includes a DataFrame for table rendering and plain summary values for KPI labels. The service rejects an incompatible schema instead of failing later inside a widget update.

```python
from dataclasses import dataclass
import pandas as pd

@dataclass(frozen=True)
class Filters:
    region: str = "All"
    minimum_revenue: float = 0.0


def summarize(frame: pd.DataFrame, filters: Filters) -> dict:
    required = {"region", "revenue"}
    missing = required - set(frame.columns)
    if missing:
        raise ValueError(f"Missing required columns: {sorted(missing)}")
    result = frame.copy()
    if filters.region != "All":
        result = result[result["region"].eq(filters.region)]
    result = result[result["revenue"].ge(filters.minimum_revenue)]
    return {
        "rows": int(len(result)),
        "revenue": float(result["revenue"].sum()),
        "average_revenue": float(result["revenue"].mean()) if not result.empty else None,
        "records": result,
    }

print(summarize(sales.assign(revenue=sales["quantity"] * sales["unit_price"]), Filters("North", 100)))
```

## 3. Statistics need a business definition

`sum`, `mean`, `median`, `min`, `max`, `count`, and `nunique` answer different questions. Does average order value mean total sales divided by number of orders, or mean revenue per line? Does a cancelled transaction contribute zero sales or disappear from the denominator? Should profit equal revenue minus cost, and are returns negative quantities? These are domain decisions, not merely syntax choices.

Name aggregation outputs clearly and test them on a tiny dataset whose expected values can be calculated by hand. For dashboards, expose row counts and filter scope. A KPI without the date range, currency, and included statuses can be mathematically correct but misleading.

### Worked example 3: Use NumPy for numerical operations

NumPy makes the missing-value policy explicit. Standardization is not automatically desirable for every analytics problem; preserve the original measure and explain transformed units.

```python
import numpy as np

values = np.array([12.0, 15.0, np.nan, 18.0])
valid = values[~np.isnan(values)]
print("Mean of valid observations:", float(np.mean(valid)))
print("Population standard deviation:", float(np.std(valid)))

# Example: normalize a numeric feature only if the analytical question calls for it.
if valid.size and np.std(valid) > 0:
    standardized = (valid - np.mean(valid)) / np.std(valid)
    print("Standardized values:", standardized)
```

## 1. Profile before modifying

### Worked example 1: Build a compact profile report

## 2. Cleaning decisions should be auditable

Common actions include trimming text, standardizing category labels, parsing dates, coercing numeric values under a controlled policy, removing true duplicates, and deriving measures. ### Worked example 2: Make cleaning steps measurable

The coercion policy converts malformed numeric text to missing values so it can be flagged; it does not silently make those records zero. In production, count failures against the original non-null mask and attach row identifiers and rule names.

```python
from pathlib import Path
from openpyxl import Workbook, load_workbook
from openpyxl.styles import Font, PatternFill, Alignment
from openpyxl.worksheet.table import Table, TableStyleInfo

output = Path("output") / "styled_sales.xlsx"
output.parent.mkdir(exist_ok=True)
wb = Workbook()
ws = wb.active
ws.title = "Sales"
rows = [
    ["Order ID", "Region", "Revenue", "Profit Margin"],
    [1001, "North", 1250.50, 0.24],
    [1002, "West", 980.00, 0.18],
]
for row in rows:
    ws.append(row)
for cell in ws[1]:
    cell.font = Font(bold=True, color="FFFFFF")
    cell.fill = PatternFill("solid", fgColor="1F4E78")
    cell.alignment = Alignment(horizontal="center")
ws.column_dimensions["A"].width = 14
ws.column_dimensions["B"].width = 16
ws.column_dimensions["C"].width = 16
ws.column_dimensions["D"].width = 18
for cell in ws["C"][1:]:
    cell.number_format = '₹#,##0.00'
for cell in ws["D"][1:]:
    cell.number_format = '0.0%'
table = Table(displayName="SalesTable", ref="A1:D3")
table.tableStyleInfo = TableStyleInfo(name="TableStyleMedium2", showRowStripes=True, showColumnStripes=False)
ws.add_table(table)
wb.save(output)
check = load_workbook(output, data_only=False)
print(check.sheetnames, check["Sales"].max_row, list(check["Sales"].tables))
```

```python
from typing import Protocol
import pandas as pd

class SalesRepository(Protocol):
    def fetch_sales(self, start_date: str, end_date: str, region: str | None = None) -> pd.DataFrame:
        """Return normalized line-level sales for a half-open date range."""
        ...

REQUIRED_COLUMNS = {"order_id", "order_date", "region", "quantity", "unit_price"}

def validate_repository_frame(frame: pd.DataFrame) -> pd.DataFrame:
    missing = REQUIRED_COLUMNS - set(frame.columns)
    if missing:
        raise ValueError(f"Repository result missing columns: {sorted(missing)}")
    result = frame.copy()
    result["order_date"] = pd.to_datetime(result["order_date"], errors="raise")
    result["quantity"] = pd.to_numeric(result["quantity"], errors="raise")
    result["unit_price"] = pd.to_numeric(result["unit_price"], errors="raise")
    return result
```

## 3. Threading and resource lifetime

Never run an unpredictable network query directly inside a UI button callback. The connection attempt itself may block before any rows return. ### Worked example 3: Write multiple related rows transactionally

The helper expects a live connection and a schema matching the SQL. Its caller should handle the raised error, log useful context without private values, and surface a user-facing failure. Verify whether the driver manages autocommit as expected.

```python
def insert_order(connection, order_row, item_rows):
    """Insert a header and all lines as one transaction; schema is illustrative."""
    cursor = connection.cursor()
    try:
        cursor.execute(
            "INSERT INTO orders (order_id, order_date, customer_id) VALUES (%s, %s, %s)",
            (order_row["order_id"], order_row["order_date"], order_row["customer_id"]),
        )
        cursor.executemany(
            "INSERT INTO order_items (order_id, product_id, quantity, unit_price) "
            "VALUES (%s, %s, %s, %s)",
            [(order_row["order_id"], item["product_id"], item["quantity"], item["unit_price"]) for item in item_rows],
        )
        connection.commit()
    except Exception:
        connection.rollback()
        raise
    finally:
        cursor.close()
```

## 1. Documents and flexible schema are not schema-free

MongoDB stores BSON documents in collections. Different documents can have different fields, which supports evolving models, but applications still need a contract for required fields, allowed types, and naming. A flexible schema does not remove the need for data validation; it can simply move more of that responsibility into application code or collection validators.

Design around query patterns. ### Worked example 1: Create a simple aggregation pipeline

This pipeline is illustrative: real BSON dates should normally be stored as dates rather than ISO strings, and the `$addToSet` array can consume memory. A simpler `$sum: 1` works if documents represent one order each; choose based on the collection grain.

```python
pipeline = [
    {"$match": {"status": "Paid", "order_date": {"$gte": "2026-01-01", "$lt": "2027-01-01"}}},
    {"$group": {
        "_id": "$region",
        "order_count": {"$addToSet": "$order_id"},
        "revenue": {"$sum": {"$multiply": ["$quantity", "$unit_price"]}},
    }},
    {"$project": {
        "_id": 0,
        "region": "$_id",
        "order_count": {"$size": "$order_count"},
        "revenue": 1,
    }},
    {"$sort": {"revenue": -1}},
]
print(pipeline)
# Execute with collection.aggregate(pipeline) after aligning field names and date types with your schema.
```

## 2. Filters, projections, and indexes

A `find()` filter selects matching documents; a projection controls returned fields; sort and limit shape results. Query documents should come from validated UI criteria, not be built from arbitrary raw strings. MongoDB operators such as `$gte`, `$in`, `$and`, and `$or` express filter semantics. Watch for the difference between a missing field and a field explicitly set to `null` when data quality matters.

Indexes can make common queries faster but increase storage and write overhead. Index fields used frequently in filters and sorts, then inspect query plans and production-like data. Avoid creating an index for every possible GUI filter combination. A filterable dashboard should start with a small number of known access paths and be measured before adding indexes.

### Worked example 2: Map UI criteria to a safe Mongo filter

Use BSON datetime objects when the collection stores BSON dates. The GUI should parse and validate user dates first. For complex criteria, construct a small known set of operators instead of accepting an arbitrary query document from user text.

```python
def build_mongo_filter(region: str, start_date, end_date, include_cancelled: bool) -> dict:
    criteria = {"order_date": {"$gte": start_date, "$lt": end_date}}
    if region != "All":
        criteria["region"] = region
    if not include_cancelled:
        criteria["status"] = {"$ne": "Cancelled"}
    return criteria

print(build_mongo_filter("North", "2026-01-01", "2027-01-01", False))
```

## 3. Aggregation pipelines and the DataFrame boundary

An aggregation pipeline applies stages such as `$match`, `$project`, `$unwind`, `$group`, `$sort`, and `$limit`. Put restrictive `$match` stages early when semantics allow, and use `$project` to omit fields that the GUI does not need. `$unwind` turns array elements into separate pipeline documents; it changes row grain and can multiply rows just as a relational join can.

The result may contain nested documents, arrays, ObjectIds, dates, and Decimal128 values. Normalize these into stable columns before constructing a pandas DataFrame. Preserve an identifier as a string for display or as an ObjectId in the repository layer; be explicit about decimal conversion to avoid silently losing financial precision.

### Worked example 2: Read with projection, sort, and limit

This function deliberately limits preview size and projects columns. It assumes `collection` is already a configured PyMongo collection and dates are stored in a sortable date type.

```python
def fetch_order_preview(collection, region: str, limit: int = 200) -> list[dict]:
    if not 1 <= limit <= 2000:
        raise ValueError("limit must be between 1 and 2000")
    criteria = {} if region == "All" else {"region": region}
    projection = {"_id": 1, "order_id": 1, "order_date": 1, "region": 1, "status": 1}
    cursor = collection.find(criteria, projection).sort("order_date", -1).limit(limit)
    try:
        return list(cursor)
    finally:
        cursor.close()
```

## 3. Projection, limiting, and index strategy

Use a projection to return the fields required for a table, not entire documents with nested private details. Sort by a deterministic field and use a limit for preview screens. For large results, paginate or iterate a cursor in bounded batches rather than eagerly loading every document into memory. Close cursors when abandoning a large query early, and normalize the output in a data adapter.

Create indexes based on real filter/sort usage; examine execution plans and data size. Compound index field order matters. Adding an index has a cost for every write and consumes storage, so indexes should be evaluated, not collected without a workload model. If an app supports region, status, and date filters, select indexes that match the most common query patterns and validate them with `explain`.

### Worked example 3: Update a specific record and verify the outcome

In real data, ensure `order_id` is uniquely indexed or filter by `_id`. A modified count of zero can mean the value was already the requested value, not necessarily that the operation failed.

```python
def set_order_status(collection, order_id: str, new_status: str) -> int:
    allowed_statuses = {"Pending", "Paid", "Cancelled", "Returned"}
    if new_status not in allowed_statuses:
        raise ValueError("Unsupported order status")
    result = collection.update_one(
        {"order_id": order_id},
        {"$set": {"status": new_status}},
    )
    if result.matched_count == 0:
        return 0
    return int(result.modified_count)
```

```python
import pandas as pd

def sales_kpis(frame: pd.DataFrame) -> dict:
    required = {"order_id", "revenue", "profit"}
    if not required.issubset(frame.columns):
        raise ValueError(f"Required columns: {sorted(required)}")
    row_count = len(frame)
    revenue = float(frame["revenue"].sum())
    profit = float(frame["profit"].sum())
    return {
        "line_count": row_count,
        "unique_orders": int(frame["order_id"].nunique()),
        "revenue": revenue,
        "profit": profit,
        "profit_margin": profit / revenue if revenue else None,
    }

sample = pd.DataFrame({"order_id": [1, 1, 2], "revenue": [100, 50, 80], "profit": [20, 10, 8]})
print(sales_kpis(sample))
```

### Worked example 2: Keep drill-down state explicit

The state object makes drill-down and reset logic testable. A GUI controller can update controls to reflect this state and re-run the data service.

### Worked example 2: A bounded HTTP request pattern

A timeout tuple separates connection establishment and reading. The function is synchronous and belongs in a worker when used by a GUI.

```python
import requests


def get_json(url: str, params: dict | None = None) -> dict:
    response = requests.get(url, params=params, timeout=(3.05, 15))
    response.raise_for_status()
    content_type = response.headers.get("Content-Type", "")
    if "json" not in content_type.lower():
        raise ValueError("Expected a JSON response")
    payload = response.json()
    if not isinstance(payload, dict):
        raise ValueError("Expected a JSON object at the response root")
    return payload

# Example call intentionally omitted: use a documented API and comply with its authentication/rate-limit rules.
```

### Worked example 1: Redact sensitive settings before diagnostic display

Use a tested allow-list when possible; secret-key substring matching is a useful illustration but can miss unusual configuration names. Never print the raw settings dictionary to debug a connection problem.

```python
def safe_config_summary(config: dict) -> dict:
    secret_keys = {"password", "token", "secret", "api_key", "uri"}
    summary = {}
    for key, value in config.items():
        if any(secret_part in key.lower() for secret_part in secret_keys):
            summary[key] = "<redacted>" if value else "<not configured>"
        else:
            summary[key] = value
    return summary

example = {"host": "db.example.internal", "password": "not-for-printing", "database": "analytics", "mongo_uri": "mongodb://user:password@host/db"}
print(safe_config_summary(example))
```

### Worked example 2: Neutralize untrusted spreadsheet text for CSV export

Spreadsheet applications vary in how they interpret CSV fields. Treat this as a starting point to test against the target software and threat model, not as a universal substitute for security review.

```python
import pandas as pd

FORMULA_PREFIXES = ("=", "+", "-", "@", "\t", "\r")

def neutralize_spreadsheet_text(value):
    if isinstance(value, str) and value.startswith(FORMULA_PREFIXES):
        return "'" + value
    return value

unsafe = pd.DataFrame({"customer_note": ['=HYPERLINK("https://example.invalid","open")', "ordinary note"], "amount": [12.5, -3.0]})
safe = unsafe.copy()
safe["customer_note"] = safe["customer_note"].map(neutralize_spreadsheet_text)
print(safe)
# This policy applies to untrusted text columns, not blindly to every numeric measure.
```

### Worked example 2: Slice a page from a filtered DataFrame

The first output shows a full page and the second a partial final page. For database-backed paging, add deterministic ordering and push the limit/offset or keyset condition into the query.

```python
import pandas as pd

def dataframe_page(frame: pd.DataFrame, page: int, page_size: int = 200) -> pd.DataFrame:
    if page < 0:
        raise ValueError("page must be zero-based and non-negative")
    if not 1 <= page_size <= 2000:
        raise ValueError("page_size must be between 1 and 2000")
    start = page * page_size
    return frame.iloc[start:start + page_size].copy()

sample = pd.DataFrame({"value": range(1050)})
print(len(dataframe_page(sample, page=4, page_size=200)))
print(len(dataframe_page(sample, page=5, page_size=200)))
```

## 1. Start with the data service

Keep the analytics pipeline independent of the Tkinter application. ### Worked example 1: Create reusable sales transformations

This is a baseline, not a complete policy for all edge cases. Add row-level quality issues for coercion failures, status rules, unknown categories, and profit consistency. A production import should preserve the original raw data when auditing is required.

```python
import pandas as pd

REQUIRED = {"order_id", "order_date", "customer_id", "region", "product", "category", "quantity", "unit_price", "unit_cost", "status"}

def normalize_sales(raw: pd.DataFrame) -> pd.DataFrame:
    frame = raw.copy()
    frame.columns = [str(column).strip().lower().replace(" ", "_") for column in frame.columns]
    missing = REQUIRED - set(frame.columns)
    if missing:
        raise ValueError(f"Missing required columns: {sorted(missing)}")
    frame["order_date"] = pd.to_datetime(frame["order_date"], errors="coerce")
    for column in ("quantity", "unit_price", "unit_cost"):
        frame[column] = pd.to_numeric(frame[column], errors="coerce")
    frame["revenue"] = frame["quantity"] * frame["unit_price"]
    frame["cost"] = frame["quantity"] * frame["unit_cost"]
    frame["profit"] = frame["revenue"] - frame["cost"]
    frame["profit_margin"] = frame["profit"] / frame["revenue"].where(frame["revenue"].ne(0))
    return frame

# Example with the course pack: raw = pd.read_csv("data/sample_sales.csv"); clean = normalize_sales(raw)
```

### Worked example 2: Build filtered data and a summary in one service

This makes the filter and KPI semantics testable. For date filters, make the UI show that end date is exclusive internally or convert the visible inclusive date to the next day deliberately.

```python
from dataclasses import dataclass
import pandas as pd

@dataclass(frozen=True)
class SalesFilters:
    start_date: pd.Timestamp | None = None
    end_date: pd.Timestamp | None = None
    region: str = "All"
    category: str = "All"
    status: str = "Paid"


def query_sales(frame: pd.DataFrame, filters: SalesFilters) -> tuple[pd.DataFrame, dict]:
    result = frame.copy()
    if filters.start_date is not None:
        result = result[result["order_date"] >= filters.start_date]
    if filters.end_date is not None:
        result = result[result["order_date"] < filters.end_date]
    if filters.region != "All":
        result = result[result["region"].eq(filters.region)]
    if filters.category != "All":
        result = result[result["category"].eq(filters.category)]
    if filters.status != "All statuses":
        result = result[result["status"].eq(filters.status)]
    revenue = float(result["revenue"].sum())
    profit = float(result["profit"].sum())
    summary = {"line_count": len(result), "orders": int(result["order_id"].nunique()),
               "customers": int(result["customer_id"].nunique()), "revenue": revenue,
               "profit": profit, "margin": profit / revenue if revenue else None}
    return result, summary
```

### Worked example 1: Create a generic data-quality profile

```python
import pandas as pd

def profile_dataframe(frame: pd.DataFrame) -> pd.DataFrame:
    total = max(len(frame), 1)
    return pd.DataFrame({
        "column": frame.columns,
        "dtype": [str(frame[name].dtype) for name in frame.columns],
        "missing_count": [int(frame[name].isna().sum()) for name in frame.columns],
        "missing_pct": [round(float(frame[name].isna().mean() * 100), 2) for name in frame.columns],
        "distinct_count": [int(frame[name].nunique(dropna=True)) for name in frame.columns],
    })

example = pd.DataFrame({"student_id": ["S1", "S2", "S3"], "score": [85, None, 91], "subject": ["Math", "Math", "Science"]})
print(profile_dataframe(example))
```

### Worked example 2: Define a normalized assessment score

This example normalizes scores but flags values outside the maximum or with invalid denominators. Weighting assessment percentages into a final grade is a separate policy decision that should be written down.

```python
import pandas as pd

def add_score_percentage(frame: pd.DataFrame) -> pd.DataFrame:
    required = {"score", "max_score"}
    if not required.issubset(frame.columns):
        raise ValueError(f"Required columns: {sorted(required)}")
    result = frame.copy()
    result["score"] = pd.to_numeric(result["score"], errors="coerce")
    result["max_score"] = pd.to_numeric(result["max_score"], errors="coerce")
    invalid = result["max_score"].isna() | result["max_score"].le(0) | result["score"].lt(0) | result["score"].gt(result["max_score"])
    result["quality_issue"] = invalid
    result["score_pct"] = (result["score"] / result["max_score"] * 100).where(~invalid)
    return result

scores = pd.DataFrame({"score": [18, 40, 105, None], "max_score": [20, 50, 100, 100]})
print(add_score_percentage(scores))
```

## 1. Tkinter-specific failure modes

A blank window can mean the event loop never started, widgets were never managed by pack/grid/place, or an exception interrupted construction. ### Worked example 1: Inspect widget and geometry state

The function assumes a real widget that belongs to a running Tk interpreter. Window dimensions can still be provisional before the event loop has processed geometry requests.

```python
def debug_widget(widget):
    """Call this from a running Tk application while debugging."""
    print("widget:", widget)
    print("class:", widget.winfo_class())
    print("mapped:", widget.winfo_ismapped())
    print("size:", widget.winfo_width(), widget.winfo_height())
    print("parent:", widget.master)
    print("manager:", widget.winfo_manager())

# Example use after constructing a widget: debug_widget(my_widget)
```

## 2. Data and database failure modes

A result that is consistently too large may be caused by a join that multiplies records or confusing order lines with orders.
