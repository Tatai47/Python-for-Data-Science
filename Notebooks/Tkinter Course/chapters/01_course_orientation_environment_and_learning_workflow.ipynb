## 1. Tkinter in the Python ecosystem

Tkinter is Python’s standard interface to the Tcl/Tk GUI toolkit. It provides a compact way to create native desktop applications: forms, menus, dialogs, tables, charts, and file workflows. It is not a web framework and it does not turn a notebook into a browser application. A Tkinter window lives in a desktop windowing system; hosted notebooks, remote servers, containers, and some IDE kernels may not have a display server available. Keep that distinction visible throughout this course.

For analytics students, Tkinter is the presentation layer around an analysis workflow. pandas and NumPy prepare data; SQL or MongoDB stores and retrieves it; Matplotlib renders plots; Tkinter makes the process accessible to a user who should not have to edit code. A useful separation is **interface → application/service logic → data access → storage**. This prevents widget callbacks from becoming tangled with SQL, cleaning rules, and business calculations.

### Worked example 1: Check the interpreter and Tk installation

The printed executable matters: it tells you which Python kernel is running. A terminal and a notebook can silently use different environments. The final command is documented as the quick visual check; keep it commented inside a notebook until you intend to test GUI support.

```python
import sys
import platform
import tkinter as tk

print("Python:", sys.version)
print("Executable:", sys.executable)
print("Platform:", platform.platform())
print("Tk version:", tk.Tcl().eval("info patchlevel"))

# Optional: run in a desktop session to open the Tcl/Tk test window.
# Run from a terminal with: python -m tkinter
```

## 2. Environment checks and dependency boundaries

Use one Python environment per course or project so package versions are explicit and reproducible. First verify the interpreter that launches the notebook; then verify Tkinter independently. The command `python -m tkinter` should open a small demonstration window in a working desktop environment. If it fails with an import error, install or repair the matching Tk support for that Python distribution rather than assuming that a pip package named `tkinter` is the solution.

Do not install database servers merely to read this book. Most chapters use local synthetic CSV data and optional dependencies. MySQL examples require a configured MySQL server and a database account; MongoDB examples require a reachable MongoDB deployment. Connection strings, passwords, and real customer records should never be placed in notebook output or committed to source control.

### Worked example 2: Check optional packages without breaking the base course

Optional packages are capability checks, not prerequisites for every chapter. Keep the lightweight Tkinter chapters usable even when a database driver or charting library is absent.

```python
from importlib.util import find_spec

optional_packages = [
    "pandas", "numpy", "openpyxl", "matplotlib", "seaborn",
    "sqlalchemy", "mysql.connector", "pymongo", "pytest",
]
for package in optional_packages:
    try:
        available = find_spec(package) is not None
    except ModuleNotFoundError:
        available = False
    print(f"{package:18} {'available' if available else 'not installed'}")
```
