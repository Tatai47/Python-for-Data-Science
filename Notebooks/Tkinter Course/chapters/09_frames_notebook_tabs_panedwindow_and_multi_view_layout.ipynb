## 1. Frames as composition boundaries

A frame is more than a box. It provides a local geometry-management context, an ownership boundary, and a unit that can be shown, hidden, or reused. Build frames around purposeful sections such as filters, KPI cards, detail forms, or an error panel. Name frames by their responsibility and keep each frame’s constructor focused on building that part of the UI.

A common architecture is a root window containing a toolbar, a main view, and a status bar. The main view may contain a tabbed notebook with Overview, Data, and Quality tabs. State such as the active DataFrame and current filter selection should be shared through a controller or model rather than duplicated in each tab.

### Worked example 1: Tabbed analytics shell

Treat each frame as a separately composed view. In a larger project, move the construction into `OverviewFrame`, `DataFrameView`, and `QualityFrame` classes.

```python
import tkinter as tk
from tkinter import ttk

root = tk.Tk()
root.title("Analytics workspace")
root.geometry("620x400")
notebook = ttk.Notebook(root)
notebook.pack(fill="both", expand=True, padx=10, pady=10)

for title, description in (("Overview", "Summary cards and charts"), ("Data", "Filtered record table"), ("Quality", "Missing values and validation issues")):
    page = ttk.Frame(notebook, padding=18)
    notebook.add(page, text=title)
    ttk.Label(page, text=title, font=("TkDefaultFont", 14, "bold")).pack(anchor="w")
    ttk.Label(page, text=description).pack(anchor="w", pady=8)
root.mainloop()
```

## 2. Tabs and panes

`ttk.Notebook` groups views whose content is conceptually parallel. It is useful for an analytics tool where users switch between a table, a chart, and data-quality details. Too many tabs can hide important navigation and complicate state management, so group related pages and keep the most common action visible.

`PanedWindow` lets users resize adjacent regions, such as a sidebar and a table. Give each pane a sensible minimum size where supported and decide which region should receive space at startup. Remember that user resizing is a preference for the current session; do not persist window/pane geometry without validating saved values when the display changes.

### Worked example 2: Resizable sidebar and content pane

The pane divider allows the user to customize the available space. Test how both regions behave at a narrow window width.

```python
import tkinter as tk
from tkinter import ttk

root = tk.Tk()
root.geometry("720x400")
panes = ttk.Panedwindow(root, orient="horizontal")
panes.pack(fill="both", expand=True, padx=8, pady=8)
sidebar = ttk.Frame(panes, padding=12, width=190)
content = ttk.Frame(panes, padding=12)
panes.add(sidebar, weight=1)
panes.add(content, weight=4)
ttk.Label(sidebar, text="Filters", font=("TkDefaultFont", 12, "bold")).pack(anchor="w")
ttk.Label(content, text="Results and charts", font=("TkDefaultFont", 12, "bold")).pack(anchor="w")
root.mainloop()
```

## 3. Navigation and data consistency

Switching a view should not unexpectedly discard filters or reload expensive data without telling the user. Define whether views share the same query result or compute their own projections. It is often effective to retain a canonical filtered DataFrame in a controller/service layer and derive a table view, chart aggregation, and quality summary from that data.

Do not use tabs as a replacement for task flow. A multi-step import wizard can be better represented by Next/Back controls with explicit validation, while a notebook-style widget is best for freely switching between parallel analytical views.
