## 1. Why ttk is the default for many controls

Themed Tk widgets (`ttk`) use a platform-aware theming system and generally provide a more native appearance than many classic widgets. Their options are not identical to classic Tk options: for example, `ttk.Button` styling is commonly configured through `ttk.Style`, not by setting arbitrary colors on each button. Use classic widgets when their specialized behavior is valuable, such as `Canvas`, `Text`, or certain advanced option combinations.

A style is a named design rule. Applying a style such as `Heading.TLabel` or `Primary.TButton` makes intent explicit and reduces repeated option dictionaries. A consistent design system should define typography, spacing, emphasis, validation feedback, disabled state, and table row density.

### Worked example 1: Define reusable ttk styles

Style names encode a role rather than a single location. For example, the same title style can be used on multiple pages and themed consistently later.

```python
import tkinter as tk
from tkinter import ttk

root = tk.Tk()
root.title("Style system")
root.geometry("420x230")
style = ttk.Style(root)
style.configure("PageTitle.TLabel", font=("TkDefaultFont", 16, "bold"), padding=(0, 0, 0, 10))
style.configure("Metric.TLabel", font=("TkDefaultFont", 12, "bold"), padding=8)
style.configure("Primary.TButton", padding=(12, 7))

page = ttk.Frame(root, padding=20)
page.pack(fill="both", expand=True)
ttk.Label(page, text="Revenue dashboard", style="PageTitle.TLabel").pack(anchor="w")
ttk.Label(page, text="Net sales: ₹ 248,500", style="Metric.TLabel", relief="groove").pack(fill="x", pady=6)
ttk.Button(page, text="Refresh data", style="Primary.TButton").pack(anchor="e")
root.mainloop()
```

## 2. Theme selection and platform verification

`ttk.Style().theme_names()` reveals the themes supported by the current Tcl/Tk installation. `theme_use()` switches the active theme. A theme available on one platform may not exist elsewhere, so inspect available themes and choose a fallback rather than assuming a theme is universal. Always test the result on a second platform or display scaling setting if the application will be distributed.

Do not rely on color alone to communicate status. Pair a warning color with a label or icon and explicit wording. Ensure foreground/background contrast, visible focus, a predictable tab sequence, and adequate control size. Native-looking widgets still need deliberate accessibility and usability review.

### Worked example 2: Inspect themes and choose a fallback

This demonstrates capability detection. Do not force a theme if it makes controls less native or less accessible on the target platform.

```python
import tkinter as tk
from tkinter import ttk

root = tk.Tk()
style = ttk.Style(root)
available = style.theme_names()
print("Available themes:", available)
preferred = "clam"
if preferred in available:
    style.theme_use(preferred)
else:
    print(f"Theme {preferred!r} unavailable; keeping {style.theme_use()!r}")
ttk.Label(root, text=f"Active theme: {style.theme_use()}", padding=20).pack()
root.mainloop()
```

## 3. Visual hierarchy for analytics interfaces

Analytics screens often have many numbers competing for attention. Use a clear title, restrained section headings, primary/secondary actions, consistent alignment, and a small set of spacing values. Reserve strong emphasis for actions such as Apply Filters and Save Report; destructive actions should be visibly distinct and confirm intent.

A result count, refresh timestamp, loading status, and validation message should have stable positions. Avoid changing control geometry on every update. Table and chart titles should state the metric and time window; a chart without units or a clear title forces readers to infer meaning.
